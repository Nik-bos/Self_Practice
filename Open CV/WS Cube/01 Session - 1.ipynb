# Show multiple images and slide show.

Showing multiple images means showing a single image multiple times in one single window.

To do this, we will need both, cv2 and numpy module.

```python
import numpy as np
```

```python
# How hstack and vstack works ?

array1 = np.array([1, 2, 3, 4])
array2 = np.array([5, 6, 7, 8])

h_stack = np.hstack((array1, array2))
v_stack = np.vstack((array1, array2))

print(f'Horizontal Stack: {h_stack}')
print()
print('Vertical Stack:', '\n', v_stack)
```

```python
img = cv2.imread(r'D:\Nikss\Sadhana photos\_20180110_122158.JPG')

# We can see in our previous codes that our image is too big to fit into our screen. So we need to resize it first

re_img = cv2.resize(img, (400, 400))    # (500,500) stands for width and height respectively.
# We can store it into same variable also. Here we used 're_img' as variable name. 

# Concenate the image using hstack() and vstack()
hstack = np.hstack((re_img, re_img))

cv2.imshow('Sadhana', hstack)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
vstack = np.vstack((re_img, re_img))

cv2.imshow('Sadhana', vstack)
cv2.waitKey(0)
cv2.destroyWindow('Sadhana')

# Figure out why error is given in cv2.destroyWindow()
```

```python
# 2 * 2 image. We have already stacked the image horizontally and stored it into variale 'hstack'.
# Now if we want to have 2*2 image, just vertical stack this 'hstack' variable.

img_2by2 = np.vstack((hstack, hstack))

img_2by2 = cv2.resize(img_2by2, (650, 600))   # We used the same variable name to store the resized image.
cv2.imshow("Sadhana", img_2by2)
cv2.waitKey(0)
cv2.destroyWindow("Sadhana")
```

# Apply slide show feature.

To apply a slide show feature in our image, we need to import 'os' module. Bcz for applying slideshow, we need to get the names of our files which is stored in some folder at some location, in our computer. This is only possible through 'os module.

```python
import os
```

```python
names = os.listdir(r'D:\Nikss\Sadhana photos')
names
```

```python
for i in names:
    path = 'D:\\Nikss\\Sadhana photos' + '\\' + i
    print(path)
    
    img = cv2.imread(path)
    re_img = cv2.resize(img, (600, 500))
    cv2.imshow('Sadhana', re_img)
    cv2.waitKey(0)
cv2.destroyAllWindows()
    
```
