# Read and show images

imread() : Reads the image

imshow() : Shows the read image.

```python
import cv2
```

```python
img = cv2.imread(r'D:\Nikss\Sadhana photos\_20180110_122158.JPG')
print(img)
print()
print()

print(img.ndim)
```

```python
cv2.imshow('Sadhana', img)    

# This will create a window and it opens only for 1 millisecond. (1000 millisecond = 1 second)
# So in order to keep it for some time, we need to use waitKey().

cv2.waitKey(0)    # 0 means window will be open untill we press any key from the keyword.
cv2.destroyAllWindows()    # destroyAllWindows(), destroyes all the open windows.

# cv2.destroyAllWindow('window_name') : destroyes only one window whose name is mentiones as an argument. 
```

```python
 img.shape
```

```python
# Create more than one windows.

img2 = cv2.imread(r'D:\Nikss\Sadhana photos\_20200107_075017.JPG')
cv2.imshow('Sadhana_2', img2)
cv2.imshow('Sadhana', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
