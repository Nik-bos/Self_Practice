# Text on an image

cv2.putText function is used. Its parameters are as follows,
1) **img**: Takes image
2) **text(str)**: Takes the text we want to insert on an image.
3) **org(Tuple or list of 2 integers)**: Means 'origin'. It takes x_axis and y_axis from bottom left corner. for e.g (5,6) means 5 pixels left and 6 pixels down.
4) **fontFace(int)**: Type of font. Allows argument between 0 to 7.
5) **fontScale(float)**: Size of fonts.
6) **color(tuple or list of three integers)**: Specify the color of the text. It is stored in BGR format. So we can give the combination of these colors. For e.g (25,50, 90).
7) **thickness(int)**: Thickness of the border line of the text.
8) **lineType(int)**: Type of line. Valid arguments are -1, 4, 8 and 16.
9) **bottomLeftOrigin(True/False)**: If 'False', it gives real text and if 'True', it gives mirror image. It is 'False' bydefault.


```python
img = cv2.imread(r"D:\Nikss\Self Pracitce\Open CV\Sadhana photos\_20200107_075017.JPG")
img = cv2.resize(img, (500,500))

text = cv2.putText(img = img, text = "Sadhana", org = [50, 50], fontFace = 6, fontScale = 1.5, color = (0, 255, 255), thickness = 2, lineType = -1, 
                   bottomLeftOrigin = False)

cv2.imshow("Sadhana", text)
cv2.waitKey(0)
cv2.destroyAllWindows()


# Our text is in yellow color on the top
```
