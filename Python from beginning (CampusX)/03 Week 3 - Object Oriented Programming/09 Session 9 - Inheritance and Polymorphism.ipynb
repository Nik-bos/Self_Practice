# New topic: `Inheritance`.
# ***Learn it from Jenny's lecture. It is very easy to understand.***

When a child class access the properties of parent class, it is called **Inheritance** .

# What all gets inherit?

* Parent's class's constructor
* Parent's class's Non private attributes and
* Parent's class's Non private methods.

```python
# Constructor example

class Animal:
    def __init__(self):
        self.no_of_legs = 4
        self.no_of_eyes = 2
        self.no_of_nose = 1

    def can_talk(self):
        return "Animals cannot talk."


class Dog(Animal):
    pass
```

```python
d = Dog()
```

```python
# Inherited constructor

print(d.no_of_legs)
print(d.no_of_eyes)
print(d.no_of_nose)
```

```python
d.can_talk()    # Inherited non-private method.
```

# Polymorphism.

`Polymorphism` means there are more than 1 methods with the same name but different behaviour. Means both methods have same name but performs different functions.

There are three concepts inside `Polymorphism`.
1) Method overriding: It is what we learnt in `Inheritance`. The concept deals with when a parent and a child class both has a method with same name, which method will get execute by the object of child class? Obviously child class's method. This is `Method overriding`.

2) `Method overloading`: This is not available in Python but we can indirectly achieve this functionality. We'll learn it later but first understand what is method overloading.<br>
Method overloading means when we have a class with two methods with same name but different no_of_parameters, then in this case, if a user calls that method, which should get execute? The execution of the method depends on the argument passed.<br>
For eg: Class Math has two methods with same name **add()** one with parameters a,b and other with parameter a,b,c.<br>
Now if a user calls add(1, 2) it calls first add() method and if a user calls add(1,2,5) it calls second add() method.<br>
This process happens in other programing languages but not in Python. Python calls the latest method defined in a class. But we can achieve this indirectly.

**See the below code to understand `Method overloading`.**

```python
# Method overloading.

class Math:

    def add(self, a, b):
        return a + b

    def add(self, a, b, c):
        return a + b + c
```

```python
m = Math()
m.add(1, 2)

# See, it is giving error that missing argument 'c'. So Python always executes the latest defined method.
```

```python
m.add(1, 2, 3)
```

# Now what if we want to achieve the functionality of `Method overloading`? 
# See the below code.

```python
# Create a default parameter.

class Math:
    def add(self, a, b, c = 0):
        if c == 0:
            return a + b
        else:
            return a + b + c
```

```python
m = Math()

print(m.add(1, 2))
print(m.add(1, 2, 3))
```

```python
# Create a variable length argument.

class Math:
    def add(self, *args):
        sum = 0
        for i in args:
            sum += i
        return sum
```

```python
m = Math()

print(m.add(1, 2))
print(m.add(1, 2, 3))
print(m.add(1, 2, 3, 4, 5))
```

# 3) 3rd concept in `Polymorphism` is ***Operator Overloading***

`Operator overloading` means there is a same operator, but its function is differet according to input.

For eg: `+`. It has different functionality for string, integer, list, dictionary each.


```python
# Example

print(5 + 1)
print('Hello' + 'World')

l1 = [1, 2, 3]
l2 = [4, 5, 6]

d1 = {'Nikunj': 'Rajkot', 'Harshit': 'Jetpur'}
d2 = {"Jenish": "Jetpur"}

print(l1 + l2)
print(d1 + d2)

# See, for all these + operator performs different function. Moreover it is not applicable to dictionaries.
```
