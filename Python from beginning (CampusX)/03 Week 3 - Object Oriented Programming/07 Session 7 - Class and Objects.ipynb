```python
class Atm:
    def __init__(self, balance):
        self.pin = 0000
        self.balance = balance
        self.menu()

    def menu(self):
        user_input = input('''
        Hii! How can I help you?
        1. Press 1 to create pin.
        2. Press 2 to change pin.
        3. Press 3 to check balance.
        4. Press 4 to withdraw.
        5. Press 5 to exit.
        ''')
    
        if user_input == '1':
            self.create_pin()
        elif user_input == '2':
            self.change_pin()
        elif user_input == '3':
            self.check_balance()
        elif user_input == '4':
            self.withdraw()
        elif user_input == '5':
            self.exit()
        else:
            print('Enter a correct number.')
            self.menu()

    def create_pin(self):
        user_pin = int(input("Set your pin: "))
        self.pin = user_pin
        print("Pin created successfully.")
        self.menu()
        
    def change_pin(self):
        current_pin = int(input("Enter your current pin: "))
        if current_pin == self.pin:
            new_pin = int(input("Enter your new pin: "))
            self.pin = new_pin
            print("Pin changed successfully.")
            self.menu()
        else:
            print("Current pin is incorrect.")
            self.menu()

    def check_balance(self):
        user_pin = int(input("Enter your pin: "))
        if user_pin == self.pin:
            print(f"Your current balance is {self.balance} rupees.")
            self.menu()
        else:
            print("Incorrect pin.")
            self.menu()
            
    def withdraw(self):
        amount_to_withdraw = int(input("Enter amount to withdraw: "))
        if amount_to_withdraw > self.balance:
            print("Insuffecient balance.")

        else:
            print("Withdraw successful.")
            self.balance = self.balance - amount_to_withdraw
        
        self.menu()

    def exit(self):
        print("Thankyou. Have a nice day.")
```

# Let's learn the concept of 'self'.

* Before moving forward, let me first tell you a **Golden rule of OOP**. It's,  `the attributes and methods of any class, can only be accessed by its objects, not even by menbers of the class`.
* This is the reason why we wrote **self.menu()** everytime we called `menu()` method inside some other method. Here, we are calling menu() method through object of the class i.e **self**.

* For eg.
* obj = Atm()    # Creating an object of `Atm` class.
* obj.create_pin()    # Calling method through an object of the class i.e `obj`. 
* obj.check_balance()    # Calling method through an object of the class i.e `obj`.
* obj.change_pin()    # Calling method through an object of the class i.e `obj`.
* obj.menu()    # Calling method through an object of the class i.e `obj`.
* obj.balance    # Calling `balance` attribute of the class through `obj`.
* obj.pin    # Calling `pin` attribute of the class through `obj`

Now, answer one question. Can I call check_balance() method inside deposite() method?<br>
The answer is **No**. As we learnt, Class's methods and attributes can only be accessed by its object, not even by its members.<br>
But here in our `Atm` class, we are calling menu() method inside other method. Then how is it possible?<br>
See carefully, we're not simply calling `menu()`, we are writting `self.menu`.<br>
Here comes the **self** into play. If we want to call any method or attribute of the class inside any other method, we take the help of **self**.<br>
See in our **Atm** class, every time we're calling `menu()`  method inside other method, we're writing **self.menu()**.<br>

Then now the question arrise is, when we can access attributes and methods of a class only through `object` and here we're accessing it through `self`, then is `self` **an object of the class**?
The answer is, ***Yes***. Self is the object of the class.

Let's prove it. We know in Python everything is an object. In this case, `self` is also an object and so it'll be stored in some memory location.<br>
Let's see what is the memory location of **self** and the object of the class we create.<br>

```python
class Fraction:
    def __init__(self, numerator, denominator):
        self.num = numerator
        self.den = denominator

    def __str__(self):
        return f'{self.num}/{self.den}'

    def __add__(self, other):
        
        new_num = (self.num * other.den) + (other.num * self.den)
        new_den = (self.den * other.den)

        return (f'{new_num}/{new_den}')

    def __sub__(self, other):
        new_num = (self.num * other.den) - (other.num * self.den)
        new_den = (self.den * other.den)

        return (f'{new_num}/{new_den}')

    def __mul__(self, other):
        new_num = (self.num * other.num)
        new_den = (self.den * other.den)

        return (f'{new_num}/{new_den}')

    def __truediv__(f1, f2):
        new_num = (f1.num * f2.den)
        new_den = (f1.den * f2.num)

        return (f'{new_num}/{new_den}')

    # Defining non-magic method.
    def convert_to_decimal(self):
        return (self.num / self.den)
```

```python
f3 = Fraction(1, 2)
f4 = Fraction(3, 4)
```

```python
# Subtraction

print(f3 - f4)
```

```python
# Multiplication

print(f3 * f4)
```

```python
# Division

print(f3 / f4)
```
