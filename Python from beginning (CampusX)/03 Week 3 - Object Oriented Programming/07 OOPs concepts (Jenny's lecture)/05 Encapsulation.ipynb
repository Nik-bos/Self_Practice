Encapsulation in Python is the concept of bundling the data (attributes) and methods (functions) that operate on the data into a single unit called a class. It allows you to restrict access to certain components of the object, preventing the direct modification of internal state from outside the class. Encapsulation helps in hiding the implementation details of a class and only exposing the necessary interfaces for interacting with the class.

`Getters and Setters:` Encapsulation often involves providing getter and setter methods to manipulate the private attributes of a class indirectly. Getter methods are used to access the values of private attributes, while setter methods are used to modify them. This allows for controlled access and modification of the internal state of objects.

```python
class Student:
    def __init__(self, name, rollno, age):
        self.name = name            # Public attribute
        self._rollno = rollno       # Protected attribute
        self.__age = age            # Private attribute

    def get_age(self):              # getter() method
        return self.__age

    def set_age(self, n):           # setter() method
        self.__age = n
        
    def __display(self):      # Private method
        print(f'Hii myself {self.name}, my roll no is {self._rollno} and my age is {self.__age} from Student class')


class Branch(Student):
    pass
        
```

```python
s = Student("Nikunj", 7, 24)
b = Branch('Jugal', 10, 20)
```

```python
print(s._rollno)
print(s.__age)
```

```python
print(b._rollno)
print(b.__age)
```

```python
s.get_age()
```

```python
s.set_age(23)
print(s.get_age())
```

```python
b.get_age()
```

```python
b.set_age(19)
b.get_age()
```
