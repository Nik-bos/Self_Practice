# What are access specifiers?

Access specifiers or access modifiers are just keywords which helps us to restrict access of class methods or attributes from outside the class.


Access specifiers, also known as access modifiers, are keywords used in object-oriented programming languages like Python to control the accessibility of class members (attributes and methods) from outside the class

There are three types of access specifiers in Python:

1) `Public:`
* Members (attributes and methods) are accessible from outside the class.
* In Python, by default, all members are public.

2) `Protected:`
* Members are accessible within the class and its subclasses (if any), but not from outside the class hierarchy.
* In Python, conventionally, members prefixed with a single underscore (_) are considered protected, indicating that they are intended for internal use    within the class and its subclasses
* We can access protected members(methods/attributes) outside the class but we should not do it. It is ethic that we should not use protected members      outside the class's area

3) `Private:`
* Members are accessible only within the class itself and not from outside the class or not even from child class.
* In Python, conventionally, members prefixed with double underscores (__) are considered private
*  However, Python name mangling is used to make t  difficult, but not impossible, to access these members from outside the clas
* Name mangling means, we know we can't access private member outside the class, but to get them accessed outside the class, we use the technique of 'Name mangling'.
* It is nothing but just a drawback of Python. Here, we create public methods within the class and access those private members on these public methods.
* Then to access the private members outside the class, we call public methods outside the class and indirectly call private members as well through i..

# Public method and attribute

```python
class Student:
    def __init__(self, name):
        self.name = name    # Public attribute
    def display_name(self):    # Public method
        print(f"My name is {self.name}")
```

```python
s = Student('Nikunj')
s.display_name()    # Accessing public method outside the class
s.name    # Accessing public attributee outside the class
```

# 2) Protected method and attribute: 

* Members are accessible within the class and its subclasses (if any), but not from outside the class hierarchy.
* In Python, conventionally, members prefixed with a single underscore (_) are considered protected, indicating that they are intended for internal use    within the class and its subclasses.
* We can access protected members(methods/attributes) outside the class but we should not do it. It is ethic that we should not use protected members      outside the class's area.

```python
class Student:
    def __init__(self, name, age):
        self.name = name    # Public attribute
        self._age = age    # Protected attribute

    def display_name(self):
        print(f"My name is {self.name}")

    def _display_age(self):    # Protected method
        print(f"My age is {self._age}")


class Demo(Student):
        def show(self):
             print('This is Demo Class')
             print(f"My name is {self.name} and age is {self._age}")
```

```python
d = Demo('Nikunj', 26)
```

```python
d.show()
```

```python
print(d.name)
print(d._age)
```

# 3) Private
* Members are accessible only within the class itself and not from outside the class or child class as well
* We can't access private attribute and methods within the derived class as well
* In Python, conventionally, members prefixed with double underscores (__) are considered private.
* However, Python name mangling is used to make it difficult, but not impossible, to access these members from outside the class
* Name mangling means, we know we can't access private member outside the class, but to get them accessed outside the class, we use the technique of 'Name mangling'.
* It is nothing but just a drawback of Python. Here, we create public methods within the class and access those private members on these public methods.
* Then to access the private members outside the class, we call public methods outside the class and indirectly call private members as well through it

```python
class Student:
    def __init__(self, name, age, contact):
        self.name = name    # Public attribute
        self._age = age    # Pritected attribute
        self.__contact = contact    # Private attribute

    def display(self):
        print(f"My name is {self.name}, age is {self._age}, contact is {self.__contact}")

    def __private_method(self):
        print("I am a private method. You cannot access me outside the class.")

class Demo(Student):
    def display_name(self):
        print(f"My name is {self.name}")

    def display_age(self):
        print(f"My age is {self._age}")    # Accessing protected attribute in child class

    def display_contact(self):
        print(f"My contact is {self.__contact}")    # Accessing private attribute in child class
```

```python
d = Demo("Nikunj", 26, 7383196410)
```

```python
d.display_name()
```

```python
d.display_age()
```

```python
d.display_contact()
```
