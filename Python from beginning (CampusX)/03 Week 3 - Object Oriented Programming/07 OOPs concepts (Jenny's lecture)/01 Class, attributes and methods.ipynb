# Methods in Python

```python
# Attribute is what one has and method is what one does like add(), len(), min(), max()
```

```python
class Instructor:
    followers = 0
    def __init__(self, name, city, state):
        self.name = name
        self.city = city
        self.state = state

    def display(self, sub_name):
        print(f'This is {self.name} and I teach {sub_name} ')

    def update_follower(self, follower_name):
        self.followers += 1
```

```python
inst1 = Instructor('Nikunj', 'Rajkot', "Gujarat")
inst1.display('Python')
```

```python
inst2 = Instructor('Jugal', 'Jetpur', "Gujarat")
inst2.display("English")
```

```python
inst1.update_follower('Nikunj')
inst1.followers
```

# What is __init__(): 
It is a constructor in Python. It gets called automatically when we create an object of a class.

```python
class Student:
    def __init__(self, name, rollno):
        self.name = name
        self.rollno = rollno
        
class Branch(Student):
    def __init__(self, branch_name):
        self.branch_name = branch_name
```

```python
b = Branch()    # will give error bcz constructor will get called and it has one required argument.
```

```python
b = Branch('DS')
b.branch_name
```
