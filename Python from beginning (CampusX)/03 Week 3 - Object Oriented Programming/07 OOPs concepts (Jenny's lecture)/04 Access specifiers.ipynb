# What are access specifiers?

Access specifiers or access modifiers are just keywords which helps us to restrict access of class methods or attributes from outside the class.


Access specifiers, also known as access modifiers, are keywords used in object-oriented programming languages like Python to control the accessibility of class members (attributes and methods) from outside the class

There are three types of access specifiers in Python:

1) `Public:`
* Members (attributes and methods) are accessible from outside the class.
* In Python, by default, all members are public.

2) `Protected:`
* Members are accessible within the class and its subclasses (if any), but not from outside the class hierarchy.
* In Python, conventionally, members prefixed with a single underscore (_) are considered protected, indicating that they are intended for internal use    within the class and its subclasses
* We can access protected members(methods/attributes) outside the class but we should not do it. It is ethic that we should not use protected members      outside the class's area

3) `Private:`
* Members are accessible only within the class itself and not from outside the class or even child class.
* In Python, conventionally, members prefixed with double underscores (__) are considered private
*  However, Python name mangling is used to make t  difficult, but not impossible, to access these members from outside the clas
* Name mangling means, we know we can't access private member outside the class, but to get them accessed outside the class, we use the technique of 'Name mangling'.
* It is nothing but just a drawback of Python. Here, we create public methods within the class and access those private members on these public methods.
* Then to access the private members outside the class, we call public methods outside the class and indirectly call private members as well through i..

# Public method and attribute

```python
class Student:
    def __init__(self, name):
        self.name = name    # Public attribute

    def display(self):    # Public method
        print(f'Hii myself {self.name} from Student class')
        
```

```python
s = Student('Nikunj')
s.display()
s.name
```

# 2) Protected method and attribute: 

* Members are accessible within the class and its subclasses (if any), but not from outside the class hierarchy.
* In Python, conventionally, members prefixed with a single underscore (_) are considered protected, indicating that they are intended for internal use    within the class and its subclasses.
* We can access protected members(methods/attributes) outside the class but we should not do it. It is ethic that we should not use protected members      outside the class's area.

```python
class Student:
    def __init__(self, name, rollno):
        self.name = name    # Public attribute
        self._rollno = rollno       # Protected attribute
    
    def display(self):      # Public method
        print(f'Hii myself {self.name} and my roll no is {self._rollno} from Student class')

class Branch(Student):
    def show(self):
        print(f'{self.name} roll_no is {self._rollno}')
        
```

```python
b1 = Branch('Nikunj', 7)
```

```python
print(b1.name)
print(b1._rollno)
```

```python
b1.display()
```

```python
b1.show()
```

# 3) Private
* Members are accessible only within the class itself and not from outside the class.
* We can't access private attribute and methods within the derived class as well
* In Python, conventionally, members prefixed with double underscores (__) are considered private.
* However, Python name mangling is used to make it difficult, but not impossible, to access these members from outside the class
* Name mangling means, we know we can't access private member outside the class, but to get them accessed outside the class, we use the technique of 'Name mangling'.
* It is nothing but just a drawback of Python. Here, we create public methods within the class and access those private members on these public methods.
* Then to access the private members outside the class, we call public methods outside the class and indirectly call private members as well through it

```python
class Student:
    def __init__(self, name, rollno, age):
        self.name = name            # Public attribute
        self._rollno = rollno       # Protected attribute
        self.__age = age            # Private attribute
    
    def display(self):      # Public method
        print(f'Hii myself {self.name}, my roll no is {self._rollno} and my age is {self.__age} from Student class')


class Branch(Student):
    def show(self):
        print(f'Age is {self.__age}')
        
```

```python
s = Student('Nikunj', 7, 24)
```

```python
s.__age    # See, we cant access private attribute outside the class
```

```python
s.display()    # Altough we can't access __age outside the class, we came to know age by accessing it in display() method inside the class
```

```python
b = Branch('Nikunj', 7, 24)
```

```python
b.show()    # See, we can't access __age(private attribute) within the derived class as well
```

```python
b.display()
```

# Ways to access private methods and attributes.

There're three ways to access private attributes and methods, they're:
* Create a public method in the class and access private attribute and method on that public method
* Use name mangling
* Use getter - setter methods. getter() to get the method or attributes and setter() to modify them.

# `1)` Through public method in the class.

```python
class Student:
    def __init__(self, name, rollno, age):
        self.name = name            # Public attribute
        self._rollno = rollno       # Protected attribute
        self.__age = age            # Private attribute
    
    def __display(self):      # Private method
        print(f'Hii myself {self.name}, my roll no is {self._rollno} and my age is {self.__age} from Student class')

    def show_private_method(self):
        self.__display()


class Branch(Student):
    pass
        
```

```python
s = Student('Nikunj', 7, 24)
```

```python
s.__display()
```

```python
s.show_private_method()
```

# `3)` getter() and setter() methods.
getter() method is used to get or access the private attributes and private methods of the class and 
setter() method is used to modify the private attributes and private methods of the class

```python
class Student:
    def __init__(self, name, rollno, age):
        self.name = name            # Public attribute
        self._rollno = rollno       # Protected attribute
        self.__age = age            # Private attribute

    def get_age(self):              # getter() method
        return self.__age

    def set_age(self, n):
        self.__age = n
        
    def __display(self):      # Private method
        print(f'Hii myself {self.name}, my roll no is {self._rollno} and my age is {self.__age} from Student class')

    def show_private_method(self):    # Public method.
        self.__display()


class Branch(Student):
    pass
        
```

```python
s = Student("Nikunj", 7, 24)
b = Branch('Jugal', 10, 20)
```

```python
print(s._rollno)
print(s.__age)
```

```python
print(b._rollno)
print(b.__age)
```

```python
s.get_age()
```

```python
s.set_age(23)
print(s.get_age())
```

```python
b.get_age()
```

```python
b.set_age(19)
b.get_age()
```
