```python
class Duck:
    def swim(self):
        print("I'm a duck and I can swim.")

    def speaks(self):
            print('Quack Quack!!')

class Dog:
    def swim(self):
        print("I'm a dog and I can't swim")

    def speaks(self):
        print("Dog barks")

def display(obj):
    obj.swim()
```

```python
o = Dog()

display(o)
```

# Another example

```python
class dog:
    def sound(self):
        print("Dog barks")
    
class cat:
    def sound(self):
        print("Cat mews")
        
class duck:
    def sound(self):
        print("Duck quacks")
        
def duck_typing(obj):
    obj.sound()
    
```

```python
for obj in dog(), cat(), duck():
    duck_typing(obj)
```

```python
d=dog()
c=cat()
```

```python
duck_typing(d)
```

```python
duck_typing(c)
```

```python
duck_typing(duck())
```
