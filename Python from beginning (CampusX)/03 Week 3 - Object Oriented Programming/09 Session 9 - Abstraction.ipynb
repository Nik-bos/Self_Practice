```python
from abc import ABC, abstractmethod
```

```python
class BankApp(ABC):

    def database(self):
        return('Database Connected.')
    
    @abstractmethod
    def security(self):
        pass

    @abstractmethod
    def display(self):
        pass

#-----------------------------------------------------------

class MobileApp(BankApp):
    
    def login(self):
        return "Login Successful"

    def security(self):
        return 'Mobile Security'
    
    def display(self):
        print('display')
```

```python
m = MobileApp()

m.database()
```
