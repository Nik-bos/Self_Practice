# Polymorphism.

`Polymorphism` means there are more than 1 methods with the same name but different behaviour. Means both methods have same name but performs different functions.

There are three concepts inside `Polymorphism`.
1) `Method overriding`: It is what we learnt in `Inheritance`. The concept deals with when a parent and a child class both has a method with same name, which method will get execute by the object of child class? Obviously child class's method. This is `Method overriding`.

2) `Method overloading`: This is not available in Python but we can indirectly achieve this functionality. We'll learn it later but first understand what is method overloading.<br>
Method overloading means when we have a class with two methods with same name but different no_of_parameters, then in this case, if a user calls that method, which should get execute? The execution of the method depends on the argument passed.<br>
For eg: Class Math has two methods with same name **add()**, one with parameters a,b and other with parameter a,b,c.<br>

Now if a user calls add(1, 2) it calls first add() method and if a user calls add(1,2,5) it calls second add() method.<br>

This process happens in other programing languages but not in Python. Python calls the latest method defined in a class. But we can achieve this indirectly.

**See the below code to understand `Method overloading`.**

```python
# Method Overloading

class Math:
    def add(self, a, b):
        return a+b
    def add(self, a, b, c):
        return a+b+c
```

```python
obj = Math()
obj.add(5,8, 5)
```

```python
# Method Overloading is not available in Python
# but we can achieve its functionality indirectly.
# See the below code

class Math:
    def add(self, a, b, c=0):
        return a+b+c


obj = Math()
obj.add(5, 4)
```

# 'Static variable (Class Variables)' VS 'Instance Variable'

We know what is **Instance variable**. Now we want to create a Bank class where we should have one Cust_id attribute and it should automatically increase by 1 each time a new customer joins a bank.

```python
class Bank:
    def __init__(self, name):
        self.cust_id = 0
        self.cust_id += 1
        self.name = name
```

```python
obj1 = Bank('Nikunj')
obj1.cust_id
```

```python
obj2 = Bank("Jugal")
obj2.cust_id
```

```python
class Bank:
    __counter = 0
    def __init__(self, name):
        self.name = name
        Bank.__counter+=1
        print(f"Counter: {Bank.__counter}")
```

```python
obj1 = Bank("Nikunj")
```

```python
obj2 = Bank('Jugal')
```

```python
obj3 = Bank('Harshit')
```

```python
Bank.counter = 'Hehehe'
```

```python
obj4 = Bank("Jenish")
```

```python
obj3._Bank__counter
```

```python
class HeadOffice:
    lunch_time = 60    # Static/Class Variable

    @staticmethod
    def set_lunch_time(new_time):
        HeadOffice.lunch_time = new_time


class Branch(HeadOffice):
    pass
```

```python
HeadOffice.lunch_time
```

```python
Branch.set_lunch_time(40)
```

```python
Branch.lunch_time
```

```python
# Class Method

class Head:
    lunch_time = 60    # Class(Static) Variable

    @classmethod
    def set_lunch_time(cls, value):
        cls.lunch_time = value


class Branch(Head):
    pass
```

```python
obj_head = Head()
obj_branch = Branch()
```

```python
print(obj_head.lunch_time)
print(obj_branch.lunch_time)
```

```python
obj_branch.set_lunch_time(40)
```

```python
print(obj_head.lunch_time)
print(obj_branch.lunch_time)
```
