# Question 5: A user can find a distance between a given 2-D point and a given line.

Formula to find the shortes distance is given below.

distance = `| Ax + By + C |/ √(A**2 + B**2)`

```python
# Question 4: A user can check if a point lies on a given line.

class Point:
    def __init__(self, x, y):
        self.x_cod = x
        self.y_cod = y

    def __str__(self):
        return f'<{self.x_cod}, {self.y_cod}>'

# Question 1 complete: A user can create and view 2-D coordinated.

    def euclidean_dist(self, other):
        return math.sqrt((self.x_cod - other.x_cod)**2 + (self.y_cod - other.y_cod)**2)

# Question 2 complete: A user can find out the distance between two coordinated.

    def dist_from_orig(self):
        return self.euclidean_dist(Point(0,0))
        # return math.sqrt((self.x_cod - 0)**2 + (self.y_cod - 0)**2)

# Question 3 complete: A user can find the distance of a coordinate from the origin.



class Line:
    def __init__(self, A, B, C):
        self.a = A
        self.b = B
        self.c = C

    def __str__(self):
        return f'{self.a}x + {self.b}y + {self.c} = 0'

    def point_on_line(line, point):
        if line.a*point.x_cod + line.b*point.y_cod + line.c == 0:    # ax + by + c = 0 
            return 'Point lies on the line'
        else:
            return 'Point does not lie on the line.'

# Question 4 complete: A user can check if a point lies on a given line.

    def shortest_dist(line, point):
        return abs(line.a*point.x_cod + line.b*point.y_cod + line.c) / math.sqrt(line.a**2 + line.b**2)
```

```python
p4 = Point(3, -2)
print(p4)
```

```python
l = Line(2, 5, 4)
print(l)
```

```python
l.point_on_line(p4)
```

```python
# We know that point(3, -2) lies on the line 2x + 5y + 4. So, shortest distance should be zero. Let's see.

l.shortest_dist(p4)

# We're getting zero. So our code is correct.
```

# Reference variable.

* Reference variables holds the objects.
* We can create an object without reference variable as well
* An object can have multiple reference variables.
* Assigning a new reference variable to an existing object does not create a new object.

```python
# What is a reference variable?

Person('Nikunj', 'India')     # This will create an object without referece variable.
```

```python
# When we store an object to any variable, that particular variable is called referenve variable

p = Person('Nikunj', 'India')    # Here, 'p' is a reference variable.

# Till now we were saying that 'p' is the object of Person class, but technically it is worng. 'p' is not the object, it is storing
# the reference of the object. (If this creates any confusion, forget it. Remember the old concept that 'p' is the object of the Person class)
```

```python
a = p    # An object can have multiple reference variables.
```

```python
print(p.name)
print(a.name)

# Asigoning a new reference variable to an existing object does not create a new object.
```

```python
# Change the value of a attribute using second object.

a.name = 'Jugal'
```

```python
print(p.name)
print(a.name)

# Since reference of both the variables is same, change in any one variable reflects into the other.
```

# 'Static variable' VS 'Instance Variable'

We know what is **Instance variable**. Now we want to create a Bank class where we should have one Cust_id attribute and it should automatically increase by 1 each time a new customer joins a bank.

```python
class Bank:
    def __init__(self, name):
        self.name = name
        self.cust_id = 0
        self.cust_id += 1
```

```python
c1 = Bank('Nikunj')
c2 = Bank('Jugal')
c3 = Bank("Jenish")
```

```python
print(c1.cust_id)
print(c2.cust_id)
print(c3.cust_id)
```

See, customer id is 1 every time a new customer joins. This should not happen. Here comes the Static variable into play. Static variables are class variables and instance variables are Object's variables.

Static variables are created inside class and outside methods. See the below code.

```python
class Bank:
    counter = 1
    def __init__(self, name):
        self.name = name
        self.cust_id = Bank.counter
        Bank.counter += 1
```

```python
c1 = Bank('Nikunj')
c2 = Bank('Jugal')
c3 = Bank("Jenish")
```

```python
print(c1.cust_id)
print(c2.cust_id)
print(c3.cust_id)
```

See, now it's working. Moreover, instance variables are accessed by self and static variables are accessed by class's name(Here, Bank.cust_id. So whereever we see any variable accessed by self understand that it is an instance variable and whenever we see any variable accessed by class's name, understand that it is a static variable.

# Static methods.
Those methods which does not require object to access and which can be accessed by *class_name* are called `Static Method`. We write `@staticmethod` to identify a static method. See the below code to understand properly. As we're accessing these methods with **class_name** we don't require **self** as its first parameter.

```python
class Bank:
    __counter = 1
    def __init__(self, name):
        self.name = name
        self.cust_id = Bank.__counter
        Bank.__counter += 1

    #@staticmethod
    def get_counter():
        return Bank.__counter
```

```python
c4 = Bank('Parth')
```

```python
# We cannot access static methods using objects. We can access it by using class_name
```

```python
c4.get_counter()    # See, we're getting an error message that one argument is given.
```

```python
Bank.get_counter()    # See, we can access it now.
```

```python
# Static Variable (Class Variable)

class Bank:
    __counter = 1
    def __init__(self, name):
        self.name = name
        self.cust_id = Bank.__counter
        Bank.__counter += 1
```

```python
c1 = Bank('Nikunj')
c2 = Bank('Jugal')
c3 = Bank("Jenish")
```

```python
print(c1.cust_id)
print(c2.cust_id)
print(c3.cust_id)
```

```python
class Team:
    members = []                # class variable, a list!

    def add(self, person):
        self.members.append(person)

india = Team()
nepal = Team()
india.add("Asha")

print(nepal.members)   # ['Asha']  <- Surprise! b never added anyone

# Here we added Asha on object india, but it added for object nepal as well
# To avoid this, instance variables are used.
```

```python
# Static Method

class Head:
    lunch_time = 60    # Class(Static) Variable

    @staticmethod
    def set_lunch_time(value):
        Head.lunch_time = value


class Branch(Head):
    pass
```

```python
Branch().set_lunch_time(40)

print(Branch.lunch_time)
print(Head.lunch_time)

# See, it changed for both the classes. BUt we only want to change the lunch time of Branch.
# Here comes class methods into play. See the below code
```

```python
# Class Method

class Head:
    lunch_time = 60    # Class(Static) Variable

    @classmethod
    def set_lunch_time(cls, value):
        cls.lunch_time = value


class Branch(Head):
    pass
```

```python
Branch.set_lunch_time(40)

print(Branch.lunch_time)
print(Head.lunch_time)

```

```python
Branch().set_lunch_time(35)    
# Here we are calling set_lunch_time using object of the class, even then it works the same way.
# This is bcz when we access class method using the object of the class, 
# Python ignores the object and passes the class to 'cls' parameter.

print(Branch.lunch_time)
print(Head.lunch_time)
```
