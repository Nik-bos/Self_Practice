# Colored Scatterplots

```python
iris = pd.read_csv(r"Datasets/iris.csv")
iris.head()
```

```python
# Find how SepalLength and petallength are co-related

plt.scatter(iris['SepalLengthCm'], iris['PetalLengthCm'])
plt.xlabel('Sepal_Length')
plt.ylabel('Petal_Length')
plt.title("Sepal_Length VS Petal_Length")
```

```python
# Now we wanna give different color to dots representing different flowers.
# In this case, we will first have to assign numbers to each unique specie
data = {'Iris-setosa': 0,
        'Iris-versicolor': 1,
        'Iris-virginica': 2}

iris['Species'] = iris['Species'].replace(data)
iris.sample(5)
```

```python
# Colored scatter plot: 'c' parameter
plt.scatter(iris['SepalLengthCm'], iris['PetalLengthCm'], c = iris['Species'])
plt.xlabel('Sepal_Length')
plt.ylabel('Petal_Length')
plt.title("Sepal_Length VS Petal_Length")
```

```python
# Colored scatter plot: 'cmap' parameter
# There are various cmap options available. You can search on google
# Cmap: plasma, magma, inferno, viridis, cividis, jet etc
# colorbar(): Shows color bar indicating which color belongs to which number
# alpha: Manages opacity of the dots. Range is between 0-1


plt.scatter(iris['SepalLengthCm'], iris['PetalLengthCm'], c = iris['Species'], cmap = 'jet', alpha = 1)
plt.xlabel('Sepal_Length')
plt.ylabel('Petal_Length')
plt.title("Sepal_Length VS Petal_Length")
plt.colorbar()
```

```python
scatter = plt.scatter(iris['SepalLengthCm'],
                      iris['PetalLengthCm'],
                      c=iris['Species'],
                      cmap='jet',
                      alpha=1)

cbar = plt.colorbar(scatter)
cbar.set_ticks([0, 1, 2])
cbar.set_ticklabels(['Setosa', 'Versicolor', 'Virginica'])

plt.xlabel('Sepal_Length')
plt.ylabel('Petal_Length')
plt.title("Sepal_Length VS Petal_Length")
plt.show()
```

# Horizontal and Vertical Lines:

- We can also plot horizontal and verticaal line on our graph.
- Supose we are selectors of the batters and we wanna select only those batters <br>
whose strike_rate is greater than 130.

**See the below code**

```python
plt.figure(figsize = (12, 6))
plt.scatter(sample_batters['avg'], sample_batters['strike_rate'], s = sample_batters['runs']/20)

plt.axhline(130, color = 'red')

# Annotation
font_dict = {'size': 10, 'color': 'purple'}
for i in range(len(sample_batters)):
    plt.text(sample_batters['avg'].values[i],
             sample_batters['strike_rate'].values[i],
             sample_batters['batter'].values[i],
             fontdict = font_dict)
    
# axhline means, axis_horizontal_line creates line by y-axis so we pass point where we wanna ccreate this line.
# Here it is 130, means we want to create a horizontal line on y-axis = 130

# Now, we can easily know batters whose strike rate is above 130

# Now supose someone comes and says, instead of selecting batters whose strike_rate is greater than 130,
# we will select those batters whose strike_rate > 130 and average > 30
# In this case, as average is in x_axis, we will create a vertical line using axvline()
# See then below code
```

```python
plt.figure(figsize = (12, 6))
plt.scatter(sample_batters['avg'], sample_batters['strike_rate'], s = sample_batters['runs']/20)

plt.axhline(130, color = 'red')
plt.axvline(30, color = 'red')

# Annotation
font_dict = {'size': 10, 'color': 'purple'}
for i in range(len(sample_batters)):
    plt.text(sample_batters['avg'].values[i],
             sample_batters['strike_rate'].values[i],
             sample_batters['batter'].values[i],
             fontdict = font_dict)
    
# We can create as manu line we want
# Now create a line for strike_rate>140 (Green color)
```

```python
plt.figure(figsize = (12, 6))
plt.scatter(sample_batters['avg'], sample_batters['strike_rate'], s = sample_batters['runs']/20)

plt.axhline(130, color = 'red')
plt.axhline(140, color = 'green')
plt.axvline(30, color = 'red')

# Annotation
font_dict = {'size': 10, 'color': 'purple'}
for i in range(len(sample_batters)):
    plt.text(sample_batters['avg'].values[i],
             sample_batters['strike_rate'].values[i],
             sample_batters['batter'].values[i],
             fontdict = font_dict)
    
# We can create as manu line we want
# Now create a line for strike_rate>140 (Green color)
```

# Subplots:
- This is a very interesting topic.
- Till now you must have noticed that whenever we plot graphs, it only plots a single graph.
- But what if we wanna create multiple graphs side by side for comparison?
- Here comes subplots into play.

`Note`: There are two ways of plotting on graphs. See both the ways. (First is which we are using, see the second carefully.)

```python
plt.figure(figsize = (10,5))

plt.scatter(sample_batters['avg'], sample_batters['strike_rate'])
plt.xlabel('Average')
plt.ylabel('Strike_Rate')
plt.title("Avg VS Strike_Rate")

plt.show()
```

```python
# 2nd method
figure, axis = plt.subplots(figsize = (10,5))    # figsize is passed here

axis.scatter(sample_batters['avg'], sample_batters['strike_rate'], color = 'orange')

axis.set_xlabel('Average')    # Here it is set_xlabel
axis.set_ylabel('Strike_Rate')    # Here it is set_ylabel
axis.set_title("Avg VS Strike_Rate")    # Here it is set_title

figure.show()    # Inplace of plt.show()
```

```python
plt.subplots(nrows = 2, ncols = 2, figsize = (6,6))
# See, we are getting a figure object
# (<Figure size 600x600 with 4 Axes>,

#  and an array of axis objects
# array(
# [
#   [<AxesSubplot: >, <AxesSubplot: >],
#   [<AxesSubplot: >, <AxesSubplot: >]
# ]

# array[0] returns
# [
#   [<AxesSubplot: >, <AxesSubplot: >],
#   [<AxesSubplot: >, <AxesSubplot: >]
# ]

# array[0][1] returns [<AxesSubplot: >, <AxesSubplot: >]
# array[0][2] returns [<AxesSubplot: >, <AxesSubplot: >]
```

```python
fig1, axis= plt.subplots(nrows = 2, ncols = 1, figsize = (10,7)) # We only want 2 row and 1 columns

# average Vs strike_rate
axis[0].scatter(batters['avg'], batters['strike_rate'], color = 'red')
axis[0].set_xlabel('Average')
axis[0].set_ylabel('Strike_Rate')
axis[0].set_title("Average VS Strike_Rate")

# runs Vs average
axis[1].scatter(batters['avg'], batters['runs'], color = 'blue')
axis[1].set_xlabel('Average')
axis[1].set_ylabel('Runs')
axis[1].set_title("Runs VS Average")

# See, x_label of 0 and title of 1 are oveerlaping.
# Note, here xlabel of both the graphs are same- "Average", so we can share them
# See below code
```

```python
fig1, axis= plt.subplots(nrows = 2, ncols = 1, figsize = (10,7), sharex= True) # We only want 2 row and 1 columns

# average Vs strike_rate
axis[0].scatter(batters['avg'], batters['strike_rate'], color = 'red')
# axis[0].set_xlabel('Average')
axis[0].set_ylabel('Strike_Rate')
axis[0].set_title("Average VS Strike_Rate")

# runs Vs average
axis[1].scatter(batters['avg'], batters['runs'], color = 'blue')
axis[1].set_xlabel('Average')
axis[1].set_ylabel('Runs')
axis[1].set_title("Runs VS Average")
```

```python
# Side by side comparison

fig1, axis= plt.subplots(nrows = 1, ncols = 2, figsize = (18,5)) # We only want 1 row and 2 columns

# average Vs strike_rate
axis[0].scatter(batters['avg'], batters['strike_rate'], color = 'red')
axis[0].set_xlabel('Average')
axis[0].set_ylabel('Strike_Rate')
axis[0].set_title("Average VS Strike_Rate")

# runs Vs average
axis[1].scatter(batters['avg'], batters['runs'], color = 'blue')
axis[1].set_xlabel('Average')
axis[1].set_ylabel('Runs')
axis[1].set_title("Runs VS Average")

# See, x_label of 0 and title of 1 are oveerlaping.
# Note, here xlabel of both the graphs are same- "Average", so we can share them
# See below code
```

```python
# Side by side comparison

fig1, axis= plt.subplots(nrows = 2, ncols = 2, figsize = (18,6)) # We only want 1 row and 2 columns

# average Vs strike_rate
axis[0, 0].scatter(batters['avg'], batters['strike_rate'], color = 'red')
axis[0, 0].set_xlabel('Average')
axis[0, 0].set_ylabel('Strike_Rate')
axis[0, 0].set_title("Average VS Strike_Rate")

# runs Vs average
axis[0, 1].scatter(batters['avg'], batters['runs'], color = 'blue')
axis[0, 1].set_xlabel('Average')
axis[0, 1].set_ylabel('Runs')
axis[0, 1].set_title("Runs VS Average")

# Histogram of average
axis[1, 0].hist(batters['avg'], color = 'red')
axis[1, 0].set_title("Histogram of Average")

# Histogram of Runs
axis[1, 1].hist(batters['runs'], color = 'blue')
axis[1, 1].set_title("Histogram of Runs")


# See, x_label of 0 and title of 1 are oveerlaping.
# Note, here xlabel of both the graphs are same- "Average", so we can share them
# See below code
```

# 3) Heatmap

```python
deliveries = pd.read_csv(r"Datasets/IPL_Ball_by_Ball_2008_2022.csv")

deliveries.head()
```

```python
# Here in deliveries['ballnumber] we are having more than 6 balls as well.
# ballnumber indicates ball of the over
# So, we are filtering only 1-6 ballnumber and where 6 is hitted

temp_df = deliveries[
    (deliveries['ballnumber'].isin([1, 2, 3, 4, 5, 6])) &
    (deliveries['batsman_run'] == 6)
]

temp_df
```

```python
# Creating pivot table

grid = temp_df.pivot_table(index = 'overs', columns = 'ballnumber', values = 'batsman_run',  aggfunc = 'count')

grid
```

```python
plt.figure(figsize=(10,6))

plt.imshow(grid, aspect = 'auto')
plt.xlabel("Ball Number")
plt.ylabel("Overs")

plt.colorbar()
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))

plt.imshow(grid, aspect='auto')
plt.colorbar()

plt.xticks(range(len(grid.columns)), grid.columns)
plt.yticks(range(len(grid.index)), grid.index)

plt.xlabel("Ball Number")
plt.ylabel("Overs")
plt.title("Heatmap of total no_of sixes in an over")

plt.show()
```
