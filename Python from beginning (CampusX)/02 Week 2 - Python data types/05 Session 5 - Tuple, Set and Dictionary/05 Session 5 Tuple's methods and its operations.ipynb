# Tuple and its operations.

-> Python's tuple are the collection of elements of different data types.

-> The process of assigning values to a tuple is called 'Packing'.

-> Besides packing, there's a special feature in python i.e 'unpacking'.


```python
# Tuple packing

a = ('A', 'B', 'C', 'D')    # Elements of same datatypes
print(f"Tuple with the elements of same datatypes: {a}\n")

b = ("A", 'B', "C", 1, 2, 3.6, 2.5)    # Elements of different datatypes i.e a string, an integer and a float'
print(f"Tuple with the elements of different datatypes: {b}\n")

c = ("A", "b", "C", (1, 2, 3))    # Nested tuple
print(f"Nested tuple: {c}\n")

d = ("A", "B", "C", [1, 2, 3])    # Tuple with list inside it as an element
print(f"Tuple with a list as an element inside it: {d}\n")
```

```python
# Note: If there's only one element in a tuple, we should end it with a comma, else it considers it a string.
# Moreover, if we enter any sequence seperated by comma, Python considers it as tuple
# for example tup = 'A', "B", "C",

e = ('Nikunj')
t = (type(e))
print(f"Without comma: {t}")

f = ("Nikunj",)
t2 =type(f)
print(f"With comma: {t2}")
```

```python
fruits = ("Apple", "Banana", "Orange")
(red, yellow, orange) = fruits

print(red)
print(yellow)
print(orange)
```

```python
fruits2 = 'Apple', "Banana", "Grapes", "Orange", "Kiwi", "Dragon",
(red, yellow, green, *colors) = fruits2
print(red)
print(yellow)
print(colors)
```

```python
'''
If the asterisk is applied to another variable name than the last, then Python keeps assigning values to that variable
untill the no.of variables left and no.of values left are equal.
'''

fruits3 = ("Apple", "Banana", "Orange", "Kiwi", "Dragon", "Mango", 'Grapes')
(red, yellow, *colors, safron, green) = fruits3

print(red)
print(yellow)
print(colors)
print(safron)
print(green)
```

```python
print(type(colors))
```

# Methods of tuples

```python
# 1) len(): This returns the length of the tuple.

# 2) count(): This method returns the no.of time element is repeated in a tuple.

# 3) index(): This method returns the index of an element passed to it.

# 4) sorted() = This function takes a tuple as an input and returns a sorted list as an output.
# Moreover, this doesn't make any changes in the original tuple.

# 5) min() : Returns the smallest number. It takes a tuple as an input
# 6} max(): Returns the greatest number. It takes a tuple as an input
# 7) sum(): Returns the sum of all the numbers. It takes a tuple as an input

t1 = ("Harshit", "Jenish", "Parth", "Jugal", "Khilan")
len(t1)
```

```python
# 2) count(): This method returns the no.of time element is repeated in a tuple

t2 = ("Harshit", "Jenish", "Parth", "Jugal", "Khilan", "Harshit", "Harshit")
t2.count("Harshit")
```

```python
# 3) index(): This method returns the index of an element passed to it.

t3 = ("Harshit", "Jenish", "Parth", "Jugal", "Khilan")
t3.index("Jenish")
```

```python
# 4) sorted() = This function takes a tuple as an input and returns a sorted list as an output.
# Moreover, this doesn't make any changes in the original tuple.

t4 = (1, 5, 2, 45, 22, 48, 12, 10, 9)

print(sorted(t4))
```

```python
# 5) min() : Returns the smallest number. It takes a tuple as an input
# 6} max(): Returns the greatest number. It takes a tuple as an input
# 7) sum(): Returns the sum of all the numbers. It takes a tuple as an input

ty = (1,2,3,4,5)
print(sum(ty))
print(min(ty))
print(max(ty))
```

# Tuples

A tuple in Python is similar to a list. The difference between the two is that we cannot change the elements of a tuple once it is assigned whereas we can change the elements of a list.

In short, a tuple is an immutable list. A tuple can not be changed in any way once it is created.

Characterstics

- Ordered
- Unchangeble
- Allows duplicate

### Plan of attack

- Creating a Tuple
- Accessing items 
- Editing items
- Adding items
- Deleting items
- Operations on Tuples
- Tuple Functions

### Creating Tuples

```python
# empty
t1 = ()
print(t1)
# create a tuple with a single item
t2 = ('hello',)
print(t2)
print(type(t2))
# homo
t3 = (1,2,3,4)
print(t3)
# hetro
t4 = (1,2.5,True,[1,2,3])
print(t4)
# tuple
t5 = (1,2,3,(4,5))
print(t5)
# using type conversion
t6 = tuple('hello')
print(t6)
```

### Accessing Items

- Indexing
- Slicing

```python
print(t3)
print(t3[0])
print(t3[-1])
```

```python
t5[-1][0]
```

### Operations on Tuples

```python
# + and *
t1 = (1,2,3,4)
t2 = (5,6,7,8)

print(t1 + t2)

print(t1*3)
# membership
1 in t1
# iteration
for i in t1:
  print(i)
```

* Tuple unpacking

```python
a, b, c = (1,2,3,)
print(a, b, c, sep = '-')

# No.of variables and the elements to unpack should be same else it will give an error.
# for e.g:
# a,b = (1, 2, 3) -  This will give an error.
```

```python
a, b, *others = (1, 2, 3, 4, 5, 6)
print(a,b)
print(others)

'''Use of asterisk : 1 and 2 will be assigned to 'a' and 'b' respectively and rest all the elements 
will be assigned to variable 'others' in the form of list.'''

print(type(others))
```

### Special Syntax

```python
# tuple unpacking
a,b,c = (1,2,3)
print(a,b,c)
```

```python
a,b = (1,2,3)
print(a,b)
```

```python
a,b,*others = (1,2,3,4)
print(a,b)
print(others)
```

```python
# Zipping a tuple

a = (1, 3, 5, 7)
b = (2, 4, 6, 8)

print(zip(a,b))

tpl = tuple(zip(a,b))
lst = (list(zip(a,b)))

print()

print(f"Zipped list: {lst}")
print(f"Zipped tuple: {tpl}")
```
