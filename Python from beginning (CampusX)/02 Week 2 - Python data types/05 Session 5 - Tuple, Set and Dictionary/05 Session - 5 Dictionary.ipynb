Dictionary in Python is a collection of keys and its values,unlike other data types which hold only a single value as an element.

For e.g: l1 = ['Nikunj', 24, 'male']

dict = { 'name' : 'nitish' , 'age' : 33 , 'gender' : 'male' }. Stored in key value pair where 'name', 'age' and 'gender' are key and 'nitish', 33, and 'male' are theirs values respectively.

Characterstics:

- Mutable
- Indexing has no meaning
- keys can't be duplicated
- keys can't be mutable items

# How to create a dictionary?

```python
# Empty dict

d = {}
print(f"Empty dict: {d}")

# 1D dict. Means dict in which there's no other dict.

d2 = {'Name': 'Nikunj', 'Age': 24, 'City': 'Jetpur'}
print(f'\n1D dict:\n{d2}')

# 2D dict    # Means dict within a dict.

d3 = {'Name' : 'Nikunj', 'Branch' : 'Data Science', 'Subjects' : {'EDAV' : 80, 'Deep Learning' : 72, 'Python' : 60}}
print(f'\n2D dict: \n{d3}')

# Create using type conversion, i.e using dict() function. To do so, we hafta pass list of lists, list of tuples, 
# tuple of lists or tuple of tuples into dict() function.

list_of_lists = [
    ['Name', "Nikunj"],
    ["Age", 24],
    ['City', 'Jetpur']
]

d4 = dict(list_of_lists)   # It will automatically create key values pairs seperated by comma.
print(f'\nDict using dict() function:\n{d4}')

# Mutable items as keys are not allowed, i.e, lists, sets, dict are mutable, so they are not allowed 
# as keys of dict. Tuples are allowed. 

# Note: You can note that mostly we are using strings as keys of dict, and see, strings are imutable.
# So mutable items as keys are not allowed in dict.
# For eg:

mut_key = {
    'Name' : 'Nikunj',
    'Age': 24,
    ['Singing', 'Reading', 'Travelling'] : 'Hobies'
    
}
# Above code will give an error bcz we are using mutable item, a list, as a key of the dict.
```

# Removing key value pair
* pop() : Deletes a specified key-value pair from the dict. (Specify only key)
* popitem() : Deletes the last key-value pair. 
* del : Deletes the specified key. (Specify only key) or we can delete the entire dict. It behaves same as pop()
* clear() : It clears the dict and make it an empty dict.

```python
d = {
    'Name' : 'Nikunj', 
    'Surname' : 'Bosamiya', 
    'Age' : 24, 
    'Hobey' : 'Singing', 
    'Fav_actor' : 'Dilip Kumar'
    }
print(d)
```

```python
# pop()

d.pop('Surname')
print(f'After pop():\n{d}')
```

```python
# popitem()

d.popitem()    # Removes key 'Fav_actor'
print(f'After popitem():\n{d}')
```

```python
# del

del d['Age']    # Removes 'Age' key-value.
print(f'After del:\n{d}')

# clear() clears the entire dict and make it an empty dict
```

# Dictionary's functions

* len() : Returns length of the dict
* sorted() : Sorts the dict into ascending order.
* update() : Used to add new key-valur pair
* items() : Returns all the key-value pairs in the form of tuple's list (List of tuples)
* keys() : Returns all the keys of the dict in list.
* values() : Returns all the values of the dict in list.

```python
dct = {
    'Name' : 'Nikunj',
    'Surname' : 'Bosamiya',
    'Age' : 23,
    'Course' : 'DataScience'
}

print(dct)
```

```python
dct2 = {
    'Name' : 'Nikunj',
    'Surname' : 'Bosamiya',
    'Age' : 23,
    'Course' : 'DataScience',
    'Marks' : {'Python' : 60, 'Stats' : 80, 'DAA' : 80, 'AI' : 45}
}

dct2
```

```python
# len()

print(len(dct))    # 4 key-value pairs

print(len(dct2))   # 5 key-value pairs
```

```python
# sorted()

print(sorted(dct, reverse = True))
print(min(dct))
print(max(dct))
```

```python
# update()

updt = {
    'College' : {'BBA' : 'Bosamiya college', 'MSc' : 'Marwadi University'}
}

dct.update(updt)

print(dct)

```

```python
a = {
    'Result' : 'Pass'
}

dct.update(a)

print()

print(dct)
```

```python
# items()

print(dct.items())
print()

# keys()

print(dct.keys())
print()

# values()

print(dct.values())
```

# Dictionary comprehension

`Problem 1` : Print first 10 numbers and their squares in the form of key - value pair.

```python
# Using zip()

''' All the elments of l1 is added with the elements of l2 at same index.
for eg: 1 + -1, 2 + -2, 3 + -3 and 4 + -4
zip(l1, l2) gives zip object so we need to convert it into list to get the pairs of elements at the same indices
'''

l1 = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
l2 = [1, 2, 3, 4, 5, 6, 7]

{i: j for (i,j) in zip(l1, l2)}
```

```python
# Using if condition

dct5 = {
    'Python' : 80,
    'Daa' : 60,
    'Stats' : 80,
    'AI' : 45,
    'Deep learning' : 72,
    'EDAV' : 69
}

{key: values for (key, values) in dct5.items() if values > 70} 
```

```python
# Nested dict comprehension. (Nested dictionary)

# Print tables of 2 to 4(including)

{i: {j: i*j for j in range(1, 11)} for i in range(2, 5)}
```
