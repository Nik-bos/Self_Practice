###`Q2:` Multiply Adjacent elements (both side) and take sum of right and lest side multiplication result.


For eg.
```
The original tuple : (1, 5, 7, 8, 10)
Resultant tuple after multiplication :

(1*5, 1*5+5*7, 7*5 + 7*8, 8*7 + 8*10, 10*8) -> (5, 40, 91, 136, 80)

output-(5, 40, 91, 136, 80)
```

````python
t = (1, 2, 3, 4, 5)
l = [t[0]*t[1]]

for i in range(1, len(t)-1):
    left_mul = t[i]*t[i-1]
    right_mul = t[i]*t[i+1]
    l.extend([left_mul+right_mul])

l.append(t[-1]*t[-2])

````

###`Q3`: Check is tuples are same or not?
Two tuples would be same if both tuples have same element at same index
```
t1 = (1,2,3,0)
t2 = (0,1,2,3)

t1 and t2 are not same
```

````python
"""
t1 = (1, 2, 3, 4)
t2 = (1, 5, 3, 4)

t1==t2
"""

t1 = (1, 2, 3, 4)
t2 = (1, 2, 3, 4)
flag = True

for i, j in zip(t1, t2):
    if i == j:
        continue
    else:
        flag = False
        break

if flag:
    print("All elements of the tuple are same and at same location")
else:
    print('Either elements are different or stored at different location.')
````

# `Q1:` Write a program to find set of common elements in three lists using sets.


**Input** : <br>
ar1 = [1, 5, 10, 20, 40, 80]    <br>
ar2 = [6, 7, 20, 80, 100]       <br>
ar3 = [3, 4, 15, 20, 30, 70, 80, 120]    <br>

### `Q2`: Replace words from Dictionary. Given String, replace it’s words from lookup dictionary.

**Example 1:**

Input:

```bash
test_str = 'CampusX best for DS students.'
repl_dict = {"best" : "is the best channel", "DS" : "Data-Science"}
```

Output:

```bash
CampusX is the best channel for Data-Science students.
```

**Example 2:**

Input:
```bash
test_str = 'CampusX best for DS students.'
repl_dict = {"good" : "is the best channel", "ds" : "Data-Science"}
```

Output:
```bash
CampusX best for DS students.
```

````python
s = "CampusX best for DS students."
d = {"best" : "is the best channel", "DS" : "Data-Science"}
result = []

for i in s.split():
    if i not in d.keys():
        result.append(i)
    else:
        result.append(d[i])

' '.join(result)
````

### `Q4`: Convert a list of Tuples into Dictionary.

**Example 1:**

Input:
```bash
[("akash", 10), ("gaurav", 12), ("anand", 14), ("suraj", 20), ("akhil", 25), ("ashish", 30)]
```

Output:
```bash
{'akash': [10], 'gaurav': [12], 'anand': [14], 'suraj': [20], 'akhil': [25], 'ashish': [30]}
```

**Example 2:**

Input:
```bash
[('A', 1), ('B', 2), ('C', 3)]
```

Output:
```bash
{'A': [1], 'B': [2], 'C': [3]}
```

````python
a = [("akash", 10), ("gaurav", 12), ("anand", 14), ("suraj", 20), ("akhil", 25), ("ashish", 30)]
d = dict()

for key, value in a:
    d[key] = [value]

print(d)
````

### `Q5`: Sort Dictionary key and values List.

**Example 1:**

Input:

```bash
{'c': [3], 'b': [12, 10], 'a': [19, 4]}
```

Output:

```bash
{'a': [4, 19], 'b': [10, 12], 'c': [3]}
```

**Example 2:**

Input:

```bash
{'c': [10, 34, 3]}
```

Output:

```bash
{'c': [3, 10, 34]}
```

````python
d = {'c': [3], 'b': [12, 10], 'a': [19, 4]}

result = {}

for i in sorted(d):
    result[i] = sorted(d[i])

print(result)
````
