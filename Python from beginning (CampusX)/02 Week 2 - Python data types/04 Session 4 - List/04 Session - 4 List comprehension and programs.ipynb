# List compression
List compression provides a concise way of creating a list.

`Program 1` : Print 1 to 10 using list compression.

```python
l = [i for i in range (1, 11)]
print(l)
```

`Program 3` : Squares of items using list compression

```python
l = [1, 2, 3, 4, 5]
sqr = [i**2 for i in l]

print(sqr)
```

`Program 4` : Print all the numbers divisible by 5 in the range of 1 to 50

```python
nums = [i for i in range(1, 51) if i % 5 == 0]

nums
```

`Program 5` : Find words which starts with letter 'N'

```python
l = ['Nikunj', 'Pizza', 'Narrow', 'Food', 'Coconut', 'Camel', 'Ninja']

a = [i for i in l if i[0] == 'N']
print(a)
```

```python
l = ['Nikunj', 'Pizza', 'Narrow', 'Food', 'Coconut', 'Camel', 'Ninja']

a = [i for i in l if i.startswith('N')]

a
```

`Program 6`: Nested if with List Comprehension

```python
basket = ['apple', 'banana', 'pineapple', 'cherry', 'orange', 'angoor']
my_fruits = ['kiwi', 'mango', 'banana', 'dragon', 'apple', 'grapes', 'angoor']

# Create a new list, and it it, insert elements of 'my_fruits' such that it shoult be present in 'basket' and also should start with 'a'.
# Without using list compression.

new_list = []

for i in my_fruits:
    if i in basket and i.startswith('a'):
        new_list.append(i)

print(new_list)
```

```python
# With list comprehenssion

basket = ['apple', 'banana', 'pineapple', 'cherry', 'orange', 'angoor']
my_fruits = ['kiwi', 'mango', 'banana', 'dragon', 'apple', 'grapes', 'angoor']

new_list = [i for i in my_fruits if i in basket if i.startswith('a')]

new_list
```

`Program 7` : Print a (3,3) matrix using list comprehension -> Nested List comprehension

```python
[[i*j for i in range(1,4)] for j in range(1,4)]
```

`Program 8` : Cartesian products -> List comprehension on 2 lists together

```python
'''Cartesian products means 1 multiply with all elements of l2, 2 with all, 
3 with all and 4 with all'''

L1 = [1,2,3,4]
L2 = [5,6,7,8]

[i*j for i in L1 for j in L2]
```

```python
l3 = [1, 2, 3, 4]
l4 = [1, 2, 3, 4]

[i*j for i in l3 for j in l4]
```

## Zip

The zip() function returns a zip object, which is an iterator of tuples where the first item in each passed iterator is paired together, and then the second item in each passed iterator are paired together.

If the passed iterators have different lengths, the iterator with the least items decides the length of the new iterator.

```python
''' All the elments of l1 is added with the elements of l2 at same index.
for eg: 1 + -1, 2 + -2, 3 + -3 and 4 + -4
zip(l1, l2) gives zip object so we need to convert it into list to get the pairs of elements at the same indices
'''




# Write a program to add items of 2 lists indexwise

L1 = [1,2,3,4]
L2 = [-1,-2,-3,-4]

print(f"Zip object: {zip(L1, L2)}")
print()

print(list(zip(L1,L2)))
print()

[i+j for i,j in zip(L1,L2)]
```

`Program 1`: Create 2 lists from a given list where 
1st list will contain all the odd numbers from the original list and
the 2nd one will contain all the even numbers 

```python
orig_l = [i for i in range (1, 11)]

even = [num for num in orig_l if num % 2 == 0]
odd = [num for num in orig_l if num % 2 != 0]

print("Even numbers: ", even)
print("Odd numbers: ", odd)
```

```python
orig_l = [i for i in range (1, 11)]

even = []
odd = []

for i in orig_l:
    if i % 2 == 0:
        even.append(i)

    else:
        odd.append(i)

print("Even numbers: ", even)
print("Odd numbers: ", odd)
```
