# List's methods

## Adding Items to a List
- append()
- extend()
- insert()

```python
# 1) append() : Adds the element at the end of the list

a = [0, 1, 2, 3, 4]

a.append(5)
print(a)
```

```python
# append(): This list's method adds the specified single element at the end of the list.
# Adding a seperate list in the old list

b = ["Simran", "Vipasha", "Kelshi", "Hima", "Dhara"]

b.append([4, 6, 3, 2, 9])
print(b)
```

```python
# 2) extend(): This list's method adds the specified more than one elements at the end of the list.

c = ['Simran', 'Vipasha', "Kelshi", "Hima", "Dhara"]
d = [4, 6, 3, 2, 9]

c.extend(d)
print(c)
```

```python
# Other ways to extend list

e = ['A', 'B', 'C', 'D']
f = ['E', "F", "G", "H"]

e+f
```

```python
# 3) insert(): This list's method adds the specified elements at the specied index's position of the list.
# a.insert(position, value)

g = ["Simran", "Vipasha", "Hima", "Dhara"]

g.insert(2, 'Kelshi')
print(g)
```

# Editing items in a List

```python
L = [1,2,3,4,5]

# editing with indexing
L[-1] = 500

# editing with slicing
L[1:4] = [200,300,400]

print(L)
```

## Deleting items from a List
- del()
- remove()
- pop()
- clear()

```python
# del
L = [1,2,3,4,5]

# indexing
del L[-1]

# slicing
del L[1:3]
print(L)
```

```python
# 4) remove(): This list's method removes the specified value from the list. It deletes based on value. 
# It removes the first value from the list that matches the specified value in the method
# a.remove(value)

h = ["Simran", "Vipasha",'Dhara', "Kelshi", "Hima",]

h.remove("Dhara")
h
```

```python
# clear
L = [1,2,3,4,5]

L.clear()

print(L)
```

```python
# 5) pop(): Bydefault, this list's method removes the last element from the list bcz index parameter is set to -1.
# We can change the value of index parameter to remove the value at specific index
# a.pop()

i = ["Simran", "Vipasha", 'Kelshi', "Hima", "Dhara"]

i.pop()
print(i)
```

```python
# 6) count(): This list's method returns the no.of times specified values appears in the list.

j = ["Simran", "Vipasha", 'Kelshi', "Hima", "Dhara", 'Simran', "Simran"]

j.count("Simran")
```

```python
# 7) reverse(): This list's method is used to reverse the entire list.

k = ["Simran", "Vipasha", 'Kelshi', "Hima", "Dhara"]

k.reverse()
k
```

```python
# 8) sort(): This list's method is used to sort the list into ascending order. To print into descending,
# set reverse parameter as True i.e reverse = True 

# sorted() : sorted does'nt make changes in original list whereas sort() does

l = ["Simran", "Vipasha", 'Kelshi', "Hima", "Dhara"]

l.sort()
l
```

```python
l.sort(reverse = True)
l
```

```python
# 9) copy(): This method of the list creates the copy of the list.

m = ["Simran", "Vipasha", 'Kelshi', "Hima", "Dhara"]

n = m.copy()
print(f"Copy of m: {n}")
```

```python
# 10)  clear(): This list's method clears the entire list but the list remains.
# del: This list's method deletes the list. Here, the list does'nt remain, it itself gets deleted.

print("Example of clear() method.")
o = ["Simran", "Vipasha", 'Kelshi', "Hima", "Dhara"]

o.clear()
print(o)
print("\n")

print("Example of del() method.")
p = ["Simran", "Vipasha", 'Kelshi', "Hima", "Dhara"]

del(p)
print(p)
```

# Palindrome's function.

```python
def palin(n):
    if type(n) == str:
        a = n.lower()
        if a == a[::-1]:
            print("It's a palindrome.")
        else:
            print("It's not a palindrome.")

    elif type(n) == int:
        orig_no = n
        rev_no = 0

        while orig_no > 0:
            digit = orig_no % 10
            rev_no = rev_no * 10 + digit
            orig_no = orig_no // 10

        if n == rev_no:
            print("This number is a palindrome.")
        else:
            print("This number isn't a palindrome.")

# Type Checking Using type(n) == str or type(n) == int
# This strict type checking is not recommended.
# Better approach: Use isinstance(n, str) and isinstance(n, int), which support subclasses too.
```

## Zip

The zip() function returns a zip object, which is an iterator of tuples where the first item in each passed iterator is paired together, and then the second item in each passed iterator are paired together.

If the passed iterators have different lengths, the iterator with the least items decides the length of the new iterator.

```python
# Write a program to add items of 2 lists indexwise

L1 = [1,2,3,4]
L2 = [-1,-2,-3,-4]

list(zip(L1,L2)) 

[i+j for i,j in zip(L1,L2)]
```
