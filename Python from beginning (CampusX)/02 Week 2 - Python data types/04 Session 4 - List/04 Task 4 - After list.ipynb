# `Problem 2:` Add new item to list after a specified item
Write a program to add item 7000 after 6000 in the following Python List
```
list1 = [10, 20, [300, 400, [5000, 6000], 500], 30, 40]
```
`Output:`
```
[10, 20, [300, 400, [5000, 6000, 7000], 500], 30, 40]
```

````python
list1 = [10, 20, [300, 400, [5000, 6000], 500], 30, 40]
print(len(list1))

for i in range(len(list1)):
    if list1[i] == 6000:
        if isinstance(list1[i + 1], list):
            list1[i + 1].insert(0, 7000)
        else:
            list1.insert((i + 1), 7000)

print(list1)
````

````python
def iterate_nested_list(nested_list):
    for element in nested_list:
        if isinstance(element, list):
            iterate_nested_list(element)
        else:
            print(element)

nested_list = [1, 2, [3, 4, [5, 6], 7], 8]
iterate_nested_list(nested_list)
````

# `Problem 3:` Update no of items available

Suppose you are given a list of candy and another list of same size representing no of items of respective candy.

i.e -  
```
candy_list = ['Jelly Belly','Kit Kat','Double Bubble','Milky Way','Three Musketeers']
no_of_items = [10,20,34,74,32]
```
`Write a program to show no. of items of each candy type.
`

`Output:`
Jelly Belly-10
Kit Kat-20
Double Bubble-34
Milky Way-74
Three Musketeers-32

````python
# Simple way.

candy_list = ['Jelly Belly', 'Kit Kat', 'Double Bubble', 'Milky Way', 'Three Musketeers']
no_of_items = [10, 20, 34, 74, 32]

for i in range(len(candy_list)):
    print(f"{candy_list[i]} - {no_of_items[i]}")
````

````python
# using zip()

candy_list = ['Jelly Belly', 'Kit Kat', 'Double Bubble', 'Milky Way', 'Three Musketeers']
no_of_items = [10, 20, 34, 74, 32]

for candy, num_items in zip(candy_list, no_of_items):
    print(f'{candy} - {num_items}')
````

# `Problem 5:` You are given a list of integers. You are asked to make a list by running through elements of the list by adding all elements greater and itself.

i.e. Say given list is `[2,4,6,10,1]`
resultant list will be `[22,20,10,23]`.

For 1st element `2` ->> these are greater `(4+6+10)` values and `2` itself so on adding becomes `22`.

For 2nd element `4` ->> greater elements are `(6, 10)` and `4` itself, so on adding `20`

like wise for all other elememts.

`[2,4,6,10,1]`-->`[22,20,16,10,23]`




````python
l = [2, 4, 6, 10, 1]
sum = 0
l2 = []

for i in l:
    for j in l:
        if i <= j:
            sum = sum + j
    l2.append(sum)
    sum = 0

print(l2)
````

### `Problem 10:` Add Space between Potential Words.

**Example:**

Input:

```bash
['campusxIs', 'bestFor', 'dataScientist']
```

Output:
```bash
['campusx Is', 'best For', 'data Scientist']
```

````python
input_list = ['campusxIs', 'bestFor', 'dataScientist']
output_list = []

for string in input_list:
    words = ''
    for char in string:
        if char.isupper():
            words += ' ' + char
        else:
            words += char
    output_list.append(words.strip())

print(output_list)
````

# `Problem 12:` Write a program that can find the max number of each row of a matrix

**Example:**

Input:

```bash
[[1,2,3],[4,5,6],[7,8,9]]
```

Output:
```bash
[3,6,9]
```

````python
l = [
    [3, 2, 1],
    [5, 6, 4],
    [7, 8, 9]
]

greater = []

for nst_lst in l:
    a = nst_lst[0]
    b = nst_lst[1]
    c = nst_lst[2]
    
    
    if a >= b and a >= c:
        greater.append(a)
    elif b >= c:
        greater.append(b)
    else:
        greater.append(c)
        

print(greater)
````
