# 4. `drop_duplicates()`: This will remove all the duplicate values.

```python
temp = pd.Series([1, 2, 3, 4, 1, 2, 5, 4, 7, 8, 4, 12, 15, 14, 16, 14])
temp
```

```python
temp[temp.duplicated()]    # duplicated() return True if a value is repeating else False

# See, 1 2 and 4 and 14 are repeating

```

```python
# Drop

temp.drop_duplicates()

# Bydefault, this keep the first occurrence of duplicate values
# There's a parameter 'keep' which we can set to last for keeping the last value and deleting all the earlier values
```

# 5. `sum()/ count()`: 
`sum()` adds the digits present and <br>
`count()` returns the total non_null values <br>

`sum()` considers True as 1 and False as 0

```python
temp2 = pd.Series([1, 2, 3, np.nan, 4, np.nan, 5])

temp2[temp2.isnull()]
```

```python
temp2.isnull().sum()
```

```python
temp2.sum()
```

```python
# Sum considers True as 1 and False as 0

temp2 = pd.Series([True, True, False, True, False])

temp2.sum()    # 1+1+0+1+0
```

```python
temp2.count()    # Returns total non_null values. Here 0 non_null values. So returns 
```

# 8. `isin()`: Returns True if value is in the passed list, else False

```python
# Supose we want to find matches where Kohli made 49 and 99 runs.

kohli[(kohli == 49) |
(kohli == 99)
]

# Here it was only two cases 49 and 99
# But what if i wanna find matches where Kohli made 48, 49, 98 and 99 runs. I'll have to make multiple conditions.
# Here comes isin(). Just pass a list containing the desired values.
```

```python
# isin()

kohli[kohli.isin([48, 49, 98, 99])]
```

# 10 `copy()`: This copies the data.

Let's first see the difference between **view** and **copy**

```python
kohli.tail()
```

```python
# Supose someone tells us taht kohli has not made 7 runs in his last match. He made 10 runs. In that case lat's change it

last5 = kohli.tail()
```

```python
last5
```

```python
last5[215] = 10

last5
```

```python
kohli.tail()

# See this made changes to the original data as well.
# It is just giving us the 'view' of the original data.
# Here comes the copy() which actually copies the data
```

```python
# Let's again change the value of last match to 7
```

```python
kohli.tail().copy()
```

```python
new = kohli.tail().copy()
```

```python
new[215] = 7
```

```python
new
```

```python
kohli[215]

# See, it remained 10 even after changing
```
