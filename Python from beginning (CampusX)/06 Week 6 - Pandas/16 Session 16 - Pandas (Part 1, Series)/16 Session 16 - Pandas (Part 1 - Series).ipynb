# `What is Pandas in Python?` **Pandas** is a tool inside Python which is used to analyse and maipulate the data. There're two objects in Pandas, they're<br>
1. `Pandas Series`: **Series** is a column of a table which can hold the data of any type.
2. `Pandas Dataframe`: **Dataframe** is a tabular representation of the data with rows and columns

**Here in this session, we'll learn `Pandas Series`. So let's start.**

```python
import pandas as pd
import numpy as np

# Pandas is built on the top of numpy, so it is a good idea to import numpy along with pandas bcz we will be using it.
```

# There're different way to form pandas series. Let's explore them one by one.

# 1) Series from list. 

```python
country = ['India', 'Japan', 'Brazil', 'Australia', 'Singapore', 'Russia']

pd.Series(country)

# Series will always have 2 things, one is 'value'(India, Japan, Brazil...) and other is the index(Bydefault numbers).
# Other thing we can see is 'dtype', which is 'object' here, it shows the datatyoe of the values. 
# By default, pandas call strings as object. Not always, but 99% of times it does.
```

```python
# List of integers

runs = [53, 42, 25, 78, 56, 54]
index = ['Sachin', 'Rahul', 'Dhoni', 'Virat', 'Yuvraj', 'Gautam']

runs = pd.Series(data = runs, index = index, name = "Runs scored by each players")
runs

# See, we can also give our own index and 'name' parameter gives name to the series.
# See tge dtype here, it's int64. We can change it using 'dtype' parameter
```

# 2) Series from dictionary.

```python
dict = {
    'Sachin': 53,
    'Rahul': 42,
    'Dhoni': 25, 
    'Virat': 78, 
    'Yuvraj': 56, 
    'Gautam': 54
}

runs = pd.Series(dict, name = 'Runs scored by each player')
runs

# See, 'keys' became indices and 'values' became value of the Series
```

# Series's attributes

```python
# size:  Returns the total no_of items in a series including null values.
# Where count() excludes null values.

print(runs.size)     
```

```python
# dtype: Returns the data type of the series

runs.dtype
```

```python
# name: Returns the name of the series

runs.name
```

```python
# is_uniqueL Returns 'True' if all the values are unique else returns 'False'

runs.is_unique
```

```python
# index: Returns the indices of the Series.

runs.index
```

```python
# values: Returns all the values of the series

runs.values
```
