# What is group_by? Groupby groups the data and then used to study the groups.

`groupby` is always performed on Categorical column. See, what is the use if we apply on numeric column. In numeric column the data will be continuous ar descrete, there's no sense to create groups of numbers.

When we create a groupby function, it creates a groupby object and then we can apply other functions to this object.

`It's mainly used with aggregation function.`

```python
import pandas as pd
import numpy as np
```

# Here, we'll learn on `imdb` dataset and will practise on cricket's dataset.

`IMDB dataset is the records of top 1000 movies`

```python
imdb = pd.read_csv("00 imdb_dataset.csv")

imdb.head()
```

```python
imdb['Genre'].unique()
```

```python
genres = imdb.groupby(by = "Genre")    # This will create groups for each category of Genre. This will create a groupby object

genres
```

```python
genres.sum()

# Index is Genre and sum() method will sum all the values of the numeric columns
```

```python
imdb[imdb['Genre'] == 'Action']['Runtime'].sum()

# See, we're getting the same result as above
```

```python
genres.min()
```

```python
genres.mean()
```

# Find out top 3 genres by total earnings

```python
imdb.groupby(by = 'Genre').sum()    ['Gross'].sort_values(ascending = False).head(3)
```

```python
# More effecient way. 
# Always perform this method bcz this requires less computation to get the top 3 genres on earnig basis

imdb.groupby(by = 'Genre')  ['Gross'].sum().head(3)
```

# Find the genre with highest average imdb ratings.

```python
imdb.groupby('Genre')  ['IMDB_Rating'].mean().sort_values(ascending = False).head(1)
```

```python
imdb.groupby('Genre')  ['IMDB_Rating'].mean().sort_values(ascending = False).idxmax()

# idxmax() only returns the index with highest value, not the value.
```

# Find top 5 most popular directors(Consider no.of votes for popularity).

```python
imdb.groupby('Director')  ['No_of_Votes'].sum().sort_values(ascending = False).head()
```

# Find no_of_movies done by each actor

```python
imdb.groupby('Star1')    ['Series_Title'].count().sort_values(ascending = False)
```

```python
# Using value counts

imdb['Star1'].value_counts()
```

# `groupby` attributes and methods.

* find total number of groups -> len()
* find items in each group -> size
* first()/last() -> nth item
* get_group() -> vs filtering
* groups
* describe()
* sample()
* nunique()

# 1) Find total no_of groups--> `len()`

```python
len(imdb.groupby('Genre'))
```

```python
imdb.Genre.nunique()
```

# 2) Find items/rows in each group--> size

```python
imdb.groupby('Genre').size()
```

```python
imdb.Genre.value_counts()
```

# 3) Find 1st and last item/ row of each group--> first()/ last()/ nth()

```python
imdb.groupby('Genre').first()
```

```python
# last()

imdb.groupby('Genre').last()
```

```python
# Find 5th movie of each group

imdb.groupby('Genre').nth(4)
```

# 4) `get_group()`--> Get records of particular group

```python
group = imdb.groupby('Genre')

group.get_group('Adventure')
```

```python
# Using filter method

imdb[imdb['Genre'] == 'Adventure']
```

# 5) groups--> This will return a key-value pairs where keys are groups' names and values are the list of dictionary containing those value.

`For eg: {'Action': [2, 5, 8, 10, 13, 14, 16, 29, 30, 31, 39, 42, 44} means,` <br>
`Index 2, 5, 8, 10, 13 etc are Action movies.`

```python
group.groups
```

# 6) `describe()`--> Gives statistical info of the group. Same as series and DataFrame

```python
group.describe()
```

```python
group.describe()  ['IMDB_Rating']
```

# 7) `sample()`--> This will return specified no_of random samples of each group

```python
group.sample(2, replace = True)

# Replace = True bcz some groups are only having one record. If False, will throw an error mgs.
```

# 8) nunique()--> Returns the no_of unique values of each group.

```python
group.nunique()
```
