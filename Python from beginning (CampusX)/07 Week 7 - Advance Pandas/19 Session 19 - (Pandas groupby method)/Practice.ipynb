# Always follow below 3 steps
`90% Problems will get solved with below 3 steps.`
1. apply groupby()
2. Get your desired column
3. Perform Aggregate operations


# `groupby` attributes and methods.

* find total number of groups -> len()
* find items in each group -> size
* first()/last() -> nth item
* get_group() -> vs filtering
* groups
* describe()
* sample()
* nunique()

```python
# Find total no of groups formed: len()

len(imdb.groupby("Genre"))
# imdb['Genre'].nunique()
```

```python
# Find items in each group: size

# imdb.groupby('Genre') ['Series_Title'].count().sort_values(ascending=False)
# imdb['Genre'].value_counts()
# imdb.groupby('Genre').size()
```

```python
genre = imdb.groupby("Genre")
```

```python
# first()/last(): Retruns the first and last movie of each Genre, respectively
# first(): Enters into the each group and returns the first movie
# last():  Enters into the each group and returns the last movie
# nth():  Enters into the each group and returns the nth movie

# genre.first()
# genre.last()
# genre.nth(9)    # Will return the 10th movie of each group
genre.nth(9).set_index("Genre")    # Will return the 10th movie of each group
```

```python
# get_group(): We can get all the rows of a particular group

genre.get_group("Action")

# We can also do this by filtering
# imdb[imdb['Genre'] == 'Action']

# get_group is falst bcz it has already created gropus and the filtering will perform filter every time.
```

```python
# groups(): This returns a dictionary where keys are the names of the groups created,
# and values are list of indices or each group
# Rarely used.

genres.groups
```

```python
# describe: Returns statistics of each group

genre.describe()
```

```python
# sample(): Returns one movie from each group

# len(genre)
# genre.sample()
# genre.sample(5)
genre.sample(2, replace = True)
```

```python
# nunique(): Returns total no of unique rows present in each group

genre.nunique()
```

# Some other functions

```python
# agg(), pasing dictionary: Till now, we were applying single aggregare function like min(), max(), sum() etc
# But if we want, we can also apply multiple aggregate functions using agg() method.

genre.agg(
    {
        'Runtime': 'mean',
        'IMDB_Rating': 'mean',
        'No_of_Votes': 'sum',
        'Gross': 'sum',
        'Metascore': 'min'
    }
)

# genre.get_group("Action")['Metascore'].min()
```

```python
# agg(): Find min, max and average of each column

genre.agg(
    {
        'Runtime': ['min', 'max', 'mean'],
        'Gross': ['min', 'max', 'mean'],
        'No_of_Votes': ['min',  'max', 'mean']
    }
)
```

```python
# looping on the group 

for group_object, data in genre:
    print(data['IMDB_Rating'])
```

```python
# apply: We can apply built-in functions but most importanr thing for apply function is, we can apply
# custom function as well.

# Find total no of movies starting with 'A' in each genre/ group
def count_a(group):
    print(group)
    return group

genre.apply(count_a)
```

```python
# continue

def count_a(group):
    return group['Series_Title'].str.startswith('A').sum()
```

```python
genre.apply(count_a)
```

```python
# Give rank to each movie in the group according to IMDB_Rating

def rank_movie(group):
    group["rank"] = group['IMDB_Rating'].rank(ascending = False)
    return group
    


genre.apply(rank_movie)
```

```python
# Find normalized imdb rating groupwise

def normalize(group):
    group['Norm_Rating'] = (group['IMDB_Rating'] - group['IMDB_Rating'].min()) / (group['IMDB_Rating'].max() - group['IMDB_Rating'].min())

    return group


genre.apply(normalize)
```

# Grouping on multiple columns

```python
duo = imdb.groupby(by= ['Director', 'Star1'])

#size()
duo.size()

# get_group()

duo.get_group(
    (
        "Aamir Khan", "Amole Gupte"
    )
)
```

```python
# Find the most earning actor director combo

duo['Gross'].sum().sort_values(ascending= False).head()
```

```python
# Find the best actor-genre combo in terms of average meta score

actor_genre = imdb.groupby(by = ['Star1', 'Genre']) ['Metascore'].mean().sort_values(ascending = False)
# print(actor_genre)
# print("="* (50))

# print(actor_genre.index)    # See, this is a multi-index series

# Below code will return a Multi-index series and series doesn't have columns to choose for sort_values()
# Sp series's sort_values does not have 'by' parameter. IT sorts only on values

# reset_index() will convert series to DataFrame
# imdb.groupby(by = ['Star1', 'Genre']) ['Metascore'].mean().sort_values(ascending = False).reset_index()

imdb.groupby(by = ['Star1', 'Genre']) ['Metascore'].mean().reset_index().sort_values(by = "Metascore", ascending = False).head()

# reset_index() converts the multi-index series to the DataFrame
```
