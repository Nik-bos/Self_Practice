```python
# Find top 10 Batsman in term of runs

df.groupby("batsman") ['batsman_runs'].sum().sort_values(ascending = False).head(5)
```

```python
# Find top 5 batsmax with maximum no of sixes

# See, here we dont need any rows where batsman did nit hit six.
# Our concern is only woth those rows where batsman hit six.
# So, let's first filter those rows

sixes_df = df[(df['batsman_runs'] == 6)]
# sixes

sixes_df.groupby("batsman") ['batsman'].count().sort_values(ascending = False).head()
```

```python
# Find batsman with most boundries in death overs

# Same solution as above.
# We are only concerned with boundries in death overs so lets filter rows of 16-20 over so filyer accordingly.
# Then apply groupby on batsman.
# take any one column and count no of rows using count()


death_overs_boundries = df[
    (df['over'].between(16,20)) &
    (df['batsman_runs'].isin([4, 6]))
]

# death_overs_boundries

death_overs_boundries.groupby('batsman') ['batsman'].count().sort_values(ascending = False).head()
```

```python
# Find V.Kholi's total runs against all the teams

# We are concerned with only V.Kholi. So lets first filter his matches

v_kohli = df[df['batsman'] == 'V Kohli']

# v_kohli

v_kohli.groupby('bowling_team') ['batsman_runs'].sum().sort_values(ascending = False).reset_index()
```

```python
# Create a function which returns the highest score of any batsman in a single match

def get_highest_score(df, batsman):

    filter_matches = df[df['batsman'] == batsman]
    
    highest_score = filter_matches.groupby('match_id') ['batsman_runs'].sum().sort_values(ascending = False).head(1).values[0]

    return highest_score
```

```python
get_highest_score(df, "CH Gayle")
```

```python
def get_highest_score(df, batsman):

    filter_matches = df[df['batsman'] == batsman]

    highest_score = (
        filter_matches
        .groupby('match_id')['batsman_runs']
        .sum()
        .max()
    )

    return highest_score
```

```python
get_highest_score(df, "CH Gayle")
```
