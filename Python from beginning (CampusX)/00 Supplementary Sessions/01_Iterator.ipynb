# What is an Iteration?
Iteration is a general term for taking each item of something, one after another. Any time we use a loop to go over a group of items, that is iteration.

```python
# Example
num = [1, 2, 3, 4]

for i in num:
    print(i)
```

# What is an iterator?
- An `iterator` is an object the allows us to iterate through sequence of the data without having to store the entire data in the memory.
- An `iterator` is the thing because of which iteration is performed.
- When we run a loop, `iterator` gets created internally which is responsible to run the loop.
- Iterator fetches an item one by one and gives to the loop.

- An iterator stores a data in the memory, performs operation and deletes it.
- Then stores the next data in the memory, performs operation and deletes it.
- Like this, an `iterator` doesn't store while data in the memory.

**See the below code and notice the size of a simple list and an `iterator`**


```python
import sys

l = [x for x in range(100000)]

print(f"Size of l: {sys.getsizeof(l)/1024}")

x = range(100000)
print(f"Size of x: {sys.getsizeof(x)/1024}")
```

# What is an iterable?

- An `iterable` is an object one can iterate over.
- Means if we are able to run a loop over any object, that object is an `iterable`
For example:
    * We can run a loop over list, range() so list and range() object is an iterator.


- An iterable generates an iterator when passed to an `iter()` method.

`See the code below`

```python
l = [1, 2, 3, 4, 5]

print(type(l))
# l-> Is in iterable

print(type(iter(l)))    # list_iterator
```

# Points to remember.

- Not all iterable is an iterator
- All iterators are iterable.

For example: <br>
    - `List` is an `iterable` because we can loop over it, but it is `not an iterator` beacuse list stores every item together but `iterable doesn't store every item together`.
    
    - So, not every iterable is an iterator.

    - We can loop over an iterator objects, so all iterators are iterables

# Trick.

- Every iterable has an `iter()` function.
- Every iterator has an `iter()` and `next()` function both

```python
l = [1, 2, 3, 4]

dir(l)    # See, it has an __iter__() method, so list is an iterable.
```

```python
dir(iter(l))    # See, this has __iter__() and __next__() method both. So, this is an iterator.
```

# Understand for `for loop` works.

```python
# Define an iterable.
l = [1, 2, 3, 4]

# Make it an iterator.
iter_l = iter(l)

# Call next function.
print(next(iter_l))    # Prints 1 and deletes it
print(next(iter_l))    # Prints 2 and deletes it
print(next(iter_l))    # Prints 3 and deletes it
print(next(iter_l))    # Prints 4 and delets it

print(next(iter_l))    # Throws error: StopIteration.
```

```python
# Making our own for loop.

def mera_for_loop(iterable):

    # Making iterable an iterator
    iterator = iter(iterable)

    # Running loop
    while True:
        try:
            print(next(iterator))
        except StopIteration:
            break
```

```python
a = [1, 2, 3, 4]
b = range(1, 11)
c = (5, 6, 7, 8)
d = {'Nikunj': 26, "Jugal": 16}
e = {9, 10, 11, 12}

mera_for_loop(e)
```

# A confusing point.

```python
l = [1, 2, 3, 4]

iterator_obj_1 = iter(l)
print(f"ID of object 1: {id(iterator_obj_1)}")

iterator_obj_2 = iter(iterator_obj_1)
print(f"ID of object 2: {id(iterator_obj_2)}")

# See, id of both iterators is same.
# This simple means when we create an iterator from iter we get an iterator_1
# But again if we create an iterator using the iterator_1, it returs itself i.e iterator_1
```
