# Lambda Functions

- Lambda functions are also known as `Anonymous Functions`.
- These are written in one line.
- We will learn when and why we use lambda function over and above normal function

**Syntax**: lambda(input: expression)

```python
x = lambda x: x**2
x(8)
```

```python
# Sum of two numbers
x = lambda a, b: a+b

x(5, 7)
```

```python
# Lambda returns the whole function unlike normal function that returns the int, float, string or some other object
x = lambda a, b: a+b
print(x)
```

# Difference between normal and lambda function

- Lambda has no return function
- Rather than returning a value, lambda function returns the whole function
- Lambda function can be written only in one line
- Unline normal functions, lambda function is not used for code reusability. We define lambda function to use only once.(Use and throw)
-  Lambda function is anonymous, means it has no name.

# When do we use lambda function?

- We don't use lambda function in normal sense.
- We use it along with higher order function
- Higher Order Functions are those functions, which takes another function as an argument or which returns a function.

`Before we understand why do we need higher order functions, let's first define some lambda functions so we understant it properly.`

```python

a = lambda s: s[0].lower() == 'a'

print(a('Nikunj'))
print(a("Archie"))
```

```python
# Check if given number is odd or even

oe = lambda x: 'Even' if x%2 == 0 else 'Odd'

oe(2)
```
