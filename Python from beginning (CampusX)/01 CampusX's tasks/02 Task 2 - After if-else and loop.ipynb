`Problem 1`: Write a program that will give you in hand monthly salary after deduction on CTC - HRA(10%), DA(5%), PF(3%) and taxes deduction as below:
* Below 5L : 0%
* 5L-10L : 10%
* 10L-20L : 20%
* above 20L : 30%

```python
n = int(input("Enter your salary: "))

ctc = (n * 10/100) + (n * 5/100) + (n * 3/100)
b4_tax = n - ctc
print("\nSalary before tax:", b4_tax,'\n')

if b4_tax < 500000:
    print(f"Salary after tax(0%): {b4_tax}")

elif 500000 <= b4_tax < 1000000:
    print(f"Salary after tax(10%): {b4_tax - (b4_tax * 10/100)}")

elif 1000000 <= b4_tax < 2000000:
    print(f"Salafy after tax(20%): {b4_tax - (b4_tax * 20/100)}")
    
else:
    print(f"Salary after tax(30%): {b4_tax - (b4_tax * 20/100)}")
```

`Problem 3`: Write a program that will take user input of cost price and selling price and determines whether its a loss or a profit.

```python
cp = int(input("Enter cost price: "))
sp = int(input("Enter selling price: "))

if sp > cp:
    print(f"\nIt's a profit of {sp - cp}.")

elif sp < cp:
    print(f"\nIt's a loss of {cp - sp}.")

else:
    print("\nNo profit, no loss.")
```

`Problem 5` - Display Fibonacci series up to 10 terms.
Note: The Fibonacci Sequence is a series of numbers. The next number is found by adding up the two preceeding numbers. The first two numbers are 0 and 1. For example, 0, 1, 1, 2, 3, 5, 8, 13, 21. The next number in this series above is 13+21 = 34

```python
a = 0
b = 1
n = int(input("Enter no.of terms for fibonacci series: "))
fib = [0,1]

if n == 1:
    print(a)
elif n == 2:
    print(a,b,sep = ', ')

else:
    for i in range(1, n-1):
        c = b + a
        fib.append(c)
        a,b = b,c
print(fib)
```

`Problem 6` - Find the factorial of a given number.

```python
n = int(input("Enter a number to get its factorial: "))

fact = 1
for i in range(1, n + 1):
    fact = fact * i

print(f"\nThe factorial of {n} is {fact}.")
```

`Problem 7` - Reverse a given integer number.

```python
n = int(input("Enter a number to get its reverse: "))
temp = n

while temp != 0:
    ln = temp % 10
    print(ln, sep = ', ', end = '')
    temp = temp // 10
```

```python
# Using list

n = int(input("Enter a number to get its reverse: "))
temp = n
reverse = []

while temp != 0:
    last_no = temp % 10
    reverse.append(last_no)
    temp = temp // 10

print(reverse)
```

#`Problem 8`: Take a user input as integer N. Find out the sum from 1 to N. If any number if divisible by 5, then skip that number. And if the sum is greater than 300, don't need to calculate the sum further more. Print the final result. And don't use **for** loop to solve this problem.

```python
n = int(input("Enter a number: "))
sum = 0
i = 1
while i <= n:
        
    if i % 5 == 0:
        i += 1
        continue

    else:
        if sum + i <= 300:
            sum = sum + i
            i += 1
        else:
            break
print(sum)
print(f'Current value of i is {i}.')
```

# Working of loops

* digits = [int(digit) for digit in str(i)]

str(i): Converts the current number i to a string. For example, if i is 2468, str(i) becomes the string '2468'.

for digit in str(i): Iterates through each character in the string, treating each character as a digit. For '2468', it 
iterates through '2', '4', '6', and '8'.

int(digit): Converts each character back to an integer. So, it converts '2' to 2, '4' to 4, and so on.

The list comprehension [int(digit) for digit in str(i)] creates a list containing the individual digits of the number. For 2468, it creates the list [2, 4, 6, 8].

* if all(digit % 2 == 0 for digit in digits):
for digit in digits: Iterates through each digit in the list created in the previous step. For the list [2, 4, 6, 8], it 

iterates through 2, 4, 6, and 8.

digit % 2 == 0: Checks if each digit is even. If the remainder when dividing the digit by 2 is 0, it means the digit is even.

The all() function is used to check if all the elements in the iterable (digits) satisfy the condition digit % 2 == 0. If all digits are even, the condition is True.

# `Problem 13`:Print all the Armstrong numbers in a given range.
Range will be provided by the user<br>
Armstrong number is a number that is equal to the sum of cubes of its digits. For example 0, 1, 153, 370, 371 and 407 are the Armstrong numbers.

```python
a = 153

temp = a
sum = 0
while temp > 0:
    rem = temp%10
    sum =sum +  rem**3
    temp = temp//10

if sum == a:
    print(f'{a} is an Armstrong number.')

else:
    print(f'{a} is not an Armstrong number.')
```
