`Program 1:` Find the length of a given string without using the len() function


```python
n = input("Enter a string : ")

length = 0

for i in n:
    length += 1

print(length)
```

`Program 2:` Extract username from a given email. 
Eg: If the email is nitish24singh@gmail.com then the username should be nitish24singh

```python
email = input("enter your email_id : ")

pos = email.index('@')

u_name = email[0 : pos]
print(f'\nUsername is {u_name}')
```

```python
email = input("Enter an email: ")

username = email.split('@')[0]

print(f'Username is {username}')
```

`Program 3` Count the frequency of a particular character in a provided string. 
 E :g 'hello how are you' is the string, the frequency of h in this string is 2.

```python
n = input("Enter a string: ")
term = input("Enter a term to get its frequency: ")

freq = 0

for i in n:
    if i == term:
        freq += 1        

print(f'\nFrequency of {term} in the given string is {freq}')
```

`Program 4:` Write a program which can remove a particular character from a string.

```python
n = input("Enter a string: ")
term = input("Enter a character or letter to remove: ")

result = ""
for i in n:
    if i != term:
        result = result + i

print(result)
```

```python
# Write a program that can convert an integer to string.

number = int(input('enter the number'))

digits = '0123456789'
result = ''
while number != 0:
  result = digits[number % 10] + result
  number = number//10

print(result)
print(type(result))
```
