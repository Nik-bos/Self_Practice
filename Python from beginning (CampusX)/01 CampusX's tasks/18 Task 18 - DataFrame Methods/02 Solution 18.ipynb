# `Question:` IPL Point Table

Make a function `point_table` which take `season` as parameter and show points table in descendng order of points and in ascending order of team name.

For winning - 2 Points<br>
For loosing - 0 Point<br>
For not result both team gets 1 points.    `Here there are no matches with no-result`

Make dataframe which will have
`TeamName`, `MatchesPlayed`, `MatchesWon`, `Points`
make `TeamName` as index.

```
season parametr should be one of these->
['2022', '2021', '2020/21', '2019', '2018', '2017', '2016', '2015', '2014', '2013', '2012', '2011', '2009/10', '2009', '2007/08']
```



# Output of two Top 2 in season 2022
```
TeamName    MatchesPlayed	MatchesWon	NoResult	Points

Gujarat Titans	    16	   12	       0	     24
Rajasthan Royals	  17	   10	       0	     20

```

# `Remember: Season column is string.`

````python
def point_table(ipl, season):
    
    subset = ipl[ipl['Season'] == str(season)]    # This code will fetch the records of passed season

    matches_won = subset['WinningTeam'].value_counts()    # This code will return matches won by each team

############################################################################

    # Names of the first two teams
    
    team_1 = matches_won.index[0]    # Name of 1st team with maximum winnings
    team_2 = matches_won.index[1]    # Name of 2nd team with maximum winnings

############################################################################

    # Total matches won by First 2 teams

    team_1st_total_winning = matches_won[0]
    team_2nd_total_winning = matches_won[1]

############################################################################

    # Total points of first 2 teams
    
    team_1st_points = subset[subset['WinningTeam'] == team_1].shape[0]*2   # Returns the no_of points by team 1
    team_2nd_points = subset[subset['WinningTeam'] == team_2].shape[0]*2   # Returns the no_of points by team 2
    
############################################################################

    # Total no_of_matches played by first two teams
    
    total_matches_team_1 = subset[    # Total no_of_matches played by 1st team
    (subset['Team1'] == team_1) |
    (subset['Team2'] == team_1)
    ].shape[0]

    total_matches_team_2 = subset[    # Total no_of_matches played by 1st team
    (subset['Team1'] == team_2) |
    (subset['Team2'] == team_2)
    ].shape[0]

############################################################################
    
    # Creating the final dataframe

    data = {'Team': [team_1, team_2],
           'Total Matches Played': [total_matches_team_1, total_matches_team_2],
            'Winnings': [team_1st_total_winning, team_2nd_total_winning],
            'Points': [team_1st_points, team_2nd_points],
            'Season': season
           }
    
    df = pd.DataFrame(data)
    
    return df.set_index('Team')   
````
