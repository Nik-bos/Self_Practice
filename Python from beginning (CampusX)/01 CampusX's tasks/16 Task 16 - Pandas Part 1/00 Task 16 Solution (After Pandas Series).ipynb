# Q-2: Write a Pandas program to add, subtract, multiple and divide two Pandas Series.

```python
class Math:
    def __init__(self, s1, s2):
        self.s1 = s1
        self.s2 = s2

    def add(self):
        return self.s1 + self.s2

    def sub(self):
        return self.s1 - self.s2

    def multiply(self):
        return self.s1 * self.s2

    def divide(self):
        return self.s1 / self.s2
```

```python
s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([1, 2, 3, 4, 5])

s1/s2
```

```python
m = Math(s1, s2)

print((m.add()))
print()

print(m.sub())
print()

print(m.multiply())
print()

print(m.divide())
```

1. Read items.csv making item_name as index.

```python
# 1) Read items.csv making item_name as index.

items = pd.read_csv(r'02 Task 16 (Items dataset).csv', index_col = 'item_name').squeeze()

type(items)
```

```python
items.head()
```

```python
# 2) Show no of nan values

items.isnull().sum()
```

# 3) Item price is given in $, so convert it to rupees without currency symbol.

```python
items.dtypes
```

```python
# Let's first remove $ symbole and convert dtype to currency.

items.replace('$',  '')
```

```python
items = items.str.replace('$', "").astype('float')    # Replacing $ with "" and converting to 'float'
```

```python
items
```

```python
items_inr = items* 85
```

```python
items_inr
```

```python
# 3) Make data type of newly made series as float32

items_inr = items_inr.astype('float32')

items_inr
```

```python
# 4) Fill nan with mean of the series.

items_inr.isnull().sum()
```

```python
mean1 = items_inr.mean()

mean2 = np.mean(items_inr)
```

```python
mean1
```

```python
mean2
```

```python
# Filling

items_inr = items_inr.fillna(mean1)

items_inr.isnull().sum()
```
