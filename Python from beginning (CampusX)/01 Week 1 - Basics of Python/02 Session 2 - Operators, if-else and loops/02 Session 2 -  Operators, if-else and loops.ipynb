# Program

```python
# Get the sum of 3 digits entered by the user. (Can use for and while loop both)

num = int(input("Enter a three digit num: "))

temp = num
sum = 0

for i in range(1,4):
    rem = temp % 10
    temp = temp // 10
    sum = sum + rem

print(sum)
```

```python
# Get the sum of all the digits of the given number.

num = int(input("Enter a number: "))

temp = num
sum = 0

while temp != 0:
    rem = temp % 10
    temp = temp // 10
    sum = sum + rem
print(sum)
```

```python
# Program to print the table of the given number

num = int(input("Enter the number: "))
i = 1

while i < 11:
    print(f'{num} * {i} =', num * i)
    i = i + 1
```

```python
# While loop with else
# In python, we can also use else with loops. After executing the whole loop, it will execute the else part

x = 1

while x <= 5:
    print(x)
    x += 1
else:
    print("Limit exceeded")
```

```python
#Game: Guess the jackpot number.

import random

jackpot = random.randint(1, 101)
print(jackpot)

guess = int(input("Guess the jackpot number: "))
no_of_attempts = 1

while guess != jackpot:
    if guess < jackpot:
        print("Guess the higher number.")
    else:
        print("Guess the lower number.")

    print('\n')

    guess = int(input("Guess the jackpot number: "))
    no_of_attempts += 1
    

print("\nCongratzzzz! You won a jackpot.")
print(f"Total attemts = {no_of_attempts}.")
        
```

# Loop Control Statements

- Break
- Continue
- Pass

```python
# Break: It terminates the further iterations when condition is met.
# Real time example where we would use break: Searching in the database

# Simple example for break.

for i in range(1, 11):
    if i == 6:
        break
    else:
        print(i)
```

```python
# Example: Get a range from the user and find all the prime numbers betweem that range

lower = int(input("Enter lower nuumber: "))
upper = int(input("Enter upper number: "))

for i in range(lower, upper+1):
    for j in range(2, i):
        if i % j == 0:
            break
        
    else:
        print(i)

# If did'nt get, refer Loops' topic's note informing 'using else with loop'.
```

```python
# continue: Where break terminates the further iterations upon condition met, continue does'nt terminate but 
# skips the particular iteration for which condition is met

# Simple example for continue.

for i in range(1, 11):
    if i == 6:
        continue
    else:
        print(i)

# We will see in the output that print(6) is skipped and 6 is'nt printed.
# Real life's example where continue is used: 
# Displaying in ecommerce website where we want to skip those products whose stock is 0
```

```python
# pass: It is used to ignore error when we do;nt wanna write any code inside loop. If we dont use pass,
# python will give and error for not completing the code.

for i in range(10):
    pass
```
