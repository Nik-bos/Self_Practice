# In this session, we'll learn 5 types of graphs.

3. Bar chart

* It is very important to know the type of data bcz then only we can decide which graphs we should plot.
* If we plot a single column, it is called Univariate Analysis.

```python
# Plotting a simple 2-d plot on numeric-categorical data
price = [50000, 54000, 59000, 63000, 60000, 55000]
year = [2015, 2016, 2017, 2018, 2019, 2020]

plt.plot(year, price, marker = '*')

# We can adjust marker size using 'markersize' parameter
```

```python
# Using pandas dataframe

batsman = pd.read_csv(r"Datasets/rohit-kohli.csv")
batsman
```

```python
# plotting V kholi's runs

plt.plot(batsman['index'], batsman['V Kohli'], marker = "*")
```

```python
# Plotting multiple plots in a single graph

plt.plot(batsman['index'], batsman['V Kohli'], marker = "*")
plt.plot(batsman['index'], batsman['RG Sharma'], marker = "*")

# Blue is Kohli's and 
# Orange is Sharma's
# We can plots as many plots we want in a single graph
```

```python
# Give title to the graph
# Give label to x-axis and y-axis

plt.plot(batsman['index'], batsman['V Kohli'], marker = "*")
plt.plot(batsman['index'], batsman['RG Sharma'], marker = "*")

plt.title("Rohit Sharma Vs Virat Kohli Career Comparison")
plt.xlabel("Season")
plt.ylabel("Total Runs Scored")
```

```python
# Give colors to the lines

# Give title to the graph
# Give label to x-axis and y-axis

plt.plot(batsman['index'], batsman['V Kohli'], marker = "*", color = 'red')
plt.plot(batsman['index'], batsman['RG Sharma'], marker = "*", color = 'purple')

plt.title("Rohit Sharma Vs Virat Kohli Career Comparison")
plt.xlabel("Season")
plt.ylabel("Total Runs Scored")

# We can also pass hex codes in "color" parameter. Search it from google for any color u want.
```

```python
# Use different line-styles with linestyle and adjust width of the line using linewidth

# Give title to the graph
# Give label to x-axis and y-axis

plt.plot(batsman['index'], batsman['V Kohli'], marker = "*", color = 'red', linestyle = 'dashed', linewidth = 3)
plt.plot(batsman['index'], batsman['RG Sharma'], marker = "*", color = 'purple', linestyle = 'dotted', linewidth = 2)

plt.title("Rohit Sharma Vs Virat Kohli Career Comparison")
plt.xlabel("Season")
plt.ylabel("Total Runs Scored")

# There are many linestyles available. You can go through the documentation.
```

```python
# How would we know which line belongs to whom?
# We can use "label" parameter in plot and then use legend to show the labels

plt.plot(batsman['index'], batsman['V Kohli'], marker = "*", color = 'red', linestyle = 'dashed', linewidth = 3, label = 'Virat')
plt.plot(batsman['index'], batsman['RG Sharma'], marker = "*", color = 'purple', linestyle = 'dotted', linewidth = 2, label = 'Rohit')

plt.title("Rohit Sharma Vs Virat Kohli Career Comparison")
plt.xlabel("Season")
plt.ylabel("Total Runs Scored")
plt.legend()

# Why legend is located to this particular place only?
# We can change the location of legend using "loc" parameter in legend function, default is 'best'.
```

```python
# Limiting axis

# Plotting a simple 2-d plot on numeric-categorical data
year = [2015, 2016, 2017, 2018, 2019, 2020, 2021]
price = [50000, 54000, 59000, 63000, 60000, 55000, 2500000]

plt.plot(year, price, marker = '*')

# See, when any point is very far from the rest of the points, it is called outlier.
# Because of this, the rest of the line becomes flat and we cannot get the actual trend of the data.
# We can limit the axis to get a better view of the data.
# Here, we will have to limit y-axis because of the outlier in price.

# See the below code.
```

```python
year = [2015, 2016, 2017, 2018, 2019, 2020, 2021]
price = [50000, 54000, 59000, 63000, 60000, 55000, 2500000]

plt.plot(year, price, marker = '*')
plt.ylim(0, 75000)

# We can also limit years in x-axis using xlim function, but it has no meaning here
# plt.xlim(2017, 2020)
```

```python
# grid: Used for better interpreteability
# We can know better which point is lying where.

plt.plot(batsman['index'], batsman['V Kohli'], marker = "*", color = 'red', linestyle = 'dashed', linewidth = 3, label = 'Virat')
plt.plot(batsman['index'], batsman['RG Sharma'], marker = "*", color = 'purple', linestyle = 'dotted', linewidth = 2, label = 'Rohit')

plt.title("Rohit Sharma Vs Virat Kohli Career Comparison")
plt.xlabel("Season")
plt.ylabel("Total Runs Scored")
plt.legend()

plt.grid()
```

# 2) Scatter Plots

image.png

- Bivariate Analysis
- numerical vs numerical
- Use case - Finding correlation

`Note: To plt 2-D plot, we first plot scatter plot and then when we draw a line connecting all the dots, it becomes scatter plot.`

```python
# plot a simple scatter plot

x = np.linspace(-10, 10, 50)
y = 10*x + 3 + np.random.randint(0, 300, 50)
print(x[0:5])
print()

print(y[0:5])
```

```python
plt.scatter(x, y)

# We can observe that there is somewhat linear relationship between x and y.
```

```python
# Scatter plot on pandas dataframe

batter = pd.read_csv(r"Datasets/batter.csv")

batter.head()

# We will plot average and strike rate of first 50 players in the dataset

df = batter.head(50)
df
```

```python
plt.scatter(df['avg'], df['strike_rate'], color = "#FF04D5", marker = '*')
plt.title("Average Vs Strike Rate of Top 50 Batters")
plt.xlabel("Average")
plt.ylabel("Strike Rate")
```

```python
# size
tips = sns.load_dataset("tips")
# tips.head()

# Here, we have 2 numeric columns, total_bill and tip.
# Let's first plot scatter plot of these two columns

plt.scatter(tips['total_bill'], tips['tip'])
plt.xlabel("Total Bill")
plt.ylabel("Tip")

# Now, suppose we want such that, the size of dots should be based on the no_of person he came with.
# For example: The dot of person who came with 5 people should be bigger than the dot of person who came with 2 people.
# We can use "s" parameter. 
```

```python
plt.scatter(tips['total_bill'], tips['tip'], s = tips['size']*20)
plt.xlabel("Total Bill")
plt.ylabel("Tip")

# We multiplied with 20 bcz size column has very small values and we want to see the difference in size of dots clearly.

# Now, this graph shows us 3 info instead of 2.
# 1. Total Bill
# 2. Tip
# 3. Size of the group with which person came
```

```python
# Scatter plot using plt.plot()
# We have many limited functiionality here but this is faster.
# If we have millions of data points, then it is better to use plt.plot()
# instead of plt.scatter() because scatter is slower than plot.

plt.plot(tips['total_bill'], tips['tip'], 'o')
```

### Bar chart

image.png

- Bivariate Analysis
- Numerical vs Categorical
- Use case - Aggregate analysis of groups (Aggregation on y-axis)

```python
# simple bar chart
children = [10,20,40,10,30]
colors = ['red','blue','green','yellow','pink']

plt.bar(colors,children,color='orange')
```

```python
# horizontal bar chart
plt.barh(colors,children,color='orange')
```

```python
# Color an label

df = pd.read_csv(r"Datasets/batsman_season_record.csv")
# df

# Suppose we want to plot runs scored in 2015 and 2016 in single graph? We would do

plt.bar(df['batsman'], df['2015'])
plt.bar(df['batsman'], df['2016'])

# See, what is happenning here is, it is plotting the second graph on top of the first graph, so we are not able to see the first graph.
# Notice Dhoni's runs. You will get the point that Dhoni scored 370 runs in 2015
# Ploted down the graph with blue bar then 2016 runs are 284 which is less than 370,
# so it is plotting the second graph on top of the first graph and we are not able to see the first graph.
```

```python
len(df['batsman'])
```

```python
np.arange(5)
```

```python
x = np.arange(len(df['batsman']))
width = 0.25

# Create figure
plt.figure()

# Plot bars
plt.bar(x - width, df['2015'], width = 0.25, label='2015', color = 'orange')
plt.bar(x, df['2016'], width = 0.25, label='2016', color = 'blue')
plt.bar(x + width, df['2017'], width = 0.25, label='2017', color = 'green')

# Labels and title
plt.xlabel('Batsman')
plt.ylabel('Runs')
plt.title('Runs Comparison (2015–2017)')
plt.xticks(x, df['batsman'], rotation=45)
plt.legend()

plt.tight_layout()
```

```python
# a problem
children = [10,20,40,10,30]
colors = ['red red red red red red','blue blue blue blue','green green green green green','yellow yellow yellow yellow ','pink pinkpinkpink']

plt.bar(colors,children,color='red')
plt.xticks(rotation='vertical')
```

```python
# Stacked bar chart
plt.bar(df['batsman'], df['2017'], label = '2017')
plt.bar(df['batsman'], df['2016'], bottom = df['2017'], label = '2016')
plt.bar(df['batsman'], df['2015'], bottom = df['2016'] + df['2017'], label = '2015')

plt.legend()
```

### Pie Chart

image.png

- Univariate/Bivariate Analysis
- Categorical vs numerical
- Use case - To find contibution on a standard scale

```python
# Simple data

data = [25, 35, 15, 12, 18]
subjects = ['Gujarati', 'Hindi', 'English', 'Maths', 'Science']

plt.pie(data, labels = subjects)
```

```python
# On dataset
df = pd.read_csv(r"Datasets/gayle-175.csv")
df
```

```python
colors = ['orange', 'green', 'blue', 'white', 'violet', 'cyan']
plt.pie(df['batsman_runs'], labels = df['batsman'], autopct = "%.2f%%", colors = colors)
plt.show()
```

```python
# explode: This parameter is used to separate the pie at particular index than the rest of the pies.colors
# Let's explode ABD's pie. His index is 0

explode = [0.2, 0, 0, 0, 0, 0]
plt.pie(df['batsman_runs'], labels = df['batsman'], autopct = "%.2f%%", colors = colors, explode = explode)
plt.show()
```
