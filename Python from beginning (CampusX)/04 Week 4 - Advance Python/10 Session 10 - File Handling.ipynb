```python
f = open('Delete this also.txt', 'r')    # Open in read mode

s = f.read()
print(s)

f.close()
```

```python
# readline(): This reads the file line by line. It means it reads one line at a tile

f = open("Delete this also.txt", 'r')    # Open in read mode.

print(f.readline())    # Prints only 'Nikunj'
print(f.readline())    # Prints 'Jugal'. If u comment this, it will print only 'Nikunj'
print(f.readline())    # Prints 'Harshit'.

# In short, we will have to print 's', as many times we want to print  a new line

# Here, we can see that we are getting gap between two names. This is bcz readline() also prints a new line 
# and we know that the 'end' parameter in print function is bydefault set to '\n'. So we are getting a gap
# To avoide that gap, modify 'end' parameter in print() function.
```

```python
# Read all the lines through readlines()

f = open('Delete this also.txt', 'r')

while True:
    data = f.readline()

    if data == '':
        break

    else:
        print(data, end  ='')

f.close()
```

# Using Context Manager (with (keyword))

- It's a good idea to close a file after usage as it will free up the resources
- If we dont close it, garbage collector would close it
- 'with' keyword closes the file as soon as the usage is over

```python
with open("Delete this.txt", 'w') as f:    # This will ai=utomatically colse the file once our work is done

    l = ['Rajesh Khanna\n', 'Dilip Kumar\n', 'Rajendra Kumar\n', 'Raj Kapoor\n', 'Manoj Kumar\n', 'Jitendra\n', 'Raj Kumar\n']

    f.writelines(l)
```

```python
f.write('Hello')    # Will give error that 'I/O operation on a closed file', bcz file is closed automatically
```

```python
# read() using 'with' keyword

with open("Delete this.txt", 'r') as f:
    s = f.read()
    print(s, end = '')

print('\n',len(s))
```

```python
# read() using 'with' keyword

with open("Delete this.txt", 'r') as f:
    print(f.read(10), end = '**')
    print(f.read(10))
    

# We can see that on reading 10 chararters again, it is printing next 10 characters. This is bcz on printing first 10 characters, the cursor is after
# 10th character and on printing the next 10 characters, it prints from 11 to 20 and now the cursor is after 20th character.
```

# seek() and tell()

* seek(): seek() takes int as an argument. It takes the cursor to the position we passed to it as an argument 
* tell(): tell() function returns the current position of the cursor

```python
# tell()

with open("Delete this also.txt", 'r') as f:

    print(f.tell())
    print(f.readline())
    print(f.tell())
f.close()
```

```python
with open("Delete this also.txt", 'r') as f:
    print(f.read(6))
    print(f.tell())
```

```python
# seek

with open("Delete this also.txt", 'r') as f:
    print(f.readline())    # prints 'Nikunj'
    f.seek(0)              # Takes cursor to 0th position.
    print(f.readline())    # Again prints 'Nikunj'
```

# Serializationa and Deserialization

* Serialization: Serialization is the process of converting python's datatype to JSON format
* Deserialization: Deserialization is the process converting JSON to Python's datatype.

# What is a JSON file?

```python
# Serialization: First of all, import 'json' module

import json

# Using list

l = ['Nikunj', 'Jugal', "Harshit", "JEnshi", "Parth"]

with open("Demo.JSON", 'w') as f:
    json.dump(l, f)    # dump() takes two arguments, what to dump and a file handler object 
```

```python
# Using dictionary

dictionary = {
    'Name': "Nikunj",
    "Age": 24,
    "City": "Jetpur",
    "Religion": "Hindu"
}

with open("Demo2.JSON", 'w') as f:
    json.dump(dictionary, f, indent = 4)    # indent = 4, means 4 spaces


# Simple, this is what Serialization is.
```

```python
# Deserialization: Deserialization is the process converting JSON to Python's datatype.

# import json
import json

with open("Demo2.JSON", 'r') as f:
    d = (json.load(f))
    print(d)
    print(type(d))   # See, now we are getting class as dictionary, whereas above, we were getting as string
```

```python
# serialize and deserialize tuple. See the twist here.

import json
t = ("Nikunj", "Jugal", "Harshit", "Jenish", "Parth")

with open("Demo3.JSON", 'w') as f:
    json.dump(t, f)

# Open the file, you will notice that it is giving us list. So the twist is even if we d=serialize tuple, it returns us a list.
```

```python
# Serializing dictionary of lists

dict_list = {
    "Name": ['Nikunj', "Jugal", "Harshit", "Jensih", "Parth"],
    "Age": [24, 20, 25, 25, 24]
}

with open("Demo4.JSON", 'w') as f:
    json.dump(dict_list, f, indent = 4)

# We can also dump such complex datatypes.
```

# Serializing and Deserializing objects

```python
# Let's create a class

class Person:
    def __init__(self, name, age, city):
        self.name = name
        self.age = age
        self.city = city

p1 = Person("Nikunj", 24, "Jetpur")
```

```python
import json
with open("obj_serial.JSON", 'w') as f:
    json.dump(p1, f)
'''
Here, we will receive an error that object of class Person is not serializable.
This is bcz when we were serializing list, dict, tuple etc built-in datatypes, the structure was fixrd.
I mean, the structure of list is [1, 2, 'Nikunj', 'Jugal'], tuple is (1, 2, 'Nikunj'), 
dictionary is {"Name": "Nikunj, "Age": 24, "City": 'Jetpur"}.

But how will ti serialize the instance of a class? That's why we will receive an error.

So, to serialize an object of a class, we will have to show the dump() function how to serialize.
'''
```

# As a string

```python


def show_object(person):
    if isinstance(person, Person):
        return f"{person.name}'s age is {person.age} years old and he lives at {person.city}"
```

```python
# dumping the instance of an object

with open("obj_serial.JSON", 'w') as f:
    json.dump(p1, f, default = show_object)   # default = show_object means serialize in this manner.
```

```python
# Deserializing 

with open("obj_serial.JSON", 'r') as f:
    a = (json.load(f))
    print(a)
    print(type(a))
```

# Pickling
`Pickling` is the process whereby a Python object hierarchy is converted into a byte stream, and `unpickling` is the inverse operation, whereby a byte stream (from a binary file or bytes-like object) is converted back into an object hierarchy.

# Pickle Vs Json

- Pickle lets the user to store data in binary format. JSON lets the user store data in a human-readable text format.

```python
class Person:

  def __init__(self,name,age):
    self.name = name
    self.age = age

  def display_info(self):
    print('Hi my name is',self.name,'and I am ',self.age,'years old')
```

```python
p = Person('Nikunj',24)
```

```python
# pickle dump
import pickle
with open('person.pkl','wb') as f:
  pickle.dump(p,f)
```

```python
# pickle load
import pickle
with open('person.pkl','rb') as f:
  p = pickle.load(f)

p.display_info()
```
