# Let's first learn some basic exceptions

1) **ZeroDivisionError**: When we divide a number by zero `(45/0)`, we will get `ZeroDivisionError`

2) **TypeError**: You’ll get a TypeError in Python when you try to do something with a data type that isn’t allowed or doesn’t make sense for that operation.

For eg: 
--> "5" + 2 `TypeError`: Cannot concenate str to int.

--> x = 47 <br>
x()    `TypeError`    int object is not callable.

--> len(45)   `TypeError`: Object of type int has no len()

--> my_list = [1, 2, 3] <br>
my_list["1"]   `TypeError`: list indices must be integers or slices, not str

3) **ValueError**: You gave it the right kind of data, but the data doesn't make sense for that operation.

--> int("Hello")  `ValueError`: invalid literal for int() <br>
--> a, b = [1, 2, 3]  `ValueError`: too many values to unpack (expected 2) <br>
--> [1, 2, 3].index(5)  `ValueError`: 5 is not in list 




3) **FileNotFoundError**: When file is not found. <br>
os.list_dir('file_path')

4) **IndexError**: When something is out of index. <br>

l = [0, 1, 2, 3] <br>
l[8]  `IndexError`: list index out of range

5) **KeyError**: In dictionary, when a key is not present. <br>
d = {'Nikunj': 26, 'Jugal': 16} <br>
d['Jenish']  `KeyError`: 'Jenish'

6) **NameError**: A NameError happens when you try to use a variable, function, or object that hasn’t been defined yet.

--> print(a)  `NameError`: name 'x' is not defined. <br>
--> a()    `NameError`: name 'x' is not defined. <br>

# There are so many errors, but these are some common errors, we usually come across.
# Now, let's see how can we handle these errors.

```python
def divide(a, b):
    try: return a/b

    except ZeroDivisionError as e:
        return f'Exception: {e}'
        # return 'Dividing with zero is not possible'
```

```python
divide(5, 0)
```

```python
# We can also do this 

def divide(a, b):

    if b == 0:
        return 'Dividing with zero is not possible'
    else:
        return a/b

divide(5, 0)

# We can do this but for handling exceptions, we should use try-except blocks. Bcz if-else cannot deal with unexpected errors
```

```python
# Get index value

def get_index_value(l, index):

    try:
        return l[index]
    except IndexError as e:
        return f'Exception: {e}'

l = [5, 7, 45, 50, 20]

get_index_value(l, 8)
        
```

```python
# Key error

def get_key(d, key):

    '''
    Arguments: d, key

    d: Takes a dictionary <br>
    key: Key from the dictionary to get its value
    '''

    try:
        return d[key]
        
    except KeyError as e:
        return f'Key is not present: {e}'
```

```python
d = {'Nikunj': 26, 'Jugal': 16}

get_key(d, 'Harshit')
```

# Finally block

finally block always gets executed, whether an exception occurs or not.

```python
# Key error

def get_key(d, key):

    '''
    Parameters: d, key

    d: Takes a dictionary <br>
    key: Key from the dictionary to get its value
    '''

    try:
        return d[key]
        
    except KeyError as e:
        return f'Key is not present: {e}'

    finally:
        print('Process completed')
```

```python
d = {'Nikunj': 26, 'Jugal': 16}

get_key(d, 'Harshit')
```

# Use custom exception: We can also create and use our own exception rather than using the built-in ones.

```python
class cust_excep(Exception):
    def __init__(self, *args):
        super().__init__(*args)
```

```python
def withdraw(balance, amount):

    if balance < amount:
        raise cust_excep ('Insuffecient balance')
    balance -= amount
    return balance
    
```

```python
a = withdraw(5000, 8000)
```

```python
try:
    a = withdraw(5000, 8000)
except cust_excep as e:
    print(e)
```
