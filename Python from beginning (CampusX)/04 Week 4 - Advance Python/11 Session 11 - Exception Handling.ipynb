# Types of errors:

1) Index error: An IndexError is thrown when trying to access an item at an invalid index.
2) ModuleNotFound error: A ModuleNotFoundError is thrown when a module could not be found.
3) Key error: A KeyError is thrown when a key is not found.
4) Type error:  A TypeError is thrown when an operation or function is applied to an object of an inappropriate type.
5) Value error: A ValueError is thrown when a function's argument is of an inappropriate type.
6) Name error: A NameError is thrown when an object could not be found.
7) Attribute error: An AttributError is thrown when an object doesn't have any particular attribute.

# Exceptions

If things go wrong during the execution of the program(runtime). It generally happens when something unforeseen has happened.

- Exceptions are raised by python runtime.
- You have to takle it on the fly.

# **Examples**

- Memory overflow.
- Divide by 0 -> Logical error
- Database error.

# Why is it important to handle exceptions:
1) For user experience: We don't want our user to encounter the error messages if he does any mistake.
2) For security purpose, bcz our error message displays many informations of our file like, file name and location in the server, line and code where we got an error etc... A smart hacker may colect these pieces of code and create problems.

# How to handle exception:
**try-except blocks**

```python
# For e.g, let's forst create a file.

with open("Delete this file.txt", 'w') as f:
    f.write("Delete this file as this was created only for teaching purpose.")
```

```python
# Now, on opeaning this file, if a user mistakenly mispells the file name, he will encounter an error.

with open("Delete this file.txt", 'r') as f:
    f.read()

# See, we're getting a 'FileNotFoundError'. But we don't want our user to recerive any such dangerous error message(He may get shock and die. 
# Instead we want him to receive just a simple message that "File not found." To do this 'try and except' blocks are used,
```

```python
# try/except demo

try:
    with open('Delete this file.txt', 'r') as f:
        print(f.read())

except:
    print("Sorry!! File not found. Check the file's name.")

# Now, on misspelled file's name, we'll not receive any dangerous error message, instead a short message.
# If we have not misspelled the file's name, it would print the contents written in the file.
```

# Note: It is a good practice to write the codes inside try/except blocks whenever we are dealing with any external elements in python like file, database etc. So that is a user does any mistake, he won't receive any error messages.

# Catching specific exception.

```python
# Consider the below code first.

try:
    with open("Delete this file.txt", "r") as f:
        print(f.read())
    print(5/0)
    print(m)
    

except:
    print("Some error occured")

# Give comment on each code of try block and see what it is giving. On every error, it is giving a same message that "Some error occured."
# This should not happen. A user should be informed about his mistake.

# So, while writing any code, we should think about all the potential errors it may give and then create that many except blocks so that
# a user can be informed about his mistake properly.
```

```python
# 5/0 gives ZeroDivisionError and if file's name is mispelled, it gives FileNotFoundError.

try:
    f = open("Delete this file.txt", 'r')
    print(f.read())
    print(5/0)
    print(m)
    l = [1, 2, 30]
    print(l[5])

except FileNotFoundError:
    print("File not found.")

except NameError:
    print("Variable not defined.")

except ZeroDivisionError:
    print("Cannot divide by zero.")

except IndexError:
    print("Index out of range")


# Comment every line of code in the try's block and see which error is occured. To avoid NameError, declare variable 'm' in try's block.
```

# How can we know what is the error's name?

For e.g, we know that divison by zero is not possible and it will surely give an error, but we don't know which error will it throw.

Let's see how to identify that.

```python
try:
    print(10/5)
    print(int(60/6))
    print(5/0)

except Exception as e:
    print()
    print(e.with_traceback)

# We can see in an error message that 5/0 is a ZeroDivisionError.

```

```python
# Now, we are humans. So it may happen that we may not figure out every potential errors. So after writing several 
# except blocks for various exceptions, we write one additional except block which handles left out errors/exceptions. 
# Let's see how.

try:
    m = 'This is a demo code by NY Bosamiya'
    f = open("lete this file.txt", 'r')
    print(f.read())
    print(5/0)
    print(m)
    l = [1, 2, 30]
    print(l[5])

except FileNotFoundError:
    print("File not found.")

except NameError:
    print("Variable not defined.")

except ZeroDivisionError:
    print("Cannot divide by zero.")

except IndexError:
    print("Index out of range")

except Exception as e:    # Catches general or say left out exceptions.
    print(e)

# Try commenting required lines in try's block such that it triggers the last except's block and see what happens.
```

# Note: An except block to catch general or left out exceptions should always be written last bcz if we write it in between, we will not receive our own's defined message.

# else:

- The code inside **else** block gets execute only if **try** block gets executed properly.
- In short, it gets execute if **except** block doesn't execute. Simple, if **try** block gets execute properly, it means there is no error and so **except** block will not execute and directly else block will execute.
- Only one will get execute, either **try** or **else**.
- Inshort, what we want to perform something after try's success, we put it into else's block.

Let's consider below example.

```python
try:
    f = open("Delete this file.txt", 'r')

except Exception as e:
    print(e)

else:
    print(f.read())
    f.close()

# Here, it will first try to open the file, if exception occurs, it will execute 'except' block and if not, then 'else' block.
# Try to misspell the file's name. It will execute 'except' block.
```

```python
# Same as above with simple changes

try:
    f = open("Delete this file.txt", 'r')

except FileNotFoundError:    # Change
    print("File not found.")
    
except Exception:    # Change.
    print("Tu kar kya raha hai bhai!!.")

else:
    print(f.read())
    f.close()

# Here, it will first try to open the file, if exception occurs, it will execute any one of the 'except' blocks and if not, then 'else' block.
# Try to misspell the file's name. It will execute 'except' block.
```

# finally block:

* finally block will always get executed.
* It is used to clear the mess we have created in the code like closing the files or clearing anything.

```python
try:
    f = open("Delete this file.txt", 'r')

except FileNotFoundError:    # Change
    print("File not found.")
    
except Exception:    # Change.
    print("Tu kar kya raha hai bhai!!.")

else:
    print(f.read())
    f.close()

finally:
    print()
    print("Apun finally hai. Apun execute hoga he. Ye hi apun ka funda hai baap.")
```

# Raise Exception.

- In Python programming, exceptions are raised when errors occur at runtime. 
- We can also manually raise exceptions using the raise keyword.
- We can optionally pass values to the exception to clarify why that exception was raised.

```python
class Bank:
    def __init__(self, balance):
        self.balance = balance

    def withdraw(self, amount):
        
        # Let's think about the possible errors here.
        # 1) Amount entered to withdraw exceeds the balance.
        # 2) amound entered is negative
        if amount < 0:
            raise Exception("-ve is not allowed.")
            
        elif amount > self.balance:
            raise Exception("Paise nhi hai tere account me lukhe.")
        else:
            print("Money withdrawn.")
            self.balance = self.balance - amount
```

```python
nikunj = Bank(10000)
```

```python
try:
    nikunj.withdraw(15000)
except Exception as e:    # Exception is a class and we are passing here the object of it which we created while creating the class and and it's method. 
    pass                  # moreover, ever exception we learnt till now, are all classes. That is the reason we are able to write inside parenthesis.
else:
    print(nikunj.balance)

# Try replacing the amount with -ve and more that the current balance
```

# Creating custom exceptions.

* Python allows us to define our own exception class which works as we want them to work.
* While creating our own exception class, we'll hafta inherit 'Exception' class else we won't be able to raise that user-defined exception.

```python
class My_Exception(Exception):
    def __init__(self, message):
        print(message)

class Bank2:
    def __init__(self, balance):
        self.balance = balance

    def withdraw(self, amount):
        
        if amount < 0:
            raise My_Exception("Amount can't be negative.")
        
        if amount > self.balance:
            raise My_Exception("Paise nhi hai tere account me lukhe.")

        else:
            print("Money withdrawn.")
            self.balance = self.balance - amount
```

```python
jugal = Bank2(25000)
```

```python
try:
    jugal.withdraw(-4000)
except My_Exception as e:
    pass
else:
    print(jugal.balance)

# Try replacing the amount with -ve and more than the current balance
```
