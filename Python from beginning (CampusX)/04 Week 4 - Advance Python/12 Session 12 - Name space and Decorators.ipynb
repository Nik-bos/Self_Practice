# What is Namespace: `Namespace` is a space that holds identifiers. Programatically speaking, namespaces are dictionary of `identifiers(keys)` and their `objects(values)`.

For eg. a = 10, b = 20<br>
**Namespace**: {'a': 10, 'b': 20}

There're four types of namespace in python. They're:
1) `Local Namespace`
2) `Enclosing Namespace`
3) `Global Namespace`
4) `BuiltIn Namespace`


# `Scope and LEGB rule`:

`Scope` is that area where a namespace(value) is directly accessable.<br>

`Note LEGB rule`: The interpreter searches for a name(variable) from inside out searching in `local scope`, then in `enclosing scope`, then in `global scope` and finally in `built-in scope`. If it doesn't find even in `built-in` scope, it will raise a `NameError`


`Let's first learn **local and global scope**.`

```python
# Local and global scope.

a = 10    # Global variable(Global scope)

def temp():
    b = 20    # Local variable(Local scope)
    print(b)

temp()
print(a)
```

```python
# Same variable name

a = 10    # Gloabl variable(Global Scope)

def temp():
    a = 20    # Local Variable (Local scope)
    print(a)

temp()
print(a)
```

```python
# Local does not have but global have

a = 10

def temp():
    print(a)    # Will search in Enclosed scope(Ignore for now), then will search in Global scope(found there)

temp()
print(a)
```

```python
# Modifying global variable inside local variable.

a = 10

def temp():
    a += 5
    print(a)

temp()
print(a)

# See, we cannot modify global variable inside local scope. But somehow we wanna do it then? See the below code
```

```python
# USe 'global' keyword

a = 10

def temp():

    global a
    a += 5
    print(a)

temp()
print(a)

# Caution! We can do this but should never modify global variables bcz then it will create problems in the other places where they are used.
```

```python
# Global variable creaated from function(Local scope)

def temp():

    global a
    a = 10
    print(a)

temp()
print(a)
```

```python
# Function's parameter is always local variable

def temp(n):
    print(n)   # This is a local variable

a = 10    # This is a global variable
temp(a)

print(a)
```

```python
a = 10

def temp():
    a = 20
    global a
    a = 30
    print(a)

temp()
print(a)

# See this will give error. The error is before global declaration of 'a' we have made local variable 'a'.
```

# 3) `Enclosing scope`: It is always seen in nested functions 

`See, the below code`

```python
def outer():
    def inner():
        print('Inside inner function.')
    inner()
    print("Outer function")

outer()
print("In global scope")


# What will be the output of the above code?

```

```python
# See this code to understand the concept of enclosing scope

def outer():
    a = 10    # Enclosing scope
    
    def inner():
        a = 20
        print(a)    # Local scope
    inner()
    print("Outer function")

outer()


# What will this give?   First it will search in Local Scope. So a = 20 and will never search in any other scopes.
```

```python
# See this

def outer():
    a = 10    # Enclosing scope
    
    def inner():
        print(a)
    inner()
    print("Outer function")

outer()


# What will this give?
```

```python
a = 10    # Global scope

def outer():
    def inner():
        print(a)
    inner()
    print("Outer function")

outer()


# What will this give?
# Firstly will search in Local scope, will not find 'a'.
# Then will search in enclosing scope, will not find 'a'.
# Then will search in global scope and will find it. Hence will not move further for searching in built-in scope
```

# Decorators.

A `decorator` is a function in Python which receives a fuction as an argument during a function's call, adds some functionality to that received function and then finally returns a function.

This can happen bcz Python's functions are `First class citizens`.
There're two types of decorators available in Python. They're,
1) `Built-in decorators` like @abstract method, @staticmethod, @classmethod, @property etc...
2) `User definied decorators` that we can create according to our needs.

```python
# Why functions in Python is a first class citizen? Go to 'Session - 6. It's explained very well there.
# Moreover, one code to understand it, is also given here.

def modify(func,num):
  return func(num)

def square(num):
  return num**2

modify(square,2)
 
```

```python
# Simple example of decorator.

def my_decorator(fun):
    def wrapper():
        print('*' * 20)
        fun()
        print('*' * 20)
    return wrapper

def greet():
    print("Hello")

def display():
    print("Hi Nikunj.")
```

```python
a = my_decorator(greet)
a()
```

```python
b = my_decorator(display)
b()
```

```python
'''
See the code and recall the explanation of decorators I gave in the beginning.
1) Decorator is a function: Here in this code 'my_decorator' is a function.
2) It takes another function as an argument: Here, 'my_decorator' takes 'greet' and 'display', which are functions, as an argument.
3) Decorator returns a function: Here, it returns a function named 'wrapper'.
Hence, my_decorator is a decorator.
'''
```

# Understand this very carefully:

Untill now we lernt that when a function returns something or after the execution of a function is completed, all the arguments it received gets destroyed. Right?

Okk. Now see the working of below code we used a while ago <br>
**a = my_decorator(greet)<br>
a()**

Here, when we call `my_decorator` with `greet` as an argument, inside variable 'a', it directly returns 'wrapper' to a. Here it just returns but doesn't execute 'wrapper' bcz it is not yet called.<br>

Now when we call it via variable 'a' in code `a()`, it executes wrapper() function and prints "Hello", which is printed on `greet()`function.<br>

Wondered, how can it print 'Hello' even after `my_decorator()` function already returned something?<br>

As per rule, after `my_decorator()` function returns `wrapper`, all its arguments should be deleted, then how can it access `greet()` function later?

This is because a inner function, even after the death of the parent function, can access the arguments of the parent function.<br>
The working of decorators is based on this special concept of Python.<br>

This whole functionality is actually called `Closour` in python.

Now we know what is decorator. But the syntax is quite indirect.<br>
a = my_decorator(greet)    <br>
a()

See we are first calling `my_decorator()` with `greet()` as an argument and storing it on variable `a`.<br>
Then it returns 'wrapper' to variable 'a'.<br>
Finally we call a() and then we get the final result.

So let's see the short way of doing these.

```python
def my_decorator(fun):
    def wrapper():
        print('*' * 20)
        fun()
        print('*' * 20)
    return wrapper
```

```python
# Example: 1

@my_decorator
def greet():
    print("Hellow")
```

```python
greet()
```

```python
# Example: 2

@my_decorator
def display():
     print('Hi Nikunj')
```

```python
display()
```

# Now let's create a meaningful decorator which prints the time taken by our function in execution.


```python
import time

def timer(fun):
    def wrapper():
        start = time.time()
        fun()
        print(f"Time taken = {time.time() - start} seconds.")
    return wrapper
         
```

```python
def greet():
    print('Hellow. My name is Nikunj')
    time.sleep(2)    # sleep() delays the execution. Here, delays for 2 seconds.
```

```python
a = timer(greet)
a()
```

```python
@timer
def display():
    print("Kem cho?? Maja ma??")
    time.sleep(2)

display()
```

```python
@timer
def square(num):
    time.sleep(2)
    print(num**2)

square(5)
```

```python
def design(function):
    def wrapper():
        print('#*' * 10)
        function()
        print("#*" * 10)
    return wrapper
```

```python
def greet5():
    print("Hii, Nikunj")
```

```python
z = design(greet5)
z()
```

```python
@design
def greet6():
    print("Jai shree Ram.")

greet6()
```
