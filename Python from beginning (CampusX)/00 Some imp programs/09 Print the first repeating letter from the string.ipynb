# How can we use `else` with `if` as we used here in the outer loop?

The concept of using an else block with a for (or while) loop in Python can be a bit counterintuitive at first because we typically associate else with if statements. However, Python allows you to attach an else block directly to loops, and this behavior is unique to Python.

How else with for Loops Works
In Python, the else block in a for or while loop is executed only when the loop completes normally—that is, when the loop iterates over all elements or iterations without encountering a break. If the loop terminates because of a break, the else block is skipped.

The key takeaway is that the else block in loops does not depend on any condition in an if statement inside the loop, but rather on whether the loop exits "naturally" or prematurely.

Here’s how it works:

The for loop starts executing.
If the loop encounters a break statement and exits prematurely, the else block is skipped.
If the loop completes all iterations (i.e., it does not encounter a break), the else block executes.


Example 1: Basic Loop with else(See below codes)

```python
for i in range(5):
    print(i)
else:
    print("Loop completed without a break.")
```

```python
for i in range(10):
    if i == 6:
        break
    print(i)
else:
    print("Loop completed without a break.")
```

```python
s = 'Nikunj'.lower()
found = False

for i in range(len(s)):
    for j in range(i+1, len(s)):
        if s[i] == s[j]:
            print(f"{s[i]} is the first repeating character")
            found = True
            break   # break inner loop
    if found:
        break       # break outer loop

if not found:
    print("No repeating characters.")
```
