```python
def is_prime(n):
    if n == 0:
        return "Insert a valid number."

    elif n == 1:
        return f'{n} is neither prime nor composite'

    else:
        for i in range (2, n):
            if n % i == 0:
                return f'{n} is a composite number.'
        return f"{n} is a prime number."
```

```python
is_prime(20)
```

```python
for i in range(0,26):
    print(is_prime(i))
```

```python
def is_prime(n):
    if n <= 0:
        return "Insert a valid number."

    if n == 1:
        return "1 is neither prime nor composite."

    for i in range(2, int(n**0.5) + 1):    # Always divide till sqrt(n)+1
        if n % i == 0:
            return f"{n} is a composite number."

    return f"{n} is a prime number."

is_prime(51)
```
