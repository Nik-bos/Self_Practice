```python
from collections import Counter

def count_letter_frequency(s):
    # Use Counter to count frequency of each character
    frequency = Counter(s)
    return frequency

# Example usage:
string = 'nikunj'
result = count_letter_frequency(string)

# Print the frequency of each character
for letter, count in result.items():
    print(f"'{letter}': {count}")
```

```python
s = 'nikunj'
d = {}
for i in range(len(s)):
    if s[i] not in d:
        d[s[i]] = 1
    else:
        d[s[i]] += 1

print(d)
```

```python
def get_freq(s):
    s = s.lower()
    d = {}
    for i in s:
        if i not in d:
            d[i] = 1
        else:
            d[i] += 1
    return d
```
