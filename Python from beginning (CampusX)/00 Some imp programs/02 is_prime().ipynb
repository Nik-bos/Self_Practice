```python
def is_prime(n):
    if n == 0:
        return "Insert a valid number."

    elif n == 1:
        return f'{n} is neither prime nor composite'

    else:
        for i in range (2, n):
            if n % i == 0:
                return f'{n} is a composite number.'
        return f"{n} is a prime number."
```

```python
is_prime(20)
```

```python
for i in range(0,21):
    print(is_prime(i))
```
