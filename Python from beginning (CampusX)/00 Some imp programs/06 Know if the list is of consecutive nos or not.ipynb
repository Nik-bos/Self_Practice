```python
# Taking input from the user as a string
input_string = input("Enter elements separated by spaces: ")

# Splitting the input string and converting it to a list of integers
lst = list(map(int, input_string.split()))
lst.sort()

wrong_nos = ''
corrected_nos = ''
# new_list = lst[:]

for i in range(len(lst) - 1):    
    if lst[i + 1] == lst[i] + 1:
        continue
    else:
        
        # wrong_nos.append(lst[i+1])
        wrong_nos = wrong_nos + ',' + str(lst[i+1])
        lst[i + 1] = lst[i] + 1
        # corrected_nos.append(lst[i] + 1)
        corrected_nos = corrected_nos + ',' + str(lst[i] + 1)
# print("Original list:", lst)
# print("New list:", new_list)
print("Wrong numbers:", wrong_nos[1::])
print()

print("Corrected numbers:", corrected_nos[1::])
print()

print(f"Change {wrong_nos[1::]} with {corrected_nos[1::]} respectively to get correct permutation.")
print()

print(lst)
```
