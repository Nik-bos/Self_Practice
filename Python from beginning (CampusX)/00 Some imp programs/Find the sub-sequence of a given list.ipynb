# Meaning of sub-sequence.

A subsequence of a list is a sequence that can be derived from the original list by deleting some or no elements without changing the order of the remaining elements.

For example, if you have a list [1, 2, 3], some of its subsequences could be [1, 2], [1, 3], [2, 3], [1], [2], [3], or even an empty list [].

```python
def subsequences(lst):
    subs = [[]]
    for item in lst:
        subs += [sub + [item] for sub in subs]
    return subs

# Example usage:
my_list = [1, 2, 3]
print("All subsequences of", my_list, "are:")
print(subsequences(my_list))
```

```python
def subsequences(lst):
    subs = [[]]
    for item in lst:
        new_subs = []
        for sub in subs:
            new_subs.append(sub + [item])
        subs += new_subs
    return subs

# Example usage:
my_list = [1, 2, 3]
print("All subsequences of", my_list, "are:")
print(subsequences(my_list))
```
