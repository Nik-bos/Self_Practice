# Function.

```python
def fib(n):

    a = 0
    b = 1
    l = []

    if n == 0:
        print("Enter a number greater than 0")
    elif n == 1:
        l.append(a)
        print(l)
    elif n == 2:
        l.extend([a, b])
        print(l)

    else:
        l.extend([a, b])
        for i in range(n-2):
            c = a+b
            l.append(c)

            a,b = b, c

        print(l)
```

```python
fib(10)
```
