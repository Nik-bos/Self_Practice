```python
l = ['Nikunj', 'Jugal', 10, 4, 3.5]

lst_int = []
lst_float = []
lst_str = []

for i in range(len(l)):
    if type(l[i]) == int:
        lst_int.append(l[i])
    elif type(l[i]) == float:
        lst_float.append(l[i])
    else:
        lst_str.append(l[i])
```

```python
print(lst_int)
```

```python
l = ['Nikunj', 'Jugal', 10, 4, 3.5, 5.5]

l_int = []
l_float = []
l_str = []

for i in l:
    if isinstance(i, int):
        l_int.append(i)
    elif isinstance(i, str):
        l_str.append(i)
    else:
        l_float.append(i)



print(l_int)
print()

print(l_str)
print()

print(l_float)
print()
print()
```

```python
# Input: [55, 41]
# Output: [10, 5]    5+5 and 4+1

l = [72, 101]

l2 = []

for i in l:
    sum = 0
    while i > 0:
        unit = i % 10
        sum = sum + unit
        i = i//10
    l2.append(sum)
    

l2
```
