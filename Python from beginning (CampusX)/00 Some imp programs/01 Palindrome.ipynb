```python
def palin(n):
    if type(n) == str:
        a = n.lower()
        if a == a[::-1]:
            print("It's a palindrome.")
        else:
            print("It's not a palindrome.")

    elif type(n) == int:
        orig_no = n
        rev_no = 0

        while orig_no > 0:
            digit = orig_no % 10
            rev_no = rev_no * 10 + digit
            orig_no = orig_no // 10

        if n == rev_no:
            print("The number is a palindrome.")
        else:
            print("The number isn't a palindrome.")
```

```python
palin(122)
```

```python
palin(12121)
```

```python
palin("ABcBa")
```

```python
palin('abcba')
```
