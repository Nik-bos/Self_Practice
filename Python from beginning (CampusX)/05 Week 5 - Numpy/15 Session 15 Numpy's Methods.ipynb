# 1) `np.sort()`: Sorts the array

`np.sort(arr)` returns numpy array and `sorted(arr)` returns list

```python
arr = np.random.randint(1, 101, 25)
print(arr)
print()
print()

print(np.sort(arr))
print()
print()

np.sort(arr)[::-1]    # descending order.
```

```python
# 2-D array

arr2 = np.random.randint(1, 50, 12).reshape((4, 3))

arr2
```

```python
print(np.sort(arr2))    # Sorting is performed row wise
print()
print()
print()


print(np.sort(arr2, axis = 0))    # Sorting is performed column wise
print()
print()
print()

```

# 6) np.where(): Returns indices of an items in an array based on the condition.
# This is also used as an if-else statements.

```python
# Return indices of an item based on condition.

a = np.random.randint(1, 50, 15).reshape(15)
a
```

```python
np.where(a>15)
```

```python
# As an if-else statement.

np.where(a>5, True, False)    # True if a>5 else False
```

```python
# Replace even numbers with 1 and odd with 0

np.where(a%2 == 0, 1, 0)
```

```python
# Replace even numbers with 1 and leave odd ones as it is.

np.where(a%2==0, 1, a)
```

# `18) Set functions`: Below given are set's funnctions what can used with numpy array but it gives result also in numpy array.<br>It gives result in an ascending order.

- np.union1d()
- np.intersect1d()
- np.setdiff1d()
- np.setxor1d()
- np.in1d()

```python
m = np.array([1, 2, 3, 4, 5, 9, 10, 14, 15])
n = np.array([6, 7, 8, 9, 10, 5, 2])

print(m)
print()
print()

print(n)
```

```python
# np.union1d(): Returns union of two arrays.

np.union1d(m, n)
```

```python
# np.intersect1d: Performs intersection.

np.intersect1d(m, n)
```

```python
# setdiff1d(): Returns difference of two set.  For eg. set A - set B means remove all the elements of B from A(if present)
# Set B - Set A means vice-versa.

np.setdiff1d(m, n)    # m - n means remove all the elements of n from b.
```

```python
# setxor(): This removes all the common elements of the two sets and returns other elements.

np.setxor1d(m, n)   # 5 and 2 are common. So will remove 5 and 2 and return 1, 3, 4, 6, 7, 8, 9, 10
```

```python
# setin1d(): Ckecks if the passed value is present in an array.

np.in1d(m, 5)
```

```python
m[np.in1d(m, 9)]
```
