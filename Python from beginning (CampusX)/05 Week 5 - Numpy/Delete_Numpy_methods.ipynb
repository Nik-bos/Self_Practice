# 1) `np.sort()`: Sorts the array

`np.sort(arr)` returns numpy array and `sorted(arr)` returns list

```python
np.random.randint(1, 101, 25)
```

```python
arr = np.random.randint(1, 101, 25)
print(arr)
print()
print()

print(np.sort(arr))
print()
print()

print(np.sort(arr)[::-1])    # descending order.
```

```python
# 2-D array

arr2 = np.random.randint(1, 50, 12).reshape((4, 3))

arr2
```

```python
np.sort(arr2, axis = 1)
# axis = 0 will sort each column
# axis = 1 will sort each row
```

```python
# 2-D array

print(arr2)
print()
print()

a = np.array([[1, 2, 3, 4]]).reshape((4, 1))
print(a)

np.append(arr2, a, axis = 1)    # axis = 1 to add a column
```

# `18) Set functions`: Below given are set's funnctions what can used with numpy array but it gives result also in numpy array.<br>It gives result in an ascending order.

- np.union1d()
- np.intersect1d()
- np.setdiff1d()
- np.setxor1d()
- np.in1d()

```python
m = np.array([1, 2, 3, 4, 5, 9, 10, 14, 15])
n = np.array([6, 7, 8, 9, 10, 5, 2])

print(m)
print()
print()

print(n)
```

```python
# np.union1d(): Returns union of two arrays.

np.union1d(m, n)
```

```python
# np.intersect1d: Performs intersection.

np.intersect1d(m, n)
```

```python
# setdiff1d(): Returns difference of two set.  
# For eg. set A - set B means remove all the elements of B from A(if present)
# Set B - Set A means vice-versa.

np.setdiff1d(m, n)    # m - n means remove all the elements of n from b.
```

# 19) `np.clip`: This will clip the array. Means if np.clip(m, a_min = 25, a_max = 75), then all the numbers greater than a_max will become 75 and all the numbers smaller than a_min will become 25 and rest will be as they are

```python
m = np.array([1, 5, 9, 12, 15, 17, 25, 24, 
              29, 35, 32, 38, 45, 42, 49, 52, 
              51, 58, 62, 68, 67, 78, 71, 76, 
              86, 82, 87, 91, 98, 94])
```

```python
np.clip(m, a_min = 15, a_max = 87)
```
