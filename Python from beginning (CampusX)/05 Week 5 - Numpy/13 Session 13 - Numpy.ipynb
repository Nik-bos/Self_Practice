# Creating numpy arrays. 
Function is `np.array()` and it takes list as an input.

**Bydefault, dtype of numpy array is `float`.**

```python
data = [1, 2, 3]

arr1 = np.array(data)

print(arr1)
print(type(arr1))

print(arr1.ndim)    # ndim attribute shows dimension of a numpy array.
print(arr1.dtype)    # dtype attribute shows datatype of a numpy array.
```

# Indexing and Slicing

* `Indexing`: Means fetching only one item.
* `Slicing`: Means fetching more than one item.

# Indexing

```python
a1 = np.arange(10)
a2 = np.arange(12).reshape((3, 4))
a3 = np.arange(8).reshape(2 ,2, 2)
```

```python
print(a1)
print()
print()
print()

print(a2)
print()
print()
print()

print(a3)
print()
print()
print()
```

```python
print(a1[4])
print(a1[-2])
print(a1[3:8])
```

```python
a2    # 2-D array
```

```python
print(a2[1][2])    # Counting of rows and columns starts from zero.
print(a2[1, 2])    # Same result as above
```

```python
a3    # 3-D array
```

```python
print(a3[1][0, 1])    # 1th matrix, 0th row and 1th column
print(a3[1, 0, 1])    # Same result as above.

print(a3[1, 1, 0])    # 6
```

# Slicing.

`Remember`: While slicing, before comma, **rows** and after comma, **columns**.

```python
a1
```

```python
print(a1[2: 5])
print(a1[2: 8: 2])

print(a1[-3: -7: -1])
print(a1[-7: -3:])

print(a1[-3: -7])    # We cannot move from right to left, like in list.
```

```python
a2    # 2-D array
```

```python
print(a2[0, :])    # Fetch 1st row
print()
print()

print(a2[:, 2])    # all rows and 2nd column.
print()
print()

print(a2[0, 0:3])    # 0th row, from 0th to 3rd item.
print()
print()

print(a2[0:2:])    # from 0th to 2nd row and all items.
print()
print()

print(a2[0:2, 0:3])    # 0th to 2nd row and 0th to 3rd item
```

```python
# Fetch [5, 6, 9, 10]

print(a2[1:3, 1:3])    # 1th to 3rd row and 1th to 3rd columns
print()
print()

# Fetch [0, 3, 8, 11]
print(a2[0:3:2, 0:4:3 ])    # 0th to 3rd row with 2 steps and 0th to 4th columns with 3 steps.

# 2 and 3 steps in rows and columns means every 2nd and 3rd row and column respectively.
```

```python
# Fetch [1, 3, 9, 11]
print(a2)
print()
print()
print()

print(a2[0:3:2, 1:4:2])

# All these codes give same output bcz, when we want to start from the beginning, we can keep it blank and when we want to
# go till the end, we can keep that place blank as well. 
```

```python
# Fetch [4, 7]
print(a2[1, 0::3])
print(a2[1, 0:4:3])    # This code is same as the above one.
print(a2[1, ::3])    # This is also same as the above 2 codes.

# All these codes give same output bcz, when we want to start from the beginning, we can keep it blank and when we want to
# go till the end, we can keep that place blank as well.
```

```python
# Fetch [1, 2, 3], [5, 6, 7]

a2[0:2, 1:]
```

```python
a2[0:2, 1:4]    # This is same as the above code.
```

# Stacking.
* `Stacking` means we can join two or more numpy arrays.

There're two types of stacking.
1) `Horizontal stacking`: Stack arrays horizontally. **hstack()**
2) `Vertical stacking`: Stack arrays vertically.    **vstack()**

**Size of both the arrays should be same**

```python
# Horizontal stacking.

b2 = np.random.random(12).reshape((3, 4)) * 10
b2 = b2.round()

print(b2)
print()
print()
print()

print(a2)
```

```python
np.hstack((a2, b2))
```

```python
# Vertical stack

np.vstack((a2, b2))
```

```python
# We can stack as many arrays we want but size of all must be same.

np.hstack((b2, a2, b2))
```

# Splitting; Splits the array into parts.

* There are two types:
1) Horizontal split: **hsplit()** and
2) Vertical split: **vsplit()**

* Split must happen into equal parts
* Takes 2 arguments, array and how many splits.

```python
print(a2)
print()
print()
print()

np.hsplit(a2, 2)    # takes 2 arguments, array and how many splits.
```

```python
a2 = np.arange(16).reshape((4, 4))
a2
```

```python
np.vsplit(a2, 2)
```
