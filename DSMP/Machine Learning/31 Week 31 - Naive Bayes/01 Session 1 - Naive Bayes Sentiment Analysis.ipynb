# Removing HTML Tags

```python

import re
clean = re.compile('<.*?>')
re.sub(clean, '', df.iloc[2].review)
```

```python
# Function to clean html tags
def clean_html(text):
    clean = re.compile('<.*?>')
    return re.sub(clean, '', text)
```

```python
df['review']=df['review'].apply(clean_html)
df['review'].iloc[0]
```

# Function to remove special characters

```python
def remove_special(text):
    x=''
    
    for i in text:
        if i.isalnum():
            x=x+i
        else:
            x=x + ' '
    return x
```

```python
remove_special("The movie @BalajiCineplex was a great experience")
```

```python
df['review']=df['review'].apply(remove_special)
```

```python
df['review'].iloc[0]
```

# Remove stopwords

```python
from nltk.corpus import stopwords
```

```python
stopwords.words('english')
```

```python
def remove_stopwords(text):
    x=[]
    for i in text.split():
        
        if i not in stopwords.words('english'):
            x.append(i)
    y=x[:]
    x.clear()
    return y
```

```python
df['review']=df['review'].apply(remove_stopwords)
df['review'].iloc[0]
```

# Performing Stemming

```python
# Perform stemming

from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
y=[]
def stem_words(text):
    for i in text:
        y.append(ps.stem(i))
    z=y[:]
    y.clear()
    return z
```

```python
stem_words(['I','loved','loving','it'])
```

```python
df['review']=df['review'].apply(stem_words)
```
