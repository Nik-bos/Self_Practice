```python
import pandas as pd
import numpy as np

np.set_printoptions(suppress=True, precision=4)
pd.set_option('display.float_format', '{:.4f}'.format)
```

```python
class Mera_Lr:
    def __init__(self):
        self.m = None
        self.c = None

    def fit(self, x_train, y_train):
        numerator = 0
        denominator = 0

        for i in range(x_train.shape[0]):
            # sum all numerator
            numerator = numerator + ((x_train[i] - x_train.mean()) * (y_train[i] - y_train.mean()))
            # sum all denominator
            denominator = denominator + ((x_train[i] - x_train.mean())**2)

        # print(numerator)
        # print(denominator)
        
        # Update m and c 
        self.m = numerator/denominator
        self.c = (y_train.mean() - (self.m * (x_train.mean())))

        print(f"Slope (m) = {self.m}")
        print(f"Intercept (c) = {self.c}")

    def predict(self, x_test):
        return (self.m * (x_test)) + self.c
```

```python
df = pd.read_csv(r"00 placement.csv")
df
```

```python
x = df.iloc[:, 0:-1].values
y = df.iloc[:, -1].values

print(x[0:5])
print()
print()

print(y[0:5])
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 1)

print(x_train.shape)
print(y_train.shape)
```

```python
# Check the class-> Mera_Lr

lr = Mera_Lr()

lr.fit(x_train, y_train)
```

```python
# Predict
lr.predict(8.6)
```

```python
y_pred = lr.predict(x_test).flatten()
y_pred
```

```python
error  = (y_pred-y_test)**2
error
```
