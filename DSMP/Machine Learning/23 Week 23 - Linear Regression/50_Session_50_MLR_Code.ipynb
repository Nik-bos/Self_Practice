```python
import pandas as pd
import numpy as np
import plotly.express as pe
import plotly.graph_objects as go
from sklearn.datasets import make_regression

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
x, y = make_regression(
    n_samples = 200,
    n_features = 2,
    n_informative = 2,
    n_targets = 1,
    noise = 50)

print(x[0:5])
print()
print()

print(y[0:5])
```

```python
df = pd.DataFrame(
    {
        'feature_1': x[:, 0],
        'feature_2': x[:, 1],
        'target': y
    }
)

df.head()
```

```python
fig = pe.scatter_3d(df, x = 'feature_1', y = 'feature_2', z='target')

fig.show()

# If we zoom, we will be able to see points are scattered, else we will see as if points are very near to each other.
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 1)
```

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()

lr.fit(x_train, y_train)
```

```python
y_pred = lr.predict(x_test)

y_pred[0:5]
```

```python
x_test.shape
```

```python
mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
rmse = np.sqrt(mse) 
r2 = r2_score(y_test,y_pred)
ar2 = 1 - ((1-r2)*(40-1)) / (40-1-2)



print("MAE", mae)
print("MSE", mse)
print("RMSE", rmse)
print("R2 score", r2)
print(f"AR2-score: {ar2}")

# See, our R2-Score is 0.6381 and Adjusted R2 Score is 0.6185.
# We know Adjusted R2-Score is more accurate, hence all independent features combined explains 0.61 i.e
# 61% variation in the target column.
```

# See features' weights and intercept

```python
print(lr.coef_)    # This shows weights of 'feature_1' and 'feature_2' respectively. Beta_1 and Beta_2
print(lr.intercept_)    # Beta_0
```

# See the hyperplane visually.

```python
import numpy as np
import plotly.graph_objects as go

# Create grid
x1_range = np.linspace(x[:, 0].min(), x[:, 0].max(), 50)
x2_range = np.linspace(x[:, 1].min(), x[:, 1].max(), 50)

x1_grid, x2_grid = np.meshgrid(x1_range, x2_range)

# Predict on grid
grid_points = np.c_[x1_grid.ravel(), x2_grid.ravel()]
z = lr.predict(grid_points).reshape(x1_grid.shape)

# Plot
fig = go.Figure()

# Surface (hyperplane)
fig.add_trace(go.Surface(
    x=x1_grid,
    y=x2_grid,
    z=z,
    opacity=0.6,
    name="Hyperplane"
))

# Scatter actual data points
fig.add_trace(go.Scatter3d(
    x=x[:, 0],
    y=x[:, 1],
    z=y,
    mode='markers',
    marker=dict(size=4),
    name="Data Points"
))

# Layout
fig.update_layout(
    title="3D Linear Regression Hyperplane",
    scene=dict(
        xaxis_title="Feature 1",
        yaxis_title="Feature 2",
        zaxis_title="Target"
    )
)

fig.show()
```
