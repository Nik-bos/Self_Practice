```python
# Filling missing values with random numbers


# Dropping time column
df.drop(columns = ['Time'], inplace = True)

for column in df.columns:
    print(f'Column: {column}')

    # Getting min and max value of the column
    min_value = df[column].min()
    max_value = df[column].max()

    # Generate random numbers withing the range
    random_values = np.random.uniform(low = min_value, high = max_value, size = df[column].isnull().sum())

    # Creating a series with generated random values
    random_series = pd.Series(random_values, index = df[column] [df[column].isnull()].index)

    # Fill NaN values with random values
    df[column].fillna(random_series, inplace = True)
```

```python
df.isnull().sum()
```

```python
# Seperating features and target variable
x = df.drop(columns = ['Pass/Fail'])
y = df['Pass/Fail']
```

```python
print(df.shape)
print()

print(x.shape)
print(y.shape)
```

# Applying Logistic Regression

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
# Train test split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 1)

print(f"X_train shape: {x_train.shape}")
print(f"X_test shape: {x_test.shape}")

# Initialize and train logistic regression model
lr = LogisticRegression(max_iter = 100000)
lr.fit(x_train, y_train)

# Predicting test data
y_pred = lr.predict(x_test)

# Getting accuracy
acc = accuracy_score(y_test, y_pred)
print(f"Test Accuracy: {acc}")
```

To perform filter-based feature selection on the "UCI SECOM" dataset, which has 592 columns and a target column called "Pass/Fail," we can utilize the following methods:

**1. Duplicate Features:**
   - Identify and remove duplicate columns from the dataset. Columns with identical values provide redundant information and do not contribute to the prediction task.

**2. Variance Threshold Method:**
   - Calculate the variance of each feature.
   - Remove features with low variance, as they tend to have little or no predictive power.
   - Set a threshold value for variance and remove features below that threshold.

**3. Correlation:**
   - Compute the correlation matrix of the features.
   - Identify highly correlated features and choose one from each highly correlated group.
   - High correlation between features indicates redundancy, and removing one from each correlated group helps reduce multicollinearity.

**4. ANOVA (Analysis of Variance):**
   - Perform an ANOVA test between each feature and the target variable ("Pass/Fail").
   - Select features with a significant impact on the target variable.
   - Set a significance level (e.g., p-value threshold) for the test to determine the importance of each feature.

**5. Chi-Squared:**
   - Apply the Chi-Squared test between each feature and the target variable, considering both features as categorical.
   - Select features with a significant association with the target variable.
   - Set a significance level (e.g., p-value threshold) to determine the importance of each feature.

`Implementing these feature selection methods in Python using the "UCI SECOM" dataset can be done as follows:`

# `Step 1: `Remove Duplicate columns

```python
from feature_engine.selection import DropDuplicateFeatures

ddf = DropDuplicateFeatures()

x_train = ddf.fit_transform(x_train)
x_test = ddf.transform(x_test)

print(f"""X_train shape: {x_train.shape}
X_test shape: {x_test.shape}""")

# Columns reduced from 590 to 478
```

```python
x_train.head()
```

# 2) Variance Threshold

**Note**: Always scale all the features first else a single variance threshold will not work for all the features.

- We may have two types of features.
    - `Constant features`: 
        - When in any feature, all values are same, such features are called **constant features**.
        - Such features have 0 variance
    - `Quasi Constant Features`:
        - Suppose there are 1000 rows. In column B, 990 rows has value `No`, only 10 rows has values `Yes`. Such features are called Quassi Constant Features.
        - Such features have variance very near to 0.

As such features have very low variance (almost 0), they cannot help to differentiate target variable. So, we drop such features.

# Steps:
1) `Decide threshold`: This is to be decided based on domain knowledge. <br>
`In general, variance threshold is set between 0.01-0.1`
2) Find variance of all the features independently.
3) If feature_variance < threshold, drop that feature. 

```python
# Scaling features

from sklearn.preprocessing import StandardScaler

ss = StandardScaler().set_output(transform = 'pandas')

x_train = ss.fit_transform(x_train)
x_test = ss.transform(x_test)
```

```python
# Applying Variance threshold

from sklearn.feature_selection import VarianceThreshold

vt = VarianceThreshold(threshold = 0.05).set_output(transform = 'pandas')

x_train = vt.fit_transform(x_train)
x_test = vt.transform(x_test)
```

```python
print(f"""X_train shape: {x_train.shape}
X_test shape: {x_test.shape}""")

# Columns reduced from 478 to 474
```

# 2) Correlation

There are two approach in correlation based filtering.

    1) Between each feature.
        - Find correlation of all the columns with each other.
        - It returns a value between -1 and 1.
        - -1 means strong negative correlation and 1 means strong positive correlation
        - We decide a threshold, say 0.95, if corr between features is >0.95 or <(-0.95), we will drop any one feature from them.
        - We mostly use this approach

    2) Between a feature and target variable.
        - Finds correlation of each feature with the target variable.
        - We then decide a cut off, say 0.3 and -0.3.
        - We keep those features where corr>0.3 and corr<-0.3 (Corr between the feature and target variable)

```python
from feature_engine.selection import DropCorrelatedFeatures
```

```python
corr_matrix = x_train.corr().abs()

columns = corr_matrix.columns[
    (corr_matrix > 0.95).any()
]

print(columns)
```

```python
dcf = DropCorrelatedFeatures(threshold = 0.95)

x_train = dcf.fit_transform(x_train)
x_test = dcf.transform(x_test)

print(f"""X_train shape: {x_train.shape}
X_test shape: {x_test.shape}""")

# Columns reduced from 474 to 366
```

# 4) ANOVA

```python
from sklearn.feature_selection import f_classif
from sklearn.feature_selection import SelectKBest


anova = SelectKBest(f_classif, k = 100).set_output(transform = 'pandas')
```

```python
x_train = anova.fit_transform(x_train, y_train)
x_test = anova.transform(x_test)

print(f"""X_train shape: {x_train.shape}
X_test shape: {x_test.shape}""")

# Columns reduced from 366 to 100
```
