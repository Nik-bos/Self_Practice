```python
from sklearn.datasets import load_iris

from sklearn.tree import DecisionTreeClassifier, plot_tree,
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load iris dataset
iris = load_iris()
x = iris.data[:, :2]  # We are considering only 2 features for better understanding.
y = iris.target

#----------------------------------------------------------------------------------------------------

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 1)

# model = DecisionTreeClassifier(max_depth = 5, min_samples_split=40)
model = DecisionTreeClassifier()
model.fit(x_train, y_train)

# Predicting x_test
y_pred = model.predict(x_test)

# Check accuracy
acc = accuracy_score(y_test, y_pred)
print(f"Accuracy: {acc: .2f}")

# Plot Decision Tree
plt.figure(figsize = (12, 8))
plot_tree(
    decision_tree = model,
    filled = True, 
    feature_names = iris.feature_names[0:2],
    class_names = iris.target_names)

plt.tight_layout()
plt.show()

# Plot Decision Boundries
plt.figure(figsize=(8, 6))
x_min, x_max = x[:, 0].min() - 1, x[:, 0].max() + 1
y_min, y_max = x[:, 1].min() - 1, x[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.01),
                     np.arange(y_min, y_max, 0.01))

Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

plt.contourf(xx, yy, Z, alpha=0.8)
plt.scatter(x[:, 0], x[:, 1], c=y, edgecolor='k')
plt.xlabel('Sepal length')
plt.ylabel('Sepal width')
plt.show()

```

# Plotting Decision Tree using Dtreeviz library

```python
import graphviz.backend as gv
from dtreeviz import model
from IPython.display import Image, SVG, display_svg
```

```python
dt= DecisionTreeClassifier(max_depth = 2)
iris = load_iris()

iris
```

```python
x_train = iris.data
y_train = iris.target
dt.fit(x_train, y_train)
```

```python
iris.target_names
```

```python
# Visualization using dtreeviz

viz = model(
    model = dt,
    X_train = x_train,
    y_train = y_train,
    feature_names = iris.feature_names,
    class_names = iris.target_names,
    target_name = 'specie'
    )

viz.view()

# If we don't pass max_depth parameter, the tree will grow till the end unless each node contains rows of only one class. 
```
