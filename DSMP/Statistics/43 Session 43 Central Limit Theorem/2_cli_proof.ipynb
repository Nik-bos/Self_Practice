```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import random
```

```python
df = pd.read_csv(r"43_ipl_dataset.csv")
df.head()
```

```python
runs = df.groupby('batter').sum() ['batsman_run']
runs
```

```python
runs = runs.values

runs
```

```python
fig, axes = plt.subplots(nrows = 1, ncols = 2, figsize = (12,6))

axes[0].hist(runs, density = True, edgecolor = 'black', color= 'orange')
axes[0].set_title("Histogram of runs")

sns.histplot(runs, kde = True, color = 'magenta')
axes[1].set_title('KDE plot of runs')
```

```python
# Taking samples: We will take 1000 sets with 50 samples in each set

samples = []

for i in range(1000):
    x = np.random.choice(runs, 50)
    samples.append(x.mean())

print(f"""Samples: {samples}

Total number of means: {len(samples)}
""")

```

```python
samples = np.array(samples)

samples
```

```python
# Plotting the samples

sns.displot(samples, kde = True, color = 'orange')
```

```python
# See, mean and variance of samples will be almost equal to population mean and variance

print(f'''Sample mean: {samples.mean()}
Population mean: {runs.mean()}
---------------------------------------------
Variance of Sample mean: {samples.var()}
Variance of Population mean: {runs.var()/50}''')


# We can also verify standard deviation.
```
