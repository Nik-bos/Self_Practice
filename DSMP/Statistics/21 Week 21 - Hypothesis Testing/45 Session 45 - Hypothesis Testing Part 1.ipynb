# `Question:1`

A manufacturer claims that the average weight of their new chocolate bar is `50 grams`. We had a doubt and  decided to verify the claim.

So we `drew out 25 random samples` and measured their weight.
* The `sample mean came to be 49.7 gram`
* `Sample standard deviation (s) was 1.2 grams`

`Consider significance level 5%`.

`(Population std dev is not given, so conduct t-test)`

# `Step 1`: Determine Hypothesis

Null Hypothesis-> H0: Pop mean = 50
Alternate Hypothesis-> Ha: Pop mean != 50    (Two tailed test)

---

# `Step 2`: Verify Assumptions: Population or atleast sample should be normally distributed
Here, n<30, so central limit theorem doesn't apply, but for learning, let's assume that the sample is normally distributed

---

# Step 3: Calculate t_cal
(x_bar-mu) / (s/root_n)

Here, t_cal is (-1.25)

---

# Calculate area for t_cal -1.25 and 1.25 both as this is two tailed test. 
`Here df = n-1 = 24`

cdf = 0.11 
so, `P-value = 0.11*2 = 0.22`

`See the code`


```python
from scipy.stats import  t

# Calculate area for t value. Pass t_value and df
cdf = t.cdf(-1.25, 24)

print(cdf)

p_value = cdf*2

print(f'P_Value: {p_value}')
```

# Conclusion:

- Here, significance value is given, so conditiod `p_value > sig_level is satisfied`
- `p_value (0.22) > sig_level(0.05)`, so, we failed to reject the null hypothesis
