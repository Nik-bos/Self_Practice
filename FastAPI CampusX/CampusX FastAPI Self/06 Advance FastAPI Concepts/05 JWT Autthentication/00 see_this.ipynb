# What timedelta is

timedelta represents a duration of time.

Not a date.
Not a clock time.
But a difference between two times.

Think:

“5 minutes”, “2 days”, “3 hours and 10 seconds”, etc.

Why it exists

You use timedelta to add, subtract, or compare time.

Because you can’t safely do math on dates and times with regular numbers.

```python
from datetime import datetime, timedelta, timezone
```

```python
from datetime import datetime, timedelta

now = datetime.now()

in_10_minutes = now + timedelta(minutes=10)
yesterday = now - timedelta(days=1)

print(now)
print(in_10_minutes)
print(yesterday)
```

# Measuring how long something took.

```python
from datetime import datetime

start = datetime.now()
start
```

```python
# do some work
end = datetime.now()

duration = end - start
print(duration)    # Hours:Minutes:Seconds
```

# Extracting values from a timedelta

```python
from datetime import timedelta

delta = timedelta(hours=2, minutes=30)

print(delta.total_seconds())   # 9000
print(delta.days)              # 0
print(delta.seconds)           # 9000
```

```python
import passlib, bcrypt
print(passlib.__version__)
print(bcrypt.__version__)
```

```python
from passlib.context import CryptContext
```

```python
cc= CryptContext(schemes = ['bcrypt'], deprecated = 'auto')

# scheme = ['bcrypt'] used to hash the password
```

```python
password = 'Nikss2611'
hashed_password = cc.hash(secret = password)

print(f'''Original Password: {password}
Hashed Password: {hashed_password}''')

# .hash() converts the orig password to hashed.
```

```python
is_correct = cc.verify(secret = password, hash = hashed_password)
is_correct

# verify() returns True if hashed is for correct password else False
```

```python
from passlib.context import CryptContext

cc = CryptContext(schemes=["bcrypt"], deprecated="auto")

password = "Nikss2611"
hashed = cc.hash(password)

print("Original:", password)
print("Hashed:", hashed)

print(cc.verify(password, hashed))
```
