# 1) Synchronous

```python
import time
from timeit import default_timer as timer
```

```python
timer()
```

```python
def run_task(name, seconds):

    start_time = timer()
    time.sleep(seconds)    # This wait is for synchronous. It will execute the codes line by line. 
    print(f"{name} completed at {timer():.2f} sec.")
```

```python
start_time = timer()

# Running three tasks for 2, 1 and 3 seconds respectively
# Notice, it will take 6 seconds to complete.

run_task(name = 'Task_1', seconds = 5)
run_task(name = 'Task_2', seconds = 2)
run_task(name = 'Task_3', seconds = 3)

end_time = timer()-start_time

print(f'''
Total time taken: {timer() - start_time:.2f} seconds
''')
```

# 2) Asynchronous

```python
import asyncio
```

```python
async def run_task(name, seconds):
    print(f"{name} started at {timer()} seconds")
    await asyncio.sleep(seconds)
    print(f"{name} completed at {timer()} seconds")


async def main():
    start_time = timer()
    
    await asyncio.gather(
        run_task(name = 'Task_1', seconds = 5),
        run_task(name = 'Task_2', seconds = 2),
        run_task(name = 'Task_3', seconds = 3)
    )

    print(f"\nTotal time taken {timer() - start_time:.2f} seconds.")
```

* If it is a normal python script, use `asyncio.run(main())`
* If in notebook, use `await main()`.

```python
await main()
```

```python
a = run_task(name = 'Task_1', seconds = 5)
b = run_task(name = 'Task_2', seconds = 2)
c = run_task(name = 'Task_3', seconds = 3)

print(a)    # coroutine object
```

```python
await a
```

# What is `co-routine`?

def add(a, b):  <br>
    return a + b <br>

result = add(2, 3) <br>
➡️ the function runs immediately  <br>
➡️ result becomes 5 <br>

`Now look at an async function` <br>

async def add_async(a, b): <br>
    return a + b <br>

x = add_async(2, 3) <br>
print(x) <br>

💥 It does NOT run <br>
Instead, x is: <br>
`<coroutine object add_async> <br>
` <br>

**A coroutine is an object that represents a function <br>
that can pause and resume later** <br>

It is: <br>

NOT a value <br>

NOT running yet <br>

Just a promise of work <br>

Key rule (memorize) <br>

Calling an async def function creates a coroutine <br>
It does NOT execute it <br>

2️⃣ How does a coroutine actually run? <br>

Only one way 👇 <br>
`await coroutine` <br>
Or via <br>
`asyncio.run(main())` <br>

If you don’t await it: <br>

* It never executes <br>
* Python may warn you <br>

# 3️⃣ What is asyncio.gather()? <br>

await asyncio.gather( <br>
    run_task('Task_1', 5), <br>
    run_task('Task_2', 2), <br>
    run_task('Task_3', 3) <br>
    ) <br>

`Step A:` What is passed into gather()? <br>

run_task('Task_1', 5)  # coroutine <br>
run_task('Task_2', 2)  # coroutine <br>
run_task('Task_3', 3)  # coroutine <br>

So effectively: <br>
asyncio.gather(coroutine1, coroutine2, coroutine3) <br>

`Step B:` What does asyncio.gather() do? <br>

* It groups multiple coroutines <br>
* Runs them concurrently <br>
* Returns one combined coroutine <br>

Think of it as: <br>

🧠 “Wait for ALL of these to finish” <br>

**Visual mental model** <br>

gather() <br>
    ├── coroutine 1 <br>
    ├── coroutine 2 <br>
    └── coroutine 3 <br>

# 4️⃣ Why do we write await asyncio.gather()? <br>

Because: <br>

asyncio.gather() itself is async <br>

It returns a coroutine <br>

So: <br>

result = asyncio.gather(...) <br>

<br>
➡️ nothing runs ❌ <br>

await asyncio.gather(...) <br>

<br>
➡️ all tasks start and run concurrently ✅ <br>
