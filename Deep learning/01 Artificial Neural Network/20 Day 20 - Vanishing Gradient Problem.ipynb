```python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense, Flatten
```

```python
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()
```

```python
print(x_train.shape)
print(x_test.shape)
```

```python
x_train[0]

# These are 28*28 = 784 pixels of first image and we have 70k such images
```

```python
# y_train contains digits on corresponding image

y_train[0]

# So first image contains digit 5
```

```python
# Pixel values are always betweeen 0 and 255. Here also our images contains values between 0 and 255.
# So we will have to convert these values between 0 and 1 
# by dividing each pixel by the maximum value which is 255
```

```python
x_train = x_train/255
x_test = x_test/255
```

```python
x_train[0]
```

```python
# Let's create our model now

model = Sequential()

model.add(Flatten(input_shape = (28, 28)))
model.add(Dense(10, activation = 'sigmoid'))    # Input layer. input_shape already passed in Flatten layer
model.add(Dense(10, activation = 'sigmoid'))    # Hidden layers till second last
model.add(Dense(10, activation = 'sigmoid'))
model.add(Dense(10, activation = 'sigmoid'))
model.add(Dense(10, activation = 'sigmoid'))
model.add(Dense(10, activation = 'sigmoid'))
model.add(Dense(10, activation = 'sigmoid'))
model.add(Dense(10, activation = 'sigmoid'))
model.add(Dense(10, activation = 'sigmoid'))    # Hidden layers end
model.add(Dense(10, activation = 'sigmoid'))    # Output layer

# Here we have 1 input, 1 output and between them 8 hidden layers.

# VGP is mostly found with Sigmoid and hyperbolic tangent(tanh) activation function.
```

```python
model.compile(loss = 'sparse_categorical_crossentropy', optimizer = 'Adam', metrics = ['accuracy'])

# We can also use los = 'categorical_crossentropy' but here we're using 'sparse_categorical_crossentropy'
# bcz in categorical_ce we have to OneHot encode our labels which is 0 to 9 but it is avoidable in
# Sparse_categorical_ce.
```

```python
# Initial weights and biases befor training

old_weights = model.get_weights()[0]
old_weights
```

```python
# Training the model with 1 epoch

model.fit(x_train, y_train, epochs = 1)
```

```python
new_weights = model.get_weights()[0]
new_weights
```

```python
# Let's find the gradient.
# Formula is (old weights - new weights) / learning rate

lrng_rate = model.optimizer.get_config()['learning_rate']
lrng_rate
```

# What is a Gradient?

In simple terms, a **gradient** is a measure of how much a function changes if you change its inputs slightly. In the context of machine learning, the gradient shows how much the model’s output (or error) will change if we adjust its weights.

Imagine you’re hiking on a mountain, and your goal is to reach the lowest point in a valley. The gradient at your current spot tells you the direction of the steepest downward slope—basically, the quickest way to go down. In machine learning, we use this “downhill direction” (the gradient) to adjust the model's weights in a way that minimizes the error, helping the model make better predictions.

Each step down the slope represents a small improvement in the model. The gradient tells us both the **direction** (which way to go) and **magnitude** (how big of a step to take). The learning rate then controls the size of the steps, making sure we don’t take steps that are too big (risking overshooting) or too small (making learning slow).

# 4th way: Batch Normalizzation. Just take an idea here, we will learn it properly later.
**Batch normalization** is a special layer often added to neural networks to help them learn faster and perform better.

Here’s the main idea in simple terms:

When training a neural network, the data moving through each layer can sometimes “shift” or become unevenly scaled, which slows down learning. Batch normalization fixes this by making sure that the inputs to each layer have a stable mean (average) and variance (spread) during training.

In batch normalization:

`Normalizing`: For each mini-batch (a small subset of training data), it calculates the mean and variance of the data. It then adjusts (or "normalizes") the values to keep them more balanced. This helps prevent the values from getting too extreme.

`Scaling and Shifting`: After normalizing, it adds two learnable parameters (scale and shift) so that the network can still learn the best range for those values, instead of sticking to a strict normalized scale.

Batch normalization is a layer that is typically placed right after the linear operations in a network, before or after the activation function. It has several benefits:

* It helps stabilize and accelerate training by reducing internal shifts in data.
* It allows for higher learning rates, which means faster training.
* It can also reduce the chances of overfitting, making the model more generalizable.
