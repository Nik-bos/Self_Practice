```python
import random

# Step 1: Create a HyperParameter class (like 'hp' in Keras Tuner)
class HyperParameter:
    def __init__(self):
        self.params = {}

    def Choice(self, name, values):
        self.params[name] = random.choice(values)  # Randomly pick one value
        return self.params[name]

# Step 2: Create a Custom Tuner class
class CustomTuner:
    def __init__(self, function, trials=3):
        self.function = function  # Function to optimize (like 'build_model')
        self.trials = trials  # Number of times to try different values
        self.best_result = None
        self.best_params = None

    def search(self):
        for i in range(self.trials):
            hp = HyperParameter()  # Create a new hp object
            result = self.function(hp)  # Call the function with hp
            
            # Store the best result
            if self.best_result is None or result > self.best_result:
                self.best_result = result
                self.best_params = hp.params

        print("\nBest Parameters:", self.best_params)
        print("Best Result:", self.best_result)

# Step 3: Define a function that uses 'hp' (like 'build_model')
def math_operation(hp):
    num1, num2 = 10, 5

    # Let hp choose between addition and subtraction
    operation = hp.Choice('operation', ['add', 'subtract'])

    result = num1 + num2 if operation == 'add' else num1 - num2
    print(f"Trial with {operation}: Result = {result}")
    
    return result  # Returning the result (like validation accuracy)

# Step 4: Run the tuner (just like Keras Tuner)
tuner = CustomTuner(math_operation, trials=5)
tuner.search()
```
