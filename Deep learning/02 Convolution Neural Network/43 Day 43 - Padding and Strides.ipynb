# What is padding?

`Padding` is applied to the image so that the output shape of the image be same as input shape.

`valid`: no padding to apply <br>
`same`: keras automatically decides how much padding to use such that the output shape of the image be same as input shape

```python
import tensorflow 
from tensorflow import keras
from keras.layers import Dense, Conv2D, Flatten
from keras.models import Sequential
from keras.datasets import mnist
```

```python
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

```python
x_train[0].shape
```

# Model without padding

```python
model = Sequential()

model.add(Conv2D(32, kernel_size = (3,3), padding = 'valid', activation = 'relu', input_shape = (28,28,1)))
model.add(Conv2D(32, kernel_size = (3,3), padding = 'valid', activation = 'relu', input_shape = (28,28,1)))
model.add(Conv2D(32, kernel_size = (3,3), padding = 'valid', activation = 'relu', input_shape = (28,28,1)))

# 32 means 32 filters. Like in dense layer we have nodes, here we have filters that slides over the image
# kernal_size (3,3) means filter size of 3 rows and 3 columns
# padding = 'valid' (by default) means no padding

model.add(Flatten())

model.add(Dense(128, activation = 'relu'))
model.add(Dense(10, activation = 'softmax'))

model.summary()
```

We can see, our original image size was (28,28), which changed to (26,26) after 1st conv2d, (24,24) after 2nd conv2d
and (22,22) after 3rd conv2d. (26,26,32) where 32 is the no_of filters applied.

Flatten layer converts the data into 1_D

# Model with padding.

```python
model = Sequential()

model.add(Conv2D(32, kernel_size = (3,3), padding = 'same', activation = 'relu', input_shape = (28,28,1)))
model.add(Conv2D(32, kernel_size = (3,3), padding = 'same', activation = 'relu', input_shape = (28,28,1)))
model.add(Conv2D(32, kernel_size = (3,3), padding = 'same', activation = 'relu', input_shape = (28,28,1)))

# 32 means 32 filters. Like in dense layer we have nodes, here we have filters that slides over the image
# kernal_size (3,3) means filter size of 3 rows and 3 columns
# padding = 'same' means apply padding

model.add(Flatten())

model.add(Dense(128, activation = 'relu'))
model.add(Dense(10, activation = 'softmax'))

model.summary()
```

# How to apply strides?

```python
model = Sequential()

model.add(Conv2D(32, kernel_size = (3,3), padding = 'same', strides = (2,2), activation = 'relu', input_shape = (28,28,1)))
model.add(Conv2D(32, kernel_size = (3,3), padding = 'same', strides = (2,2), activation = 'relu', input_shape = (28,28,1)))
model.add(Conv2D(32, kernel_size = (3,3), padding = 'same', strides = (2,2), activation = 'relu', input_shape = (28,28,1)))

# 32 means 32 filters. Like in dense layer we have nodes, here we have filters that slides over the image
# kernal_size (3,3) means filter size of 3 rows and 3 columns
# padding = 'same' means apply padding
# strides = (2,2) means filter will move 2 steps left and after that 2 steps bottom
# We can also keep stride = (2,1) for 2 steps horizontally and after that 1 step vertically

model.add(Flatten())

model.add(Dense(128, activation = 'relu'))
model.add(Dense(10, activation = 'softmax'))

model.summary()
```
