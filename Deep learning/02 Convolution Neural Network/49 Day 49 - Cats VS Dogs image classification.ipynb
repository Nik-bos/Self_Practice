```python
import pandas as pd
import numpy as np

import tensorflow as tf
from tensorflow import keras
from keras.models import Sequential
from keras.layers import Dense, Conv2D, Flatten, MaxPooling2D
```

```python
train_images_path = r'D:\Nikss\Self Pracitce\Deep learning\02 Convolution Neural Network\49 Day 49 - Cats_VS_Dogs\train'

test_images_path = r'D:\Nikss\Self Pracitce\Deep learning\02 Convolution Neural Network\49 Day 49 - Cats_VS_Dogs\test'
```

```python
train_data = keras.utils.image_dataset_from_directory(
    directory = train_images_path,
    labels = 'inferred',
    label_mode = 'int',    # Will assign 0 for cats and 1 for dogs
    batch_size = 32,
    image_size = (256,256)    # Will convert all images to the size of (256,256) irrespective of the size of original image
    
)

###################################################################################################################################

validation_data = keras.utils.image_dataset_from_directory(
    directory = test_images_path,
    labels = 'inferred',
    label_mode = 'int',    # Will assign 0 for cats and 1 for dogs
    batch_size = 32,
    image_size = (256,256)    # Will convert all images to the size of (256,256) irrespective of the size of original image
    
)
```

```python
train_data.class_names
```

```python
for images, labels in train_data.take(1):    # take(1) selects 1 batch
    print(images.shape)

# See, image shape is same (256, 256) which we set earlier
# 32 shows batch size. Means 32 images in one batch. This is also same as we set earlier
```

```python
# Normalize

def normalize(image, label):

    image = tf.cast(image/255, tf.float32)
    return image, label
```

```python
train_data = train_data.map(normalize)
validation_data = validation_data.map(normalize)
```

# Creating model

```python
model_1 = Sequential()

model_1.add(Conv2D(filters = 32, kernel_size = (3,3), padding = 'valid', activation = 'relu', input_shape = (256,256,3)))
model_1.add(MaxPooling2D(pool_size = (2,2), padding = 'valid', strides = 2))

model_1.add(Conv2D(filters = 64, kernel_size = (3,3), padding = 'valid', activation = 'relu'))
model_1.add(MaxPooling2D(pool_size = (2,2), padding = 'valid', strides = 2))

model_1.add(Conv2D(filters = 128, kernel_size = (3,3), padding = 'valid', activation = 'relu'))
model_1.add(MaxPooling2D(pool_size = (2,2), padding = 'valid', strides = 2))

model_1.add(Flatten())

model_1.add(Dense(128, activation = 'relu'))
model_1.add(Dense(64, activation = 'relu'))
model_1.add(Dense(1, activation = 'sigmoid'))

model_1.summary()
```

```python
model_1.compile(loss = 'binary_crossentropy', optimizer = 'adam', metrics = ['accuracy'])
```

```python
history_1 = model_1.fit(train_data, epochs = 10, validation_data = validation_data)
```
