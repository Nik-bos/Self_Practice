# Day 64- What is Gated Recurrent Network?

The purpose of GRU is same as of LSTM but the only difference is, GRU's architecture is somewhat simpler than LSTM. Both is used to achieve the same objective i.e to remember long sequential information, which is not done by simple RNNs. But as GRUs are simpler than LSTM, it has less no_of parameters to train.

# This session (Day 65) - What is Deep RNN?
When we use more than one RNN layer, it is called `deep RNN` or `stacked RNN`. 

# `Note`: Mostly in real world we don't use deep RNN, istead we use deep LSTM or deep GRU Let's see the code.
# `Let's see the code.`

```python
import tensorflow as tf
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense, SimpleRNN, Embedding, LSTM, GRU

from keras.datasets import imdb
```

```python
# Load the dataset

(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words = 10000)

# num_words = 10000 means take top 10000 words with hightest frequency
# Means it creates a dictionary of top 10000 words with highest frequency
```

```python
x_train.shape
```

```python
# Here, data is already index encoded but padding is not applied, so, we will do it.
```

```python
# Padding

from keras.preprocessing.sequence import pad_sequences
```

```python
x_train_padded = pad_sequences(x_train, maxlen = 100)
x_test_padded = pad_sequences(x_test, maxlen = 100)
```

# 1) Deep RNN

```python
deep_rnn = Sequential([
    Embedding(input_dim = 10000, output_dim = 32, input_length = 100),
    SimpleRNN(units = 10, return_sequences = True),
    SimpleRNN(units = 5, return_sequences = True),
    SimpleRNN(units = 5),

    Dense(1, activation = 'sigmoid')
])

# Embedding(input_dim = 10000, output_dim = 32, input_length = 100)
# input_dim = 10000. takes no_of words in the vocab.Here we have 10000. Remember while loading the data...
# output_dim = 32, means for each word, we want vectors of 32 integers
# input_len = 100 is the length of each sentence in the data. While padding, we limited it to 100

# SimpleRNN(units = 10, return_sequences = True),
# units is nothing but nodes in dense layers
# return_sequences = True means pass this output to next RNN

# Note: We dont assing True to the last RNN layer

deep_rnn.compile(loss = 'binary_crossentropy',
                optimizer = 'adam',
                metrics = ['accuracy'])
```

```python
deep_rnn.summary()
```

```python
deep_rnn.fit(x_train_padded, y_train,
            validation_split = 0.2,
            epochs = 10 )
```

# 2) Deep LSTMs

```python
deep_lstm = Sequential()
deep_lstm.add(Embedding(input_dim = 10000, output_dim = 32, input_length = 100))

deep_lstm.add(LSTM(units = 10, return_sequences = True))
deep_lstm.add(LSTM(units = 5, return_sequences = True))
deep_lstm.add(LSTM(units = 5))

deep_lstm.add(Dense(1, activation = 'sigmoid'))






# Embedding(input_dim = 10000, output_dim = 32, input_length = 100)
# input_dim = 10000. takes no_of words in the vocab.Here we have 10000. Remember while loading the data...
# output_dim = 32, means for each word, we want vectors of 32 integers
# input_len = 100 is the length of each sentence in the data. While padding, we limited it to 100

# LSTM(units = 10, return_sequences = True),
# units is nothing but nodes in dense layers
# return_sequences = True means pass this output to next RNN

# Note: We dont assing True to the last RNN layer

deep_lstm.compile(loss = 'binary_crossentropy',
                optimizer = 'adam',
                metrics = ['accuracy'])
```

```python
deep_lstm.summary()
```

```python
deep_lstm.fit(x_train_padded, y_train,
            validation_split = 0.2,
            epochs = 10 )
```

# 3) Deep Gated Recurrent Units (GRUs)

```python
deep_gru = Sequential([
    Embedding(input_dim = 10000, output_dim = 32, input_length = 100),

    GRU(units = 10, return_sequences = True),
    GRU(units = 5, return_sequences = True),
    GRU(units = 5),

    Dense(units = 1, activation = 'sigmoid')
])







# Embedding(input_dim = 10000, output_dim = 32, input_length = 100)
# input_dim = 10000. takes no_of words in the vocab.Here we have 10000. Remember while loading the data...
# output_dim = 32, means for each word, we want vectors of 32 integers
# input_len = 100 is the length of each sentence in the data. While padding, we limited it to 100

# Gru(units = 10, return_sequences = True),
# units is nothing but nodes in dense layers
# return_sequences = True means pass this output to next RNN

# Note: We dont assing True to the last RNN layer

deep_gru.compile(loss = 'binary_crossentropy',
                optimizer = 'adam',
                metrics = ['accuracy'])
```

```python
deep_gru.summary()
```

```python
deep_gru.fit(x_train_padded, y_train,
            validation_split = 0.2,
            epochs = 10)
```
