# Building encoder and decoder

```python
from keras.models import Model
from keras.layers import Input, Embedding, LSTM, Dense, Dropout
```

```python
# Encoder

encoder_input = Input(shape = 56)    # 56 is our input sequence length (x). See x.shape

embedding_output = Embedding(input_dim = total_words+1, output_dim = 100)  (encoder_input)

#LSTM layer

encoder_lstm_output, state_h, state_c = LSTM(units = 128, activation = 'tanh', return_state = True)   (embedding_output)

encoder_states = [state_h, state_c]

# Encoder state is the context vector we heard about, in the theory.
```

```python
# Decoder

decoder_input = Input(shape = (1, ))
decoder_embedding_output = Embedding(input_dim = total_words +1, output_dim = 100)   (decoder_input)

# Decoder LSTM
decoder_lstm_output = LSTM(units = 128, activation = 'tanh')   (decoder_embedding_output, initial_state = encoder_states)

final_output = Dense(total_words+1, activation = 'softmax')    (decoder_lstm_output)
```

```python
model = Model(inputs = [encoder_input, decoder_input], outputs = final_output)

model.compile(loss = 'categorical_crossentropy', optimizer = 'adam', metrics = ['accuracy'])
model.summary()
```

```python
def build_model(hp):
    # Tune LSTM units
    lstm_units = hp.Choice("lstm_units", values=[32, 64, 128])

    # Encoder
    encoder_input = Input(shape=(56,))
    embedding_out = Embedding(input_dim=total_words + 1, output_dim=100)(encoder_input)
    
    x = LSTM(lstm_units, return_sequences=True)(embedding_out)

    if hp.Boolean("use_encoder_dropout"):
        x = Dropout(hp.Float("encoder_dropout_rate", 0.2, 0.5, step=0.1))(x)

    encoder_output, state_h, state_c = LSTM(
        lstm_units,
        return_state=True,
        dropout=hp.Float("encoder_recurrent_dropout", 0.0, 0.5, step=0.1)
    )(x)

    encoder_states = [state_h, state_c]

    # Decoder
    decoder_input = Input(shape=(1,))
    decoder_emb = Embedding(input_dim=total_words + 1, output_dim=100)(decoder_input)

    decoder_output = LSTM(
        lstm_units,
        dropout=hp.Float("decoder_recurrent_dropout", 0.0, 0.5, step=0.1)
    )(decoder_emb, initial_state=encoder_states)

    if hp.Boolean("use_decoder_dropout"):
        decoder_output = Dropout(hp.Float("decoder_dropout_rate", 0.2, 0.5, step=0.1))(decoder_output)

    if hp.Boolean("use_batchnorm"):
        decoder_output = BatchNormalization()(decoder_output)

    final_output = Dense(
        total_words + 1,
        activation="softmax",
        kernel_initializer=hp.Choice("kernel_initializer", ["glorot_uniform", "he_normal"]),
        kernel_regularizer=l2(0.001)
    )(decoder_output)

    model = Model([encoder_input, decoder_input], final_output)
    model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
    return model
```
