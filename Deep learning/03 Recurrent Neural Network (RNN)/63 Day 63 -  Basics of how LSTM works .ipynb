```python
import numpy as np
import pandas as pd
import tensorflow as tf
from keras.models import Sequential
from keras.layers import LSTM, Dense, Embedding
from keras.preprocessing.text import Tokenizer
```

```python
l = 'My name is Nikunj'
```

```python
tokenizer = Tokenizer()

tokenizer.fit_on_texts([l])
```

```python
tokenizer.word_counts
```

```python
tokenizer.word_index
```

```python
seq = tokenizer.texts_to_sequences([l]) [0]
seq
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
padded_seq = pad_sequences(sequences = [seq], maxlen = len(tokenizer.word_index), padding = 'pre')
```

```python
padded_seq.shape[1]
```

```python
model = Sequential()
```

```python
embedding = Embedding(input_dim = 5, output_dim = 8, input_length = 4)  (padded_seq)
```

```python
embedding
```

```python
lstm = LSTM(units = 3, activation = 'tanh', return_state= True)  (padded_seq)
```

```python
# LSTM layer expects 3D input but our padded_seq is 2d

padded_seq.ndim
```

```python
# Converting to 3d

padded_seq = np.expand_dims(padded_seq, axis = -1)
```

```python
padded_seq.shape
```

```python
padded_seq = padded_seq.astype('float32')
```

```python
# lstm_output, state_h, state_c = LSTM(units = 3, activation = 'tanh', return_state= True)  (padded_seq)

# return_sequence = False(By-default)  This returns state_h  of last time step onle
# If yes, returns state_h of all time steps


lstm_output, state_h, state_c = LSTM(units = 3, activation = 'tanh', return_sequences = True, return_state= True)  (padded_seq)
```

```python
# lstm_output, state_h, state_c = LSTM(units = 3, activation = 'tanh', return_state= True)  (embedding)
```

```python
x = [lstm_output, state_h, state_c]

for i in x:
    print(i)
    print()
    print()

# See, we are getting 3 outputs bcz we passed 3 units in LSTM
```
