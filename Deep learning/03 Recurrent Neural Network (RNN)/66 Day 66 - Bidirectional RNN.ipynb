```python
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense, Embedding, SimpleRNN, LSTM, GRU, Bidirectional

from keras.datasets import imdb
```

```python
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words = 10000)

# Here, data is tokenized but padding is not done. Each sentence is of different size.
# So we will have to apply padding and make them of same siz

from tensorflow.keras.utils import pad_sequences
```

```python
x_train_padded = pad_sequences(x_train, maxlen = 100, padding = 'pre', truncating = 'pre')
x_test_padded = pad_sequences(x_train, maxlen = 100, padding = 'pre', truncating = 'pre')
```

# 1) Bi-directional RNN

```python
model_rnn = Sequential([
    Embedding(input_dim = 10000, output_dim = 32, input_length = 100),
    Bidirectional(SimpleRNN(units = 5)),
    Dense(units = 1, activation = 'sigmoid')
])

model_rnn.compile(loss = 'binary_crossentropy',
                 optimizer = 'adam',
                 metrics = ['accuracy'])
```

```python
model_rnn.fit(x_train_padded, y_train,
         validation_split = 0.2,
         epochs = 5)
```

# 2) Bi-directional LSTM `(BiLSTM)`

```python
model_lstm = Sequential([
    Embedding(input_dim = 10000, output_dim = 32, input_length = 100),
    Bidirectional(LSTM(5)),
    Dense(1, activation = 'sigmoid')
])

model_lstm.compile(loss = 'binary_crossentropy',
                  optimizer = 'adam',
                  metrics = ['accuracy'])
```

```python
model_lstm.fit(x_train_padded, y_train,
         validation_split = 0.2,
         epochs = 5)
```

# Bi-directional GRU `(BiGRU)`

```python
model_gru = Sequential([
    Embedding(input_dim = 10000, output_dim = 32, input_length = 100),
    Bidirectional(GRU(5, return_sequences = True)),
    Bidirectional(GRU(5)),    # Stacked bi-directional Grus
    Dense(1, activation = 'sigmoid')
])

model_gru.compile(loss = 'binary_crossentropy',
                  optimizer = 'adam',
                  metrics = ['accuracy'])
```

```python
model_gru.fit(x_train_padded, y_train,
             validation_split = 0.2,
             epochs = 5)
```

# `Note`: Bi-directional RNN is mostly not used in realtime, instead mostly `BiLSTM` and `BiGRU` are used.
