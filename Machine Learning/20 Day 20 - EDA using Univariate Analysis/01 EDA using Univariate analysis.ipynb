# What is `Univariate analysis`?

When we are analysing a single variable/column from various available columns, it is called as **Univariate analysis**

While performing the Univariate analysis, the very first question we should ask is, **"What's the data type of the column?""**

The values of any column can only be one of the two types.
1) Qualitative(Categorica)
2) Quantative(Numerical)

We need to performe analysis based on the types of data a column is holding.

Here, we will perform Univariate analysis on famous **"Titanic dataset**", where we predict that whether a passenger will survive or not.

Here, we'll first analysis the columns with categorical data and then the ones with numeric data, by ploting.

```python
import pandas as pd
```

```python
df = pd.read_csv("19 titanic.csv")
df.sample(10)
```

# `Ploting columns with categorical data.`

a) Countplot<br>
b) Bar chart<br>
c) Pie chart<br>

# a) Countplot.

```python
# Here, 'Survived' column has values as 0 or 1 and so we can treat it as categorical column.
# Let's plot and identify how many people survived and how many died.
```

```python
import seaborn as sns
```

```python
# Using seaborn library.

sns.countplot(x = 'Survived', data = df, color = 'turquoise')    # We can give only one color.

# Try with different categorical columns like 'Pclass', "Sex", 'Embarked' etc.
```

```python
# Getting the same result using plot() of DataFrame object.

a = df['Embarked'].value_counts()
print(a)

print()
print()


a.plot(kind = "bar", color = ['orange', 'turquoise', 'green'])    # Can give as many colors as no.of bins.
```

# b) Pie chart

In the above charts we received the values in numbers. But if we want the values in percentage, use 'Pie chart'.

```python
a = df['Pclass'].value_counts()
a
```

```python
a.plot(kind = 'pie')

# We can see that we are getting a pie chart but not the percentages. So to get the percentages 'autopct' parameter is used.
```

```python
a.plot(kind = 'pie', autopct = '%.2f')

# We cam see that we're getting percentage as well.
# 'autopct' parameter is only valid for pie chart.
```

b) Dist plot<br>

# a) Histogram.

When our data are continuous, like age, we use Histogram; where x-axis represents 'range' and y-axis represents 'count'.

```python
import matplotlib.pyplot as plt
```

```python
plt.hist(df['Age'], bins = 5)
```

```python
sns.histplot(df['Age'], bins = 5)
```

# b) Distplot

```python
sns.distplot(df['Age'])
```
