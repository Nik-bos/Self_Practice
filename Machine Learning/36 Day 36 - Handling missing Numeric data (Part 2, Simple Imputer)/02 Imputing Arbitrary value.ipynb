```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
```

```python
df = pd.read_csv("00 titanic.csv")
```

```python
df.sample(10)
```

```python
df.isnull().sum()
```

```python
x = df.iloc[:, 0:3]
y = df.drop(x, axis = 'columns')
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2)
```

```python
trans = [
    ('Age', SimpleImputer(strategy = 'constant', fill_value = 9), [0]),
    ('Fare', SimpleImputer(strategy = 'constant', fill_value = 99,), [1])
]
```

```python
ct = ColumnTransformer(transformers = trans, remainder = 'passthrough')
```

```python
x_train_trans = ct.fit_transform(x_train)
x_test_trans = ct.fit_transform(x_test)
```
