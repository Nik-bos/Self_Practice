# Creating Column Transformers

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.impute import KNNImputer
from sklearn.preprocessing import StandardScaler
```

```python
# Imputing missing values on 'bmi' using KNNImputer

data_imputer = [
    ('Impute_bmi', KNNImputer(), [8])
]

trans_imputer = ColumnTransformer(transformers = data_imputer, remainder = 'passthrough')
```

```python
# Applying OneHotEncoder

data_ohe = [
    ("OHE", OneHotEncoder(sparse = False, handle_unknown = 'ignore'), [0, 4, 5, 6, 9])
]

trans_ohe = ColumnTransformer(transformers = data_ohe, remainder = 'passthrough')
```

```python
# Applying Standard Scaler

data_std_scaler = [
    ("StandardScaling", StandardScaler(), slice(0, 10))
]

trans_scaler = ColumnTransformer(transformers = data_std_scaler)
```

```python
from sklearn.tree import DecisionTreeClassifier

trans_dt = DecisionTreeClassifier()
```

# Making Pipeline

```python
from sklearn.pipeline import Pipeline
```

```python
pipe = Pipeline([
    ('Trans1', trans_imputer),
    ('Trans2', trans_ohe),
    ('Trans3', trans_scaler),
    ('Trans4', trans_dt)
])
```

# Train and make predictions

```python
pipe.fit(x_train, y_train)
```

```python
# Making prediction

pred = pipe.predict(x_test)

# Here it is predicting only 0 bcz the class variable is imbalaced. Class 0 hase 4861 rows and 1 has only 249 rows.
```

```python
from sklearn.metrics import accuracy_score\

accuracy = accuracy_score(pred, y_test)
accuracy
```
