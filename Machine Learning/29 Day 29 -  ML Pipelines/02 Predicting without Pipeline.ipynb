```python
import pickle
import numpy as np
```

```python
ohe_sex = pickle.load(open(r'D:\Nikss\Self Pracitce\Machine Learning\models\ohe_sex.pkl', 'rb'))
ohe_embarked = pickle.load(open(r'D:\Nikss\Self Pracitce\Machine Learning\models\ohe_embarked.pkl', 'rb'))

dt = pickle.load(open(r'D:\Nikss\Self Pracitce\Machine Learning\models\dt.pkl', 'rb'))
```

```python
# User input
# Pclass/ sex/ age/ SibSp/ Parch/ Fare/ Embarked

user_input = np.array([2, 'male', 31.0, 0, 0, 10.5, 'S'],dtype=object).reshape(1,7)    # reshape(1, 7) means 1 row and 7 columns.

print(user_input)
print(user_input.ndim)
```

```python
sex = user_input[:,1]
print(sex)
print(sex.ndim)

# Sex is in 1D but ohe expects 2D so we'll convert it to 2D using 'reshape()' while applying ohe
```

```python
user_input_sex = ohe_sex.transform(user_input[:, 1].reshape(1, 1))    # reshape(1, 1) means 1 row and 1 column and also converts from 1D to 2D
# bcz ohe expects 2D input.

print(user_input_sex)
```

```python
embarked = user_input[:, 6]
print(embarked)
print(embarked.ndim)

# Embarked is in 1D but ohe expects 2D so we'll convert it to 2D using 'reshape()' while applying ohe
```

```python
user_input_embarked = ohe_embarked.transform(embarked.reshape(1, 1))

print(user_input_embarked)
```

```python
# Concatenate

user_input_transformed = np.concatenate((user_input[:,[0,2,3,4,5]],user_input_sex, user_input_embarked),axis=1)
user_input_transformed
```

```python
user_input_transformed.shape
```

```python
# Prediction

dt.predict(user_input_transformed)
```
