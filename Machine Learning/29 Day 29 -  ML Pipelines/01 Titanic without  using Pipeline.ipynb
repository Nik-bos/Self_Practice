# What are pipelines?

* Pipelines chains together multiple tasks so that the output of each task is used as an imput to the next task
* Pipelines makes it easy to apply same preprocessing to train and test.
* It's ok if not understand clearly, we will work with and without pipelines both and see the difference.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("29 titanic.csv")
```

```python
df.sample(10)
```

```python
df.isnull().sum()
```

```python
# Our aim is not model building but to understand the concept. So we are droping some unnecessary columns

df.drop(columns = ['PassengerId', 'Name', 'Cabin', "Ticket"], axis = 'columns', inplace = True)
```

```python
df.head()
```

```python
# Splitting into 'x' and 'y'

x = df.iloc[:, 1:8]
y = df.drop(x, axis = 'columns')

print(x.head())
print()
print()
print(y.head())
```

```python
# Train test split.

from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2)


print(x_train.isnull().sum())
```

```python
# Applying imputatiion

from sklearn.impute import SimpleImputer
```

```python
si_age = SimpleImputer()    # Imputing with mean(By default)
si_embarked = SimpleImputer(strategy = 'most_frequent')    # Imputing most frequent repeting element.
```

```python
x_train_age = si_age.fit_transform(x_train[['Age']])
x_train_embarked = si_embarked.fit_transform(x_train[['Embarked']])
```

```python
x_test_age = si_age.transform(x_test[['Age']])
x_test_embarked = si_embarked.transform(x_test[['Embarked']])
```

```python
print(x_train_age[0:4])
print()
print()
print()
print(x_train_embarked[0:4])
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe_sex = OneHotEncoder(sparse = False, handle_unknown = 'ignore')
ohe_embarked = OneHotEncoder(sparse = False, handle_unknown = 'ignore')
```

```python
x_train_sex = ohe_sex.fit_transform(x_train[['Sex']])
x_train_embarked = ohe_embarked.fit_transform(x_train_embarked)
```

```python
x_test_sex = ohe_sex.fit_transform(x_test[['Sex']])
x_test_embarked = ohe_embarked.fit_transform(x_test_embarked)
```

```python
print(x_train_embarked[0:4])

print()
print()
print()

print(x_test_embarked[0:4])

print()
print()
print()

print(x_train_sex[0:4])

print()
print()
print()

print(x_test_sex[0:4])
```

# Concetenation

```python
# Extraxting extra columns

x_train_rem = x_train.drop(columns = ['Age', 'Sex', 'Embarked'])
x_test_rem = x_test.drop(columns = ['Age', "Sex", 'Embarked'])
```

```python
x_train_rem
```

```python
x_train_trans = np.concatenate((x_train_rem, x_train_age, x_train_sex, x_train_embarked), axis = 1)
x_test_trans = np.concatenate((x_test_rem, x_test_age, x_test_sex, x_test_embarked), axis = 1)
```

```python
x_train_trans[1]
```

```python
x_test_trans[1]
```

```python
y_train.head(2)
```

```python
y_test.head(6)
```

# Model building

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dt = DecisionTreeClassifier()
```

```python
dt.fit(x_train_trans, y_train)
```

```python
x_test_trans.shape
```

```python
y_test.shape
```

```python
pred = dt.predict(x_test_trans)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy = accuracy_score(pred, y_test)
print(accuracy)
```
