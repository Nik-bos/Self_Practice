```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
df = pd.read_csv("00 startups.csv", usecols = ['R&D Spend', 'Administration', 'Marketing Spend', 'Profit'])

df = np.round(df/10000)
df = df.sample(5)
df
```

```python
# Seperating Target variable

df = df.iloc[:, 0:-1]
```

# Function `iterative_imputation()`

```python
def iterative_imputation(df, iterations):

    # Give an initial DataFrame without performing anything.
    
    df.iloc[1, 0] = np.nan
    df.iloc[3, 1] = np.nan
    df.iloc[-1, -1] = np.nan

    # Step 1: Impute all missing values with the mean of respective columns.
    mean_rd = df['R&D Spend'].mean()
    mean_add = df['Administration'].mean()
    mean_market = df['Marketing Spend'].mean()

    
    df['R&D Spend'] = df['R&D Spend'].fillna(mean_rd)
    df['Administration'] = df['Administration'].fillna(mean_add)
    df['Marketing Spend'] = df['Marketing Spend'].fillna(mean_market)

    # df_1 = df.copy()    Remove comment from this and give comment to the same code inside loop, gives same result.
    
    for i in range(iterations):
        df_1 = df.copy()
        
        # Converting R*D Spend into np.nan.
        df_1.iloc[1, 0] = np.nan
        x_1 = df_1.iloc[[0, 2, 3, 4], [1,2]]    # 'Administration' and 'Marketing Spend' as predictor variable
        y_1 = df_1.iloc[[0, 2, 3, 4], 0]      # 'R&D Spend' as target variable.
        pred_x_1 = df_1.iloc[1, 1:3].values.reshape(1, 2)    # Missing value to be predicted.
        
        lr.fit(x_1, y_1)

        # Making predicting.
        pred_value = lr.predict(pred_x_1)

        # Filling missing value with the predicted value.
        df_1.iloc[1, 0] = pred_value.round(2)


        # Converting 'Administration's column as np.nan.
        df_1.iloc[-2, -2] = np.nan

        # Spliting into x and y.
        x_1 = df_1.iloc[[0, 1, 2, 4], [0, 2]]    # 'R*D Spend' and 'Marketing Spend' as Predictor variable and
        y_1 = df_1.iloc[[0, 1, 2, 4], 1]         # 'Administration' as class variable

        pred_x_1 = df_1.iloc[3, [0, 2]]          # Missing value to be predicted.

        # Training the model
        lr.fit(x_1, y_1)

        # Prediction
        pred_value = lr.predict(pred_x_1.values.reshape(1, 2))

        # Filling missing value
        df_1.iloc[-2,  -2] = pred_value.round(2)

        # Changing 'Marketing Spend's column as np.nan
        df_1.iloc[-1, -1] = np.nan

        # Spliting into x and y
        
        x_1 = df_1.iloc[[0, 1, 2, 3], [0, 1]]    # 'R*D Spend ' and 'Administration' as predictor variable and
        y_1 = df_1.iloc[[0, 1, 2, 3], -1]        # 'Marketing Spend' as class varibale
        pred_x_1 = df_1.iloc[-1, [0, 1]].values.reshape(1, 2)    # Missing value to be predicted.

        lr.fit(x_1, y_1)
        
        # Making prediction.
        pred_value = lr.predict(pred_x_1)

        # Filling missing value
        df_1.iloc[-1, -1] = pred_value.round(2)

        print(df_1 - df)

        df = df_1.copy()
```

```python
iterative_imputation(df, 6)
```
