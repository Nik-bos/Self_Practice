# 9) error_bad_lines parameter.

* It may happen sometimes that there are for e.g 10 columns in our dataset. But some 10-15 rows in between may have 11 columns. In this case, we will get an error message and we won't be able to load our data in pandas DataFrame.
* error_bad_lines parameter skips those rows and loads the other rows.  
* So to avoid that error message and to load our data, set **error_bad_lines** parameter as False.

# 10) dtypes paraeter.

* Python automatically set the data types of the columns based on the types of data inserted in it.
* But if we want to change that automatically set datatype, we can do it using **dtype** parameter.

* For e.g: Here in our dataset named "aug_train.csv" there's a column named 'target'. Its data type is float, but the values are 0 and 1. But as its data type is float, it simply takes more memory. So we can change its dtype to 'int' and save the memory.
* dtypes takes a dictionary as an input where columns' names are keys and desired data types are values.

```python
df9 = pd.read_csv("aug_train.csv")
df9.info()

# See target columns's data tyoe is float. So let's change it to int.
```

```python
df9 = pd.read_csv("aug_train.csv", dtype = {'target': int})
df9.info()

# See, now target column's data type is integer.
```

# 12) converters.

* converter parameter is used to transform the values of the columns. 
* for e.g: In the IPL dataset, if we want to change 'Royal Chalengers Banglore' to 'RCB', we can use this parameter.
* It takes a dictionary where **key** is a **column_name** and **value** is a *function**.

* So let's first create a function that renames the values of the column.
* Here, we will rename **Royal Challengers Banglore** to **RCB**

```python
def rename(name):
    if name == "Royal Challengers Bangalore":
        return "RCB"
    elif name == "Mumbai Indians":
        return "MI"
    else:
        return name
```

```python
rename ("Royal Challengers Bangalore")
```

```python
df11 = pd.read_csv("IPL Matches 2008-2020.csv", converters = {"team1": rename})
df11["team1"]
```
