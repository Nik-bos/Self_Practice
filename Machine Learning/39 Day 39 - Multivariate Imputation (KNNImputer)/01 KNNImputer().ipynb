# Difference between `Univariate and Multivariate imputation`.

`Univarriate imputation`: When we're filling missing values of any column, and to do so, we consider data of only that particular column, it is called **Univarate imputation**.<br>

`Multivariate imputation`: When we're filling missing values of any column, and to do so, we consider entire data, then this is called **Multivariate imputation**. What KNNImputer does is, it fills missing value of the raw with the value of similar raw based on similarity.

There're many methods for `Multivariate imputation`. But we'll study only two **1) KNNImputer()** and **2) IterativeImputer** .

Here, in this session, we'll learn **KNNImputer()** and in next session we'll learn **IterativeImputer()**

**Advantage and disadvantages of KNNImputer()**:<br>

* It's more accurate than other methods.
* More computation, so takes more time(**Disadvantage**)
* As we are considering the whole dataframe for imputation, we'll have to put whole training set to the server while deployment(**Disadvantage**)

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('00 titanic.csv')[['Age','Pclass','Fare','Survived']]
```

```python
df.head()
```

```python
(df.isnull().mean())* 100

# Age column is having missing value, so let's fill it.
```

```python
# Train/ test split.

x = df.drop(columns = ['Survived'])
y = df['Survived']
```

```python
x.head()
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test,  y_train, y_test = train_test_split(x, y, test_size = 0.2)
```

```python
x_train.head()
```

```python
y_train.head()
```

```python
# KNNImputer()

from sklearn.impute import KNNImputer

knn = KNNImputer()

# Try making changes in the parameters on KNNImputer() and see the accuracy/

'''
parameters:
1) n_neighbors: No.of neighbors to consider. (Bydefault 5)
2) add_indicator: To add missing indicator(Bydefault False). We studied this concept in our previous session.
3) weights: Weights to be calculated with formula('uniform', 'distance' or 'user defined function'. Bydefault 'uniform')
4) metric: Formula to be used('naneuclidean' or 'user defined function'. Bydefault 'naneuclidean')
'''
```

```python
# Imputing

x_train_1 = knn.fit_transform(x_train)
x_test_1 = knn.transform(x_test)
```

```python
# Model training

from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()
```

```python
lr.fit(x_train_1, y_train)
```

```python
pred_1 = lr.predict(x_test_1)
```

```python
# Accuracy_score

from sklearn.metrics import accuracy_score
```

```python
acc_1 = accuracy_score(y_test, pred_1)
acc_1
```

# Running loop for `n_neighbors`.

```python
highest_acc = 0
best_n_neighbors = 0
```

```python
for n in range(1, 11):
    knn = KNNImputer(n_neighbors = n, add_indicator = True)
    x_train_trans = knn.fit_transform(x_train)
    x_test_trans = knn.transform(x_test)

    lr.fit(x_train_trans, y_train)
    pred = lr.predict(x_test_trans)

    acc = accuracy_score(y_test, pred)

    print(f'{n}: {acc}')

    if acc > highest_acc:
        highest_acc = acc
        best_n_neighbors = n

print()
print()
print()
print(f'Highest accuracy: {highest_acc}')
print(f'Best n_neighbors: {best_n_neighbors}')

print(x_train_trans.shape)    # To check if it's adding one extra column for missing indictor or not.
```
