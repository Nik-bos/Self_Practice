# What are `Outliers`?
`Outliers` are those datapoints which are extremely different from the other datapoints in the dataset.  <br>
For eg: In an examination of 100 students, the percentage of 90 students falls between **50-70**. But there're 5 students who got percentage between **95-98** and there are other 5 students whose percentage is between **30-35**.  <br>
We can clearly see that the precentage of these 10 students are extremely different than the other students of the class. In this case, these 10 students are the `Outliers`.  <br>

It's very important to deal with the outliers bcz if outliers are present in the training set, the model tries to fit the outliers as well and will not give accurate results on unforseen data.  <br>

# When's `Outliers` dangerous?

Whether the outliers are dangerous or not depends on the type of problem. For example, If we are dealing with the prediction problem where we are prediction the `income` of the person based on **Age**, in this case if someone's **Age** is 300, it is an outlier.  <br>

Consider the other scenerio where we are creating the `Anomaly detection model`, in this case we are actually catching the outliers, so we can't remove or make any changes to them.

# Effects of outliers on ML Algorithms.

* ***Makes effect***

1) Linear Regression
2) Logistic Regression
3) Adaboost
4) Deep Learning  <br>

Here it makes effect bcz these algorithms deal with weights in its calculations. When any algorithm considers **weights** in its calculations, `outliers` will create problem to such algorithms.

***Doesn't make any effect***

1) Tree like algorithms
2) Decision Tree
3) Random Forest
4) XG Boost

# How to deal with `Outliers`?

There are 2 most commonly used ways to deal with the outliers. They're,
1) `Remove(Trimming)`: Simply remove the data having outliers, from the dataset.
2) `Capping`: The values that lie **outside 1.5 - IQR or upper limit**, we can replace them with **95th percentile**  <br>
and the values that lie **outside 1.5 + IQR or lower limit**, we can replace them with **5th percentile**.

There're some other ways to deal with outliers which are not commonly used. They're,
1) Replace outliers with **np.nan** and then deal with it as we deal with the missing values
2) `Discretization`: We know that **Descretisation** is used for numeric data. Here, it creates bins and classifies the data points in the respective bins. So for example, if 5, 6, 9, 2 and 95, 96, 97 and 98 are outliers, it will put 5, 6, 9 and 2 in the bin of **0-10** and 95, 96, 97, and 98 in the bin of **90-100**

# How to detect `Outliers`?

1) `Normal distribution`: If the data is normally distributed, the the data **outside mean - 3 std.dev** and the data **outside mean + 3 std.dev**, are considered as `outliers`.
2) `Skew distribution`: If the distribution of the data is skewed, then the data falling **outside Q1 - 1.5(IQR)** and **Q3 + 1.5(IQR)** are considered as `outliers`
3) `Simple manner`: If our data is outside 2.5 and 97.5 percentile(Numbers can be changed as per the dataset), we can treat these data as `outliers`.
