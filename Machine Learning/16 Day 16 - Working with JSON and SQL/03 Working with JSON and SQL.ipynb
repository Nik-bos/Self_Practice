# What is JSON?

* The full form of JSON is **Java Script On Notation**
* It is a universal accepted file format, understandable by most of the  programming languages.
* The files which we obtain through API(We will learn API in one of our our later sessions) is generally in JSON format.
* pd.read_json()  function is used for reading JSON file.
* Most of the parameters of **read_csv()** are available in **read_json()**.
