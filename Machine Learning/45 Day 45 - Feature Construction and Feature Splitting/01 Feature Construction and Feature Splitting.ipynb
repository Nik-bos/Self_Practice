# What is `Feature Construction` and `Feature Splitting`?

1) `Feature Construction`: Constructing a whole new feature with the help of existing feature in the dataset. For example, a dataset is having features named **Height** and **Width**, then from these we can create a new feature named **Square feet**. <br>
Another example is, if a dataset is having some features as **Business_income**, **MF_income**, **Interest_income**, we create a new column named **Total_income**. This is called `Feature Construction`.

2) `Feature splitting`: Sometimes there may be more values inside one single column, in this case we split these values and assign each of them to new column. <br>
For example, there's a column named **Name** and the value is **Mr. Nikunj**, we can create two new columns named **Title** and **Name** and split this value as Mr and Nikunj. After splitting, we'll insert Mr in Title column and 'Nikunj' in Name column.

```python
import pandas as pd
import numpy as np
import seaborn as sns
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score
```

```python
df = pd.read_csv('00 titanic.csv', usecols = ['Age', 'Pclass', 'SibSp', 'Parch', 'Survived'])
```

```python
df.dropna(inplace = True)
```

```python
df.sample(10)
```

```python
df.isnull().sum()
```

```python
x = df.drop(columns = 'Survived')
y = df.drop(x, axis = 'columns')
```

```python
x.head()
```

```python
y.head()
```

```python
cross_val_score(estimator = LogisticRegression(), X = x, y = y, cv = 20, scoring = 'accuracy').mean()
```

# Applying `Feature Construction`.

```python
# Creating a new column named 'family' with the help of ParCh and SibSp. +1 is added bcz we're considering that particular person as well

x['Family_size'] = x['Parch'] + x['SibSp'] + 1
```

```python
x.head()
```

```python
def type(num):
    if num == 1:
        return 0    # Alone
    elif num > 1 and num <= 4:
        return 1    # Small size
    else:
        return 2    # Big family
```

```python
type(10)
```

```python
# Creating a new column 'Family_type' considering the above function.

x['Family_type'] = x['Family_size'].apply(type)
```

```python
x.head()
```

```python
# Now when we've created a column 'Family_type' from 'Family_size' which was created from 'SibSp' and 'Parch', we can drop those columns

x.drop(columns = ['SibSp', 'Parch', 'Family_size'], inplace = True)
```

```python
x.head()
```

```python
cross_val_score(estimator = LogisticRegression(), X = x, y = y, cv = 20, scoring = 'accuracy').mean()
```

# Feature splitting.

```python
df2 = pd.read_csv('00 titanic.csv', usecols = ['Name', 'Ticket', 'SibSp', 'Parch', 'Survived'])
```

```python
df2['Title'] = df2['Name'].str.split(',', expand = True)[1].str.split('.', expand = True)[0]
```

```python
df2.groupby('Title').mean()['Survived'].sort_values(ascending = False)
```

```python
df2[df2['Title'] == ' Mrs']
```

```python
# There're trailing space before values in 'Title' column, so removing it.

df2['Title'] = df2['Title'].str.strip()
```

```python
df2['Is_Married'] = 0
df2['Is_Married'].loc[df2['Title'] == 'Mrs'] = 1
```
