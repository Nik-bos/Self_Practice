```python
# KNNImputer is good for imputing missing values of numeric columns as here our 'Age' column is Ordinal categorical.
# So first we'll have to apply ordinalencoder and then KNNImputer.

from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import KNNImputer
from sklearn.preprocessing import LabelEncoder
```

```python
from sklearn.compose import ColumnTransformer

transformer = [
    ('trans1', OrdinalEncoder(), ['Age', 'Income', 'Creditrating']),
    ('trans2', OneHotEncoder(sparse=False, drop='first'), ['Student'])
]
```

```python
ct = ColumnTransformer(transformers = transformer, remainder = 'passthrough')
```

```python
x_train = ct.fit_transform(x_train)
```

```python
x_test = ct.transform(x_test)
```
