# Concept.
Here in **Mssing indicator** we'll create a new column which will have values as `True` or `False`.<br>
For eg: There's an `Age` column with some missing values. Then in this case, we'll create a new column named `Age_missing`. Corresponding to filled value in `Age` column, it will have `True` and correspond to null values in `Age` column, it will have False.<br>

***See the below DataFrame for better understanding.***

```python
import pandas as pd
import numpy as np
```

```python
data = {
    'Age': [15, 16, np.nan, 14, 25, np.nan, 19, 21, np.nan],
    'Age_missing': ['False', 'False', 'True', 'False', 'False', 'True', 'False', 'False', 'True']
}
```

```python
pd.DataFrame(data)
```

# What is the use of this.

Actually this was the method use by one participants' group in some competition and the accuracy of their model really increased. There's not any proper answer regarding how's accuray increasing in this but one conception is that, the model is learning to differentiate between **Missing values and filled values**. <br>

This technique doesn't always work but still one should try if required.<br>
Let's see one simple code.

```python
df = pd.read_csv("00 titanic.csv", usecols = ['Age', 'Fare', 'Survived'])
```

```python
df.sample(8)
```

```python
df.isnull().sum()
```

```python
# Train/ test split
x = df.drop(columns = 'Survived', axis = True)
y = df.drop(x, axis = 'columns')
```

```python
x.head()
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2)
```

```python
x_train.head()
```

```python
y_train.head()
```

# Let's compare accuracies with and without applying `MissingIndicator`.
# Filling  missing value with SimpleImputer

```python
from sklearn.impute import SimpleImputer    
```

```python
si = SimpleImputer(strategy = 'mean')
```

```python
x_train_trans = si.fit_transform(x_train)
x_test_trans = si.transform(x_test)
```

```python
x_train_trans
```

```python
# Training the model

from sklearn.linear_model import LogisticRegression
```

```python
lr = LogisticRegression()
```

```python
lr.fit(x_train_trans, y_train)
```

```python
pred_lr = lr.predict(x_test_trans)
```

```python
# Calculating accuracy.

from sklearn.metrics import accuracy_score
```

```python
acc_lr = accuracy_score(y_test, pred_lr)
acc_lr
```

# Using `MissingIndicator`.

```python
from sklearn.impute import MissingIndicator
```

```python
mi = MissingIndicator()
```

```python
mi.fit(x_train)
```

```python
mi.features_    # array[0] indicated that column at 0 position in our dataset has missing values i.e 'Age' and it's true.
```

```python
x_train_missing = mi.transform(x_train)

# This should have the values as True or False respectively for missing and filled values in Age column.
# Let's create a table to corss-check it
```

```python
x_train_missing
```

```python
x_train_temp = x_train

x_train_temp.insert(2, 'Age_missing', x_train_missing)    # Insert at loc = 2nd,  column = 'column_name', value = 'values'
```

```python
x_train_temp.sample(10)
```

```python
# Transfering x_test
x_test_missing = mi.transform(x_test)
```

```python
x_test_missing
```

```python
# Adding these new columns to the original x_train and x_test.

x_train['Age_missing'] = x_train_missing
x_test['Age_missing'] = x_test_missing
```

```python
x_train.sample(10)
```

```python
x_test.sample(10)
```

```python
 # Applying SimpleImputer after MissingIndicator.
si = SimpleImputer()

x_train_trans_2 = si.fit_transform(x_train)
x_test_trans_2 = si.transform(x_test)
```

```python
x_train_trans_2.shape
```

```python
x_train.shape
```

```python
# Training the model

lr2 = LogisticRegression()
```

```python
lr2.fit(x_train_trans_2, y_train)
```

```python
pred_mi = lr2.predict(x_test_trans_2)
```

```python
pred_mi
```

```python
acc_mi = accuracy_score(y_test, pred_mi)
acc_mi

# We can see that accuracy of our model increased from 0.70 to 0.71. But it is not so that everytime accuracy increases. We should always try this method.
```

# This same process of applying MissingIndicator() can be obtain just by passing `add_indicator` parameter in `SimpleImputer()`.
# Let's see this also.

```python
si3 = SimpleImputer(add_indicator = True)
```

```python
x_train_trans_3 = si.fit_transform(x_train)
x_test_trans_3 = si.transform(x_test)
```

```python
x_train_trans_3
```

```python
# Training the model

lr3 = LogisticRegression()

lr3.fit(x_train_trans_3, y_train)
```

```python
pred3 = lr3.predict(x_test_trans_3)
```

```python
acc_add_imputer = accuracy_score(y_test, pred3)

acc_add_imputer
```

# Conclusion:
* Accuracy without applying `MissingIndicator()`: 0.7094972067039106
* Accuracy after applying `MissingIndicator()`: 0.7150837988826816
* Accuracy after passing `add_indicator` parameter in `SimpleImputer()`: 0.7150837988826816

# Note:
* Accuracy of our model increased from **0.7094** before `MissingIndicator()` to **0.7150**  after **MissingIndicator()**.
* Moreover, we are getting same result and accuracy by using `MissingIndicator()` class and by passing `add_indicator` parameter to `SimpleImputer()` class.
# `add_indicator` parameter is also available in KNNImputer() and IterativeImputer() class, which we'll learn next.
# It's advisable to use `add_indicator` parameter bcz it is much cleaner that the other method.
