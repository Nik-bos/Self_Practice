Encoding categorical data means converting qualitative data to quantitative data.<br>
There're two types of categorical data. They're,
1) Nominal: Where there is no order in the data. It is simple qualitative data. For eg. State, Country, Qualities etc... 
2) Ordinal: Where order is available. For eg. Grades(A-B-C-D), Rank(One-Two-Three), Level(First-Second-Third) etc...

As we know, machine learning algorithm only understands numbers, we need to encode categorical data if any. There are two methods used to encode the categorical data.
1) Ordinal encoding (For Ordinal data) and
2) One-Hot Encoding(For Nominal data).

Here, in this session, we'll learn `Ordinal encoding` and in the next session we will learn `One-Hot encoding`.<br>
Today we'll also learn `Label encoding`.<br>

# What is Label encoding?
When we need to encode target variable, we use label encoding whereas while encoding predictor variables, we use `OrdinalEncoding`.<br>
There is a class named **LabelEncoder** in scikitlearn.<br>
* from skearn.preprocessing import LabelEncoder.

# How to decide which encoding to perform, ordinal or one-hot?

See the data, if it is in the order, go for ordinal encoding, else, one-hot encoding.<br>
For eg. There's a column named `Education` in our dataset and the values are among 'HS', 'UG', 'PG'. Then here we can clearly identify that `PG > UG > HS. So we should apply Ordinal encoding and assign 0, 1 and 2 to HS, UG and PG respectively.

```python
df = pd.read_csv("26 customer.csv")
df.sample(5)
```

```python
print(df['review'].unique())
print(df['education'].unique())
print(df['purchased'].unique())
```

```python
# Here, we have 3 categorical columns, namely 'gender', review', 'education' and 'purchased'.
# 'gender' has values like 'Female' or 'Male'. So there's no order in this. Hence it is a Nominal column
# 'review' is Ordinal column.(Average-Poor-Good)
#'education' is also ordinal column.(School-UG-PG)
# Purchase is a target variable and has no order(NO and Yes). So Nominal.
```

# Here we will apply One-Hot encoder for Nominal columns, Ordinal encoder for ordinal columns and LabelEncoder for target(Label) column.

In this video we'll learn `Ordinal encoding` and `Label encoding` . So here, we will deal with `review`, `education` and `purchased` columns.

```python
df = df.iloc[:, 2:]    # All the rows from column 2 to last column
df.sample(5)
```

```python
x = df.iloc[:, 0:2]
y = df.drop(x, axis = 'columns')

print(x.head())
print()
print()
print()
print(y.head())
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
oe = OrdinalEncoder(categories = [['Poor', 'Average', 'Good'], ['School', 'UG', 'PG']])

# Here, categories is a parameter of OE class and we need to  pass no.of lists we wanna encode.
# The list should be in the order of lower to higher status.

# Here we passed one list as ['Poor', 'Average', 'Good'] bcz we are telling OE class to give lower importance to Poor and highest importance  to good.
# Second list is ['School', 'UG', 'PG'], means give lower imp to School and higher to PG
```

```python
oe.fit_transform(x)
```

```python
oe.categories_
```
