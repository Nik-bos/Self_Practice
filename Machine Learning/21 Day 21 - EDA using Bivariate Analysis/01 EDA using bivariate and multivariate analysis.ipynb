# Bivariate analysis:
When we perform analysis on two columns it is called bivariate analysis. There are three types of senerios between data types of columns, they're
* Numerical-Numerical
* Categorical-Categorical
* Numerical-Categorical

Considering the senerio, we need to select the chart.
1) Numerical-Numerical:

   a) Scatterplot

# 1) Numerical-Numerical (Scatterplot)

```python
sns.scatterplot(x = tips['total_bill'], y = tips['tip'])
```

```python
sns.scatterplot(x = 'total_bill', y = 'tip', data = tips)

# We can see that there's a linear positive relationship between the bill's amount and the tip given.

'''
Now to perform multivariate analysis with this same columns like to differentiate between
males and females, then use 'hue' parameter of scatterplot class.
'''
```

```python
sns.scatterplot(x = 'total_bill', y = 'tip', hue = 'sex', data = tips)
```

```python
'''
Now what if we want additional informatin that whether the customer is a smoker or not.
Use 'style' parameter and assign 'smoker' column to it.
'''
sns.scatterplot(x = 'total_bill', y = 'tip', hue = 'sex', style = 'smoker', data = tips)

```

```python
'''Now again what if we want another additional informatin that with how many people customer visite the restaurant.
Use 'size' parameter and assign 'size' column to it.
'''

sns.scatterplot(x = 'total_bill', y = 'tip', hue = 'sex', style = 'smoker', size = 'size', data = tips)
```

```python
'''
Now here in the above code, we are getting 4 pieces of information. They're\n",
1) Relation between bill's amount and the tip given,
2) The person is male or a female,
3) Does he/she smoke or not and
4) He/she came with how many persons.
'''
```

# 1) Numerical-Numerical(Line plot).
Another graph we can use if both the columns are nnumerical is a lineplot. But remember, we sould use lineplot only when our x-axis is a time series data like days, months, year, time, date etc...Generally time is taken on x-axis

```python
'''
In flights dataset there are two columns namely 'year' and 'passengers' and both are numerical. Also, 'year' colums shows time as well.
So we can plot these data in lineplot taking 'year' on x-axis.

For this, we'll have to perform groupby on 'year'.
'''
```

```python
flight_group = flights.groupby('year').sum().reset_index()

flight_group
```

```python
sns.lineplot(x = 'year', y = 'passengers', data = flight_group)
```

```python
# Let's see another example from tips dataset. But let's first group it by day and then plot day and tip 
# column on lineplot taking day on x-axis and tip on y-axis.
```

```python
tips_group = tips.groupby('day').sum().reset_index()
tips_group
```

```python
sns.lineplot(x = 'day', y = 'total_bill', data = tips_group)
```

# 2) Numerical-Categorical(Barplot)

When we have one categorical and one numerical column, we can use bar plot.

Here generally on x-axis we kepp categorical column and on y-axis we keep numerical column.

```python
titanic.iloc[0:2,:]
```

```python
# Let see what is the average age in each Pclass.

sns.barplot(x = 'Pclass', y = 'Age', data = titanic)

# The black rod we can see here displays confidence interval. It is the topic of Statistics and we'll learn it there.
```

```python
# We can also do multivariate analysis here as well using 'hue' parameter.

sns.barplot(x = "Pclass", y = 'Age', hue = 'Sex', data = titanic)
```

# 2) Numerical-Categorical(Boxplot).

We can plot two box plots of different columns as well.

```python
sns.boxplot(x = 'Sex', y = 'Age', data = titanic)
```

```python
# We can also get box plot of three different columns

sns.boxplot(x = 'Sex', y = 'Age', hue = 'Survived', data = titanic)
```

# 2) Numerical-Categorical(Distplot)

In distplot, y-axis shows probability.

```python
sns.distplot(x = titanic["Age"])

# distplot() doesn't have 'data' parameter.
```

```python
# There's no 'hue' parameter in distplot(). So we will hafta filter and plot

# Here we want to plot the age of survived == yes and survived = no.

sn = titanic[titanic['Survived'] == 0]['Age']    # Survived == No and extracted 'Age column'
sy = titanic[titanic['Survived'] == 1]['Age']    # Survived == Yes and extracted 'Age column'
```

```python
sns.distplot(sn, hist = False, label = 'Not Survived')
sns.distplot(sy, hist = False, label = 'Survived')
plt.legend()

# We made 'hist' parameter as False bcz it was creating trouble for us to understand.
```

# 3) Categorical-Categorical(Heatmap)

```python
# Pclass and Survived are both categorical columns and we wanna find out how many people died and survived in each Pclass.

a = pd.crosstab(titanic['Pclass'], titanic["Survived"])
print(a)

sns.heatmap(a)
```

# 3) Categorical-Categorical(Clustermap)

```python
a = pd.crosstab(titanic['Parch'], titanic['Survived'])
print(a)
```

```python
sns.clustermap(a)
```

# Pairplot.

Is is applied between numeric columns only. It's just a collection of various scatter plot.

Say we have large no.of numeric columns and we want to find the relationship of each column with the other one. Then to plot a scatterplot for every column becomes very tedious. So here we can use 'Pairplot'.
What **Pairplot** does is, it finds all the numeric columns in our dataset and then plots scatterplot of every column with the other column.

Let's see an example.

```python
iris.sample(5)
```

```python
sns.pairplot(iris)
```

```python
# We can also pass a hue parameter here and give categorical column ti it(As all the numeric columns are already taken into consideration)

sns.pairplot(iris, hue = 'species')
plt.title("Pairplot of iris's dataset")
```
