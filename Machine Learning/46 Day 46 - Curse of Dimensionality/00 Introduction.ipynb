# What is Curse of Dimensionality?

***Note: `Features` means `dimensions` and `columns` as well***

In our previous session, we studied feature engineering and now we will learn `Feature extraction`. But before moving towerds it, we'll learn `Curse of Dimensionality`.<br>

In machine learning, the accuracy or say model will perform better with more number of features. But there's an irony in this. It stops performing better infact deteriorates after certain number of features. So machine learning algorithm works better with optimum features after which if you increase features, it doesn't make any difference and sometimes become worse.

`Let's understand this with an example`.<br>
Imagine you lost your wallet in your college. Which can be the scenerio in which you will find it faster? It can be among the given three.<br>
1) You know you lost it somewhere in between the college gate and the main building, which is a straight road.
2) You know you lost it in the playground.
3) You lost it in the main building.

The fastest scenerio is 1, you lost it somewhere between the college gate and the main building bcz it is one straight road. Whereas playground is a wide area and the main building has many number of floors and rooms. What i mean to say is, when you are searching something in lesser dimension, you will get better results. Here, straight road from college gate to the main building is just straight line(Imagine it as 1 dimension).<br>

The same goes with the machine learning algorithms. After optimum number of dimensions(features), you you further increase the dimensions, it will not create and difference instead may worsen the model's preformance.<br>

Imagine you are applying KNN model. We know it clasifies into class considering the distance from the neighbours. Suppose that we're having 2 or 3 features. It will calculate the distances and will give the result.<br>
Now suppose we've 150 features. Obviously it will calculate the distances and give the results. But here distances will be much higher that we can't logically consider them as neighbours.
# This is called `Curse of Dimensionality`.

# How to overcome it?
There're two ways,`1) Fearure selection` and `2) Feature extraction`.<br>

1) `Feature selection`: Here we select subset of features. Suppose we have **F1 to F50** features. Then in this case, we select **F5, F4, F21, F6, F18, F34** and train the model with these selected features. It's methods are,
* Forward selection and
* Backward selection

2) `Feature extraction`: Here, we create completely new columns with the help of existing columns. Some common techniques to perform these are,
* Principle Component Analysis(PCA)
* Linear Discriminent Analysis(LDA)
