What is `One-Hot encoding`?

Supose we have a column **Color** and it has values like Red, Yellow and Green. What we do in one-hot encoding is, we create 3 new columns named Red. Yellow and Green and then to show Red we insert 1 and 0 to others.
For eg. Red = [1, 0, 0]<br>
Yellow = [0, 1, 0]<br>
Green = [0, 0, 1]

Now what if we have 50 unique values in a column? Yes, we will create 50 columns for that.<br>
But then the dimension of our dataset will increase. Yes but there's no other way. But if some values are repeating frequently, we can create columns of those frequent values and all the other(non frequent values) values can be put under one single column named `Other`.<br>

Moreover, we delete one column from all the columns to avoid Multicolinearity.`Multicolinearity` means all the columns are dependent on each other.
It ceates problem when we deal with Linear models like Linear regression or Logistic regression. We'll learn what is multicoliniarty and how it affects Linear models n detail when we'll study Linear models. But as for now, just remember, we remove one column to avoid the problem of Multicolinearity.
