```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_diabetes
```

```python
data = load_diabetes()
print(data.DESCR)
```

```python
# splitting data into class and predictors variable

x = data.data
y = data.target

# performing train test split

from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 1) 
```

# Appplying Linear Regression

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()
```

```python
# training the model

lr.fit(x_train, y_train)
```

```python
pred = lr.predict(x_test)
```

```python
from sklearn.metrics import r2_score, mean_squared_error
```

```python
r2 = r2_score(y_test, pred)
mse = mean_squared_error(y_test, pred)
rmse_score = np.sqrt(mse)
print(f'R2 score: {r2}')
print(f'Root Mean Squared Error: {rmse_score}')
```

# Ridge(L2) Regression

```python
from sklearn.linear_model import Ridge, Lasso
```

```python
ridge = Ridge(alpha = 0.0001)

ridge.fit(x_train, y_train)
```

```python
pred_ridge = ridge.predict(x_test)
```

```python
r2_ridge = r2_score(y_test, pred_ridge)
mse_ridge = mean_squared_error(y_test, pred_ridge)
rmse_score_ridge = np.sqrt(mse_ridge)
print(f'R2 score: {r2_ridge}')
print(f'Root Mean Squared Error: {rmse_score_ridge}')
```

```python
# Using Lasso
```

```python
lasso = Lasso(alpha = 0.0001)

lasso.fit(x_train, y_train)
```

```python
pred_lasso = lasso.predict(x_test)
```

```python
r2_lasso = r2_score(y_test, pred_lasso)
mse_lasso = mean_squared_error(y_test, pred_lasso)
rmse_score_lasso = np.sqrt(mse_lasso)
print(f'R2 score: {r2_lasso}')
print(f'Root Mean Squared Error: {rmse_score_lasso}')
```

# Here we are not getting much difference because data is not much complex.

# Note: It is good to apply regularization by-default bcz it don't create any problem while training the model or in performance.
