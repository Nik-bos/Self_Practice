# Advantage of `Doubly Linked List` over `Singly Linked List`

* In `Doubly LinkedList`, we store the reference of both, preceding and succeeding node, so traversing is easier than in the `Singly LinkedList`.

* Also, it is very easier to traverse backward as we store reference of both nodes, Whereas it is very difficult to traverse backward in Singly LinkedList

Here also we will perform three operations,
* **Traversal**
  * Forward
  * Backward
* **Insertion**
  * Beginning
  * End
  * Middle (By value)
* **Deletion**
  * Beginning
  * End
  * Middle (By value)

```python
class Node:
    '''
    This class is used to create nodes for Doubly LinkedList.

    **Parameters**:
    data: Data to be inserted
    
    '''
    def __init__(self, data):

        self.data = data
        self.next_ref = None
        self.pref = None

        
```

```python
class Node:
    
    def __init__(self, data):

        self.data = data
        self.next_ref = None
        self.pref = None

# ------------------- Class Doubly Linked List -------------------

class DoublyLL:
    def __init__(self):
        self.head = None

# ============================= Traversing Forward =============================

    def traverse_f(self):

        # If list is empty
        if self.head is None:
            raise IndexError("Linked List is empty.")
            
        n = self.head
        l = []
        while n is not None:
            l.append(str(n.data))
            n = n.next_ref
        
        return '-->'.join(l)

# ============================= Traversing Backward =============================

    def traverse_b(self):
        
        # Case 1: If list is empty
        if self.head is None:
            raise IndexError("Linked List is empty.")

        
        # Case 2: If list is not empty
        
        # Going to the last node
        n = self.head
        
        while n.next_ref is not None:
            n = n.next_ref
        # Now n is pointing to the last node
        
        # Traversing backward from the last node
        l = []
        while n is not None:
            l.append(str(n.data))
            n = n.pref
        return '-->'.join(l)

# ============================= Inserting at head =============================

    def add_head(self, data):

        new_node = Node(data)
        
        # Case 1: If linked list is empty
        if self.head is None:
            self.head = new_node

        # Case 2: If list is not empty
        else:
            n = self.head
            new_node.next_ref = self.head
            n.pref = new_node
            self.head = new_node    # Now new node is the head node
        
        return f"{data} inserted on the head"

# ============================= Inserting at tail =============================

    def add_tail(self, data):

        
        new_node = Node(data)

        # Case 1: If linked list is empty
        if self.head is None:
            self.head = new_node
            return f"{data} inserted on the tail"

        # Case 2: If list is not empty
        n = self.head
        while n.next_ref is not None:
            n = n.next_ref
        
        n.next_ref = new_node
        new_node.pref = n
        
        return f"{data} inserted on the tail"

# ============================= Inserting at middle =============================

    def add_after(self, data, x):

        # If list is empty
        if self.head is None:
            raise IndexError("Cannot add to an empty linked list.")

        
        n = self.head
        while n is not None:
            if n.data == x:
                new_node = Node(data)
                new_node.pref = n
                new_node.next_ref = n.next_ref

                # If x is the last node, then next_ref will be None.
                # Also, if x is the only node, means it is also the last node. Hence next_ref will be None
                
                # So if it is not the last node, then next to next node will get the link to new_node
                # and if x is the last node, then just assign last_node's next_ref to new_node (Means avoid 'if' block)
                if n.next_ref is not None:    
                    n.next_ref.pref = new_node
                
                n.next_ref = new_node

                return f"{data} inserted after {x}"

            n = n.next_ref

        else:
            raise ValueError(f"{x} not in the list")

# ============================= Inserting at middle =============================

    def add_before(self, data, x):

        if self.head is None:
            raise IndexError("Cannot add into an empty linked list.")

        n = self.head
        while n is not None:
            if n.data == x:
                new_node = Node(data = data)
                new_node.next_ref = n
                new_node.pref = n.pref

                # If x is the first node, then pref will be None.
                # Also, if x is the only node, means it is also the first node. Hence pref will be None
                
                # So if it is not the first node, then previous to previous node will get the link to new_node
                # and if x is the first node, then just assign its next_ref to new_node (Means avoid 'if' block)
                # Caution: If x is the first node and we are ading new_node before first node,
                # then now new_node is the first node, so self.head = new_node
                
                
                if n.pref is not None:    # IF x is not the first node
                    n.pref.next_ref = new_node
                else:                     # If x is the first node, adding new node b4 it, new_node becomes the first node
                    self.head = new_node
                
                n.pref = new_node

                return f"{data} inserted before {x}"
            
            n = n.next_ref
        else:
            raise ValueError(f"{x} not in the list")

# ============================= Deleting head node =============================

    def del_head(self):

        if self.head == None:
            raise IndexError("Cannot delete from an empty lisked list.")
        
        # Case 1: If there is only one node in the list
        deleted_value = self.head.data
        if self.head.next_ref == None:
            self.head = None
            return f"{deleted_value} deleted (Only one node was present in the list)"


        # Case 2: Otherwise mode head to the next node
        self.head = self.head.next_ref
        self.head.pref = None
        
        return f"{deleted_value} deleted."

# ============================= Deleting tail node =============================

    def pop(self):
        
        if self.head == None:
            raise IndexError("Cannot delete from an empty lisked list.")

        # Case 1: If there is only one element in the list
        if self.head.next_ref == None:
            return self.del_head()

            '''
            We can also write the same code we wrote in del_head()

            deleted_value = self.head.data
            if self.head.next_ref == None:
                self.head = None
                return f"{deleted_value} deleted (Only one node was present in the list)"
            '''
        # Case 2: If there are more than one elements in the list.
        n = self.head
        while n.next_ref != None:
            n = n.next_ref
            
        deleted_value = n.data
        n.pref.next_ref = None
        return f"{deleted_value} deleted."

# ============================= Deleting by value =============================

    def del_x(self, x):

        # Let's reuse the methods which we already created

        if self.head == None:
            raise IndexError("Cannot delete from an empty linked list.")

        n = self.head
        while n is not None:
            
            if n.data == x:
                
                deleted_value = n.data

                # Case 1: If only one node is present in the list
                if n.pref == None and n.next_ref == None:
                    self.head = None
                    return f"{deleted_value} deleted (Only one node was present in the list)."

                # Case 2: If x is the head node.
                if n.pref == None:
                    return self.del_head()

                
                # Case 3: If x is the last node
                if n.next_ref is None:
                    return self.pop()

                # Case 4: Deleting middle nodes
                n.next_ref.pref = n.pref
                n.pref.next_ref = n.next_ref

                return f"{deleted_value} deleted."

            n = n.next_ref

        else:
            return f"{x} not in the list."
                
        
```
