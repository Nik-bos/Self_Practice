```python
class Node:
    def __init__(self, data):
        self.data = data
        self.ref = None

#######################################################################

class LinkedList:
    def __init__(self):
        self.head = None

#--------------------------- Printing the LinkedList ---------------------------
    
    # Traversing a linked list
    def __str__(self):
        if self.head is None:
            return "Linked List is empty"
        else:
            n = self.head
            values = []
            while n is not None:
                values.append(str(n.data))
                n = n.ref
            return " -> ".join(values)

#--------------------------- Adding the element in the beginning ---------------------------
        
    # Adding a node at the beginning
    def add_begin(self, data):
        new_node = Node(data)
        new_node.ref = self.head
        self.head = new_node

#--------------------------- Adding the element at the end ---------------------------
    
    # Adding a node at the end
    def add_end(self, data):
        
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
        else:
            n = self.head
            while n.ref is not None:
                n = n.ref
            n.ref = new_node

#--------------------------- Adding after element x ---------------------------

    # Adding a node after a specific value
    def add_after(self, data, x):

        if self.head is None:
            raise IndexError('Linked List is empty.')

            
        n = self.head
        while n is not None:
            if n.data == x:
                new_node = Node(data)
                new_node.ref = n.ref
                n.ref = new_node
                return "Value inserted"
            else:
                n = n.ref
        else:
            raise ValueError(f"Value '{x}' not found in the list.")

#---------------------------- Adding before element x ----------------------------

    def add_before(self, data, x):
        
        if self.head is None:
            raise IndexError('Linked List is empty.')

        n = self.head

        # If 'x' is the only one element in the list, in that case, n.ref will be None.
        # So, n.ref.data will raise error bcz None.data is not possible.
        # So we will have to apply another logic for it or we can simple call self.add_begin(data)
        if n.data == x:
            self.add_begin(data)
            return "Value inserted"

            
            '''
            This is the same logic present is add_begin(). We can simply place these lines instead of calling
            self.add_begin(data)
            
            new_node = Node(data)
            new_node.ref = self.head
            self.head = new_node
            return f'Value inserted' 
            '''
        
        while n.ref is not None:
            if n.ref.data==x:
                self.add_after(data = data, x = n.data)
                return "Value inserted"

                '''
                This is the same logic present is add_after(). We can simply place these lines instead of calling
                self.add_begin(data)
                
                new_node = Node(data)
                new_node.ref = n.ref
                n.ref = new_node
                return "Value inserted"
                '''
            else:
                n = n.ref
        else:    # Value not found in the list
            raise ValueError(f"Value '{x}' not found in the list.")

#---------------------------- Delete the first element ----------------------------

    def del_first(self):

        if self.head is None:
            raise IndexError('Linked List is empty. Cannot delete from an empty Linked List')
        else:
            deleted_value = self.head.data
            self.head = self.head.ref
            return f"Node with value {deleted_value} deleted from the beginning."

#---------------------------- Delete the last element ----------------------------

    def pop(self):

        if self.head is None:
            raise IndexError('Linked List is empty. Cannot pop from an empty Linked List')

        # If there is only one element in the Linked List, then the else part will not work,
        # So we will have to deal it saperately.
        elif self.head.ref == None:
            deleted_value = self.head.data
            self.head = None
            return f"Node with value {deleted_value} deleted from the end."

        else:
            n = self.head
            while n.ref.ref != None:
                n = n.ref
            
            deleted_value = n.ref.data
            n.ref = None
            return f"Node with value {deleted_value} deleted from the end."

#---------------------------- Delete a specified element (by value) ----------------------------


    def del_x(self, x):

        if self.head == None:
            raise IndexError("Cannot delete from an empty Linked List.")

        # If x is the very first node or it is the only node, in the list.
        if self.head.data == x:
            deleted_value = self.head.data
            self.head = self.head.ref
            return f"Node with value {deleted_value} deleted (was head node)."
            
            '''
            We can also call the del_first() method

            self.del_first()
            '''
        
        
        n = self.head
            
        while n.ref is not None:
            if n.ref.data == x:
                deleted_value = n.ref.data
                n.ref = n.ref.ref
                return f"Node with value {deleted_value} deleted."
                
            n = n.ref
            
        else:    # IF value not found in the list
            raise ValueError(f"{x} not found in the list.")
                


# First 'if' is True, it will raise error and function stops, doesn't execute remaining code.
# Second 'if' is True, it will 'return' and function stops, doesn't execute remaining code.
# If none of the 'Ifs' are executed, it will execute the rest of the code.
```
