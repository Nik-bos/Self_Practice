# What are `output parsers`? 
`Output parsers` help us to get output in structured format like csv, json, pydantic and many more. They ensure consistancy, validation and ease of use. We can generate structured output from both type of models, which has capabality to generate structured output and which doesn;t have capability.

There're many types of output parsers but we'll learn four mostly used. Thery are:
1) String output parser
2) Structured output parser
3) JSON output parser
4) Pydantic output parser

# !) String output parsers:
This is the most simplest output parser in Langchain. It's used to parse the output of an LLM and return it as a string.

See, we we pass any input to LLM, it returns us many meta data like content, prompt_tokens_details, response_meta_data, reasoning_tokens etc. <br>
From this, to get our answer, we print 'content'. <br>
Now, if we use `string output parser`, we will not have to print `content`. We will automatically get `content` from response of the LLM <br>
Now, the question is, it is not a big deal to print `content` then why to use parser?

See, if we ask LLM to give us detailed report on `BlackHole` it will generate. Then we ask it to convert that report in 5-6 lines. <br>
Here we'll need output parser.

`If not understood, see below code`

```python
# from langchain_google_genai import GoogleGenerativeAI
from langchain_google_genai import GoogleGenerativeAI
from dotenv import load_dotenv

from langchain_core.prompts import ChatPromptTemplate, PromptTemplate
```

```python
load_dotenv()
```

```python
model = GoogleGenerativeAI(
    model = 'gemini-1.5-flash'
)
```

```python
# Creating both templates and prompts for getting detailed response
template1 = PromptTemplate(
    template = 'Write a detailed report on {topic}',
    input_variables = ['topic'],
    validate_template = True
)

# Creating prompt1 and getting detailed response
prompt1 = template1.invoke(
    {
        'topic': 'blackhole'
    }
)

response = model.invoke(prompt1)

print(response)
```

```python
# Creating template2 and prompt 2 foor summary
template2 = PromptTemplate(
    template = 'Convert the below text into 5 lines summary.\n{text}',
    input_variables = ['text'],
    validate_template = True
)


prompt2 = template2.invoke(
    {'text': response}
)


summary = model.invoke(prompt2)

print(summary)
```

# Using String Output parsers

```python
from langchain_google_genai import GoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

from dotenv import load_dotenv
```

```python
model = GoogleGenerativeAI(
    model = 'gemini-1.5-flash'
)
```

```python
# Creating parser's object
parser = StrOutputParser()
```

```python
# Creting both the templates. Here we don't need to create prompts implicitly

template1 = PromptTemplate(
    template = 'Write a detailed report on {topic}',
    input_variables = ['topic'],
    validate_template = True
)

template2 = PromptTemplate(
    template = 'Convert the below text into 5 lines summary.\n{text}',
    input_variables = ['text'],
    validate_template = True
)
```

```python
# Parcser works with chains, so, creating a chain

chain = template1| model | parser | template2 | model | parser

response = chain.invoke(
    {'topic': 'blackhole'}
)

print(response)

# Input > Template1 > Prompt_1 > Model > Response > Parser > Parsed_Result > template2 > Prompt_2 > Model > Response > Parser > Parsed_Result 
```

# Difference:
The same output we got in the previous code but there code was lengthy and here we implemented chain. We are able to implement chains bcz we used parser. If not used parser, we would have to create two chains like this

`Chain_1`:  Template1 > Prompt_1 > Model > Response  From here take response.content <br>
`Chain_2`: Template2 > Prompt_2 (Pass response.content) > Model > Response. From here, take final response.content
