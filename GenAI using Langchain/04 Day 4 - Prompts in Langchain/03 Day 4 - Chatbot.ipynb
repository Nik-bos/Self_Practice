# This is bcz the model is not having memory of past conversations. So, lets give it memory along with each new prompt.


```python
history = []

while True:
    user_input = input('You: ')
    history.append(user_input)
    
    if user_input == 'exit':
        break
    else:
        result = model.invoke(history)
        history.append(result)
        print(f'AI: {result}')
print()
print(history)
```

# Note:
Uptill now, we had one problem and we solved it. Now, we have one more problem. Looking at the chat history, how will we know which message is of user and which is of AI? Moreover, here model is giving proper answers bcz the history is small, but when it will increase, AI also will face problem in knowing which message is of User and which are its responses.

Hence, we should create a dictionary where key will be 'user'/ 'AI' and values will be their messages.

{  <br>
"User": "Message"  <br>
"AI": "Message"  <br>
"User": "Message"  <br>
"AI": "Message"  <br>
"User": "Message"  <br>
"AI": "Message"  <br>
}  <br>

`We don't have to do this manually, langchain have built-in classes`

# For this, we'll have to learn types of messages available in langchain. There're 3 types of message in Langchain
1) System message
2) AI message
3) Human message

Go to `messages in langchain` file to understand this.

```python
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
```

```python
history = [
    SystemMessage("You're an expert historian.")
]

history
```

```python
while True:
    user_input = input("You: ")
    history.append(HumanMessage(content = user_input))

    if user_input == 'exit':
        break

    else:
        response = model.invoke(history)
        history.append(AIMessage(content = response))
        
        print(response)
        print()
```

# Consider this scene:
We are building a chatbot or an e-commerce app. A customer came and asked for refund on his order and we responded 'Refund initiated and amount will be credited within 2 working days'

After 2 working days, the user again came and said amount is not credited into his account. Here, chatbot will not understand his query bcz it is not having any history of past conversation.

So, everytime user initiates a chat, along with user_input, we should also give history to the chat bot so that it can know what conversation they had earlier. This histories of users are maintained in cloud.

See the below example

```python
from langchain_google_genai import GoogleGenerativeAI
from dotenv import load_dotenv

from langchain_core.messages import AIMessage
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
```

```python
load_dotenv()

model = GoogleGenerativeAI(
    model = 'gemini-1.5-flash',
    max_tokens = 300,
    temperature = 0.2
)
```

```python
# Load chat history

history = []

with open (r'chat_history.txt') as f:
    history.extend(f.readlines())

```

```python
# Creating chat_template

chat_template = ChatPromptTemplate([
    ('system', 'You are an helpful customer support agent.'),
    MessagesPlaceholder(variable_name = 'chat_history'),
    ('human', '{query}')
]
)
```

```python
# Creating prompt

prompt = chat_template.invoke(
    {'chat_history': history, 'query': 'Where is my refund?'}
)

prompt
```

```python
response = model.invoke(prompt)
```

```python
response
```

```python
history.append(AIMessage(response))
```

# Using for loop

```python
from langchain_google_genai import GoogleGenerativeAI
from dotenv import load_dotenv

from langchain_core.messages import AIMessage, HumanMessage
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
```

```python
# Load chat history

history = []

with open (r'chat_history.txt') as f:
    history.extend(f.readlines())
```

```python
chat_template = ChatPromptTemplate([
    ('system', 'You are an helpful customer assistant.'),
    MessagesPlaceholder(variable_name = 'chat_history'),
    ('human', 'query')
]
)
```

```python
prompt = chat_template.invoke(
    {'chat_history': history, 'query': 'Where is my refund?'}
)

prompt
```

```python
while True:
    user_input = input("Enter your query: ")

    history.append(HumanMessage(content = user_input))

    if user_input == 'exit':
        break
    else:
        prompt = chat_template.invoke(
            {'chat_history': history, 'query': user_input}
        )

        response = model.invoke(prompt)
        history.append(AIMessage(content = response))

        print(response)
    
    
```
