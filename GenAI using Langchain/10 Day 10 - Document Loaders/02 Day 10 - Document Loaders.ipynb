# 1) `TextLoader`

* `TextLoader` is the most common type of document loader. 
* It used to load .txt files into a python's object.
* It is idle for loading chai logs, scrapped text, transcript, code snippet or any other plain text data.

```python
from langchain_community.document_loaders import TextLoader
```

```python
# Creating an object of TextLoader
text_loader = TextLoader(r"00 Day 10 - Gaargi and Yaagyavalakya.txt", encoding = 'utf-8')
```

```python
# LOading the text
docs = text_loader.load()

print(docs)
```

```python
print(type(docs))
print(type(docs[0]))
print(len(docs))
```

```python
from langchain_google_genai import GoogleGenerativeAI
from dotenv import load_dotenv
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
```

```python
load_dotenv()
```

```python
model_gemini = GoogleGenerativeAI(
    model = 'gemini-2.5-flash'
)

parser = StrOutputParser()

temp1 = PromptTemplate(
    template = "Give the summary of the given poem: {poem}",
    input_variables = ['poem']
)

chain = temp1 | model_gemini | parser

response = chain.invoke(docs[0])

print(response)
```

# 2) PyPDFLoader:

* It is used to load PFD files.
* It returns a list of document object and we receive as many document object as there are no_of pages in a PDF file.
* `For example`: If pdf has 5 pages, this will return a list of 5 document objects

# Limitations:

* This uses the PYPDF library under the roof and so, it cannot read any complex PDF files which contains images and other things.
* It can only read PDFs having simple texts.
* We have other document loaders for such PDFs which we will learn later.
