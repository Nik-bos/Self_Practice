# 2) PyPDFLoader:

* It is used to load PFD files.
* It returns a list of document object and we receive as many document object as there are no_of pages in a PDF file.
* `For example`: If pdf has 5 pages, this will return a list of 5 document objects

# Limitations:

* This uses the PYPDF library under the roof and so, it cannot read any complex PDF files which contains images and other things.
* It can only read PDFs having simple texts.
* We have other document loaders for such PDFs which we will learn later.

```python
from langchain_community.document_loaders import PyPDFLoader
```

```python
pypdf_loader = PyPDFLoader(r"00 DL Curriculum.pdf")
```

```python
docs = pypdf_loader.load()
```

```python
print(len(docs))    # Means we have these many pages in our PDF file
print()
print()

print(docs[0].page_content)# returns 1st page's content
print()
print()

print(docs[0].metadata)    # Returns metadata of the 1st page
```

# Two main types of PDFs
`1️⃣ Text-based PDF (Digital PDF)`

Created directly from Word, Excel, HTML, or other digital sources.

The text is stored as actual characters inside the PDF.

You can select, copy, and search the text.

Example: An e-ticket PDF you download from a website.

`2️⃣ Image-based PDF (Scanned PDF)`

Created by scanning a paper document.

Every page is basically a photograph of the page — the text is just pixels in the image.

You cannot select or search the text — dragging the cursor just selects the whole image.

Example: Old book scans, photocopies, or scanned handwritten notes.

# `DirectoryLoader`

Suppose we have a folder containing PDF files and we wanna read them all, in that case, we can use DirectoryLoader.
It takes 3 arguments,
1. `path`: Directory's path
2. `glob`: '*.pdf' (To read all the PDFs present in the directory. Takes pattern of files' name)
3. `loader_cls`: Loader's class we wanna use. For PDF, PyPDFLoader

For example: Our PDF files are present in directory named "GENAI PDFs"

dir_loader = DirectoryLoader(
path = "GENAI PDFs",
glob = '*.pdf',
loader_cls = PyPDFLoader
)

docs = dir_loader.load()

if total no of pages from all pdf is 1000, `docs` will have 1000 `document objects`

# Patterns:

1. `'*.pdf'`: All PDF files from `main folder`
2. `'**/*.pdf'`: All PDF files from `sub-folders`
3. `'data/*.pdf'`: All PDF files from the `data` folder present in `main folder`

`Note`: Inplace of PDF, we can also extract `.txt` files as well. `loader_cls` will become `TextLoader`

# 3. `WebBaseLoader`

* This is used to load web pages from the websites.
* This works well if the web page is almost text based like wikipedia articles, research articles, news articles, blogs etc.

`Now, before moving towards limitations, learn two main types of web pages.`

1. Static page
2. Dynamic page

1. `Static page`:
Those pages, where we can immediately see everything as soon as website gets loaded, such are static pages.
`For example`: Wikipedia articles, research articles, news articles, blogs etc

2. `Dynamic page`: Those pages which doesn't load the contents as soon as website is loaded. It takes time to load contents of the page.
`For example`: Facebook, Instagram, LinkedIn etc loads content as we scroll.

 `Note`: If you turn off JavaScript in a browser, you will not be able to load Dynamic websites like Facebook, Insta, Tweeter, even Google


* So, `WebBaseLoader` can only load `Static pages`
* To load `Dynamic pages`, we have `SeleniumURLLOader`


`Let's load an amazon's page, it is a Static page`

# What is `lazy_load()`?

`Imagine this scene.` Supose in a folder, there are 500 or 1000 PDFs and we want to load all of them, Then? <br>
Correct, it is not possible to load these many documents in the RAM as it ma exceed storage limit of the RAM, and also, this will take very long time to load.

So, to solve this problem, we have `lazy_load()` function which returns a `generator` of documents.
Then, at a time, we can load a single document in the RAM, perform necessary operations and the document gets remove from the RAM.
