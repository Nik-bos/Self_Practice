# What is Text Splitters?

`Text Splitting` is the process of breaking the large text, PDFs, HTML pages or books into small chunks that an LLM can handle effectively. <br>
What performs this process are called `Text Splitters`.

When working with LLM, it is always advisible to split text data into small chunks so that LLM can perform properly.

# Why text splliting is required?

* `Token limitation`:
LLMs and Embedding models have limits on how many tokens it can take as input at a time. We cannot input texts which exceeds this limit. <br>
`For example`: If the limit is set to 50k tokens and out PDF or text file has 1 lakh tokens, we won't be able to pass our PDF as input. We will need to break it into small chunks.

# Types of Text Splitting.

1. Length Based
2. Text Structure Based
3. Document Structure Based and
4. Sementic Meaning Based

# Length Based:

* This is the most easiest splitting techniques.
* It splits the text based on character or words.
`For example`: If splitting is based on 500 characters/ words, it will make chunks of 500 characters/ words.

# Limitation

* It splits based on characters/ words. Supose we are ssplitting based on 500 characters/ words, it will simply split from 500 characters/ chunks.
`For example`: If 500th character is `Swi`, C1 will have `Swi` and C2 will start from `mming` (Same goes when splitting with words)

* This doesn't understand the sementic meaning while splitting. So when we pass each chunks to LLM, one by one for processing, the quality of output declines a lot.

* This is the main reason, `Length Based` text splitting is not used much in reality.

````python
from langchain.text_splitter import CharacterTextSplitter
````

````python
text = '''
The most fascinating philosophical dialogues from the Brihadaranyaka Upanishad (Yajnavalkya–Gargi Samvada).
This debate is not just history — it’s part of India’s ancient intellectual tradition, where women like Gargi Vachaknavi were revered as profound scholars of Vedic philosophy.

Let me narrate the full context, sequence, and philosophical essence for you.

1. Context of the Debate
The event takes place in the court of King Janaka of Videha, who was a patron of knowledge and philosophy.

King Janaka organized a brahmayajña (philosophical assembly) to discuss the nature of Brahman — the ultimate reality.

The hall was filled with the era’s greatest scholars, sages, and philosophers, including Yajnavalkya, a celebrated Vedic rishi, and Gargi Vachaknavi, one of the most learned women of her time.

The winner of the debate would receive a thousand cows, each with gold ornaments on their horns.
'''
````

````python
cts = CharacterTextSplitter(
    chunk_size = 200,    # 200 characters
    chunk_overlap = 0,
    separator = ''
)

result = cts.split_text(text)

print(result)
````

````python
print(len(result))
print()
print()

print((result[0]))
print()
print()

print((result[1]))
print()
print()

# Note: Length of each 5 chunks will be 200 as we splitted based on 200 characters.
````

# Parameter: `chunk_overlap`

Supose the text is `NikunjYogeshkumarBosamiya`, <br>
chunk_size = 10 <br>
chunk_overlap = 4 <br>

`Chunk 1`: NikunjYoge <br>
`Chunk 2`: Yogeshkuma <br>
`Chunk 3`: kumarBosam <br>
`Chunk 4`: osamiya <br>

`Note`: 4 Characters will be retained from the previous chunk.

`Generally, it is proved by experiments that **chunk_overlap** should be set as 10%-20% of the chunk_size.`

# 3. Document Structure Based Splitter:

When we wanna split some codes of python, java, x=c or any other programing language or any Markdown texts, we use `from_language()` method of `RecursiveCharacterTextSpliter`

````python
from langchain.text_splitter import RecursiveCharacterTextSplitter, Language
````

````python
text = """
class Student:
    def __init__(self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade  # Grade is a float (like 8.5 or 9.2)

    def get_details(self):
        return self.name"

    def is_passing(self):
        return self.grade >= 6.0


# Example usage
student1 = Student("Aarav", 20, 8.2)
print(student1.get_details())

if student1.is_passing():
    print("The student is passing.")
else:
    print("The student is not passing.")

"""

# Initializing the splitter

rcts = RecursiveCharacterTextSplitter.from_language(
    language = Language.PYTHON,
    chunk_size = 150,
    chunk_overlap = 0
)

chunks = rcts.split_text(text)
````

````python
print(len(chunks))
print()
print()

print((chunks[0]))
print()
print()

print(chunks[1])
print()
print()
````

# 4) Sementic Meaning Based Text Splitter.

It splits the text on sementic meaning.

For example, see the below paragraphs:

***Farmers were working hard in the fields, preparing the soil and planting seed for the next season. The sun was bright, smell of earth and fresh grass. The Indian Premier League (IPL) is the biggest cricket league in the world. People from all over the world come to watch the matchs and cheer for their favourite teams.***

***Terrorism spreads fear and violence, threatening peace in society. It destroys innocent lives and disrupts nations’ stability. Born from hatred and extremism, it ignores humanity’s values. Unity, awareness, and justice are the strongest weapons against it.***

We can clearly see that in first paragraph, initially, it is talking about farmers but suddenly it moved to IPL. The second paragraph is totally about terrorism. <br>
If we split this texts based on Recursive Character, it doesn't have the ability to recognize semantic meanings. It will simply split the text into 2 or 3 paragraphs if we give it proper chunk_size. <br>

Here comes the `Sementic Meaning Based Text Splitter` into play. It splits the data based on sementic. It will detect the sudden sementic changes even in same paragraph and split from there. Under the hood, it used embeddings to perform the splits.

# `Note: This is still under experiment`

````python
from langchain_experimental.text_splitter import SemanticChunker
from langchain_openai.embeddings import OpenAIEmbeddings
from dotenv import load_dotenv
````

````python
load_dotenv()
````

````python
text = """
Farmers were working hard in the fields, preparing the soil and planting seed for the next season.
The sun was bright, smell of earth and fresh grass. The Indian Premier League (IPL) is the biggest 
cricket league in the world. People from all over the world come to watch the matchs and cheer for their favourite teams.

Terrorism spreads fear and violence, threatening peace in society. It destroys innocent lives and 
disrupts nations’ stability. Born from hatred and extremism, it ignores humanity’s values. 
Unity, awareness, and justice are the strongest weapons against it.
"""

splitter = SemanticChunker(
    embeddings = OpenAIEmbeddings(),
    breakpoint_threshold_type = 'standard_deviation',
    breakpoint_threshold_amount = 1    # Means if similarity is > 1 std, split from there
)

chunks = splitter.split_text(text)

print(len(chunks))
print()
print()

print((chunks[0]))
print()
print()

print(chunks[1])
print()
print()
````
