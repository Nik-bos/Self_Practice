```python
from dotenv import load_dotenv
from langchain.schema import Document
```

```python
load_dotenv()
```

```python
# Creating document objects

doc1 = Document(page_content = """Avenged Lala Lajpat Rai’s death by killing Saunders and
shook the British by bombing the Central Assembly.""",
               metadata = {'freedom_fighter': 'Bhagat Singh'})


doc2 = Document(page_content = '''Organized youth through the Hindustan Socialist Republican 
Association (HSRA) and played a key role in revolutionary activities against the British.''',
               metadata = {'freedom_fighter': 'Sukhdev'})


doc3 = Document(page_content = '''Actively participated in the plan to kill Saunders and fought
as a fearless revolutionary alongside Bhagat Singh and Sukhdev.''',
               metadata = {'freedom_fighter': 'Rajguru'})


doc4 = Document(page_content = '''Led HSRA after the death of Ram Prasad Bismil, carried out armed 
resistance, and fought till his last breath to avoid British capture.''',
               metadata = {'freedom_fighter': 'Chandra Shekhar Azad'})


doc5 = Document(page_content = '''Fired the first shot of revolt in 1857 against British officers,
sparking India’s first large-scale armed uprising for independence.''',
               metadata = {'freedom_fighter': 'Mangal Pandey'})


doc6 = Document(page_content = '''Avenged the Jallianwala Bagh massacre by assassinating Michael O’Dwyer in London.''',
               metadata = {'freedom_fighter': 'Udham Singh'})


# We can also pass 'id' parameter
```

```python
docs = [doc1, doc2, doc3, doc4, doc5, doc6]
```

# 1) Using Chroma

```python
from langchain_openai import OpenAIEmbeddings
from langchain_chroma import Chroma
```

```python
# Creating an object of Chroma vecor store
vector_store = Chroma(
    embedding_function = OpenAIEmbeddings(),
    persist_directory = 'Freedom Fighters DB',    # File where our data will be stored
    collection_name = 'India')
```

```python
print(vector_store)
```

```python
# Adding documents to vector_store

vector_store.add_documents(docs, ids = ['1', '2', '3', '4', '5', '6'])

# This will add documents to the vector store and also a unique id will be assigned to each document
# We can also give custom ids by using 'ids' parameter

# If 'id' is present in Document and we pass 'ids' here also, these new ids will get the preference
```

```python
# View documents

print(vector_store.get(include = ['embeddings', 'metadatas']))
```

```python
vector_store.get(ids = '1',
                include = ['embeddings'])
```

```python
# View documents

vector_store.get(include = ['embeddings', 'metadatas'],
                      limit = 2)
```

```python
# Searching document

vector_store.similarity_search(
    query = 'Who took revenge of Jaliyanwala baugh mass killing by a british officer?',
    k = 1    # No_of results to return
)
```

```python
# Searching document with similarity score as well.
# The lesser the score the more better result because this score represents distance

result = vector_store.similarity_search_with_score(
    query = 'Which freedom fighter wa snever caught alive by britishers',
    k = 1    # No_of results to return
)

print(result)
```

```python
# Filtering

vector_store.similarity_search_with_score(
    query = '',
    filter = {'freedom_fighter': 'Bhagat Singh'}
)
```

```python
# Update a document

new_data = Document(
    page_content = '''In 1929, while imprisoned, Bhagat Singh and his fellow revolutionaries went on a 
    116-day hunger strike demanding equal rights and humane treatment for Indian political prisoners
    ''',
    metadata = {'freedom_fighter': 'Bhagat Singh'}
)

vector_store.update_document(document_id = '1', document = new_data)
```

# 2) FAISS

```python
from langchain_community.vectorstores import FAISS
from langchain_openai import OpenAIEmbeddings

from dotenv import load_dotenv
from langchain.schema import Document
```

```python
load_dotenv()
```

```python
# Creating document objects

doc1 = Document(page_content = """Refused to surrender Jhansi to the British in 1857 and 
fought with unmatched courage; escaped on horseback with her infant son and continued leading 
battles at Kalpi and Gwalior.""",
               metadata = {'freedom_fighter': 'Rani Laxmibai'})


doc2 = Document(page_content = '''Disguised as Bhagat Singh’s wife to help him escape British police; 
personally led an armed attack on a British officer’s car in 1929.''',
               metadata = {'freedom_fighter': 'Durga bhabhi'})


doc3 = Document(page_content = '''Took charge of the Malwa kingdom after her husband’s death and protected
it from invasions; ensured law, justice, and built forts/temples that symbolized cultural resistance.''',
               metadata = {'freedom_fighter': 'Ahilyabai Holkar'})


doc4 = Document(page_content = '''Unfurled the first Indian flag in Stuttgart, Germany in 1907 as a call for
independence; raised international awareness of British atrocities through fiery speeches and writings.''',
               metadata = {'freedom_fighter': 'Bhikaiji Cama'})


doc5 = Document(page_content = '''Led an armed rebellion in 1824 against British annexation of Kittur; defeated
the British in her first battle and became an icon of women’s resistance.''',
               metadata = {'freedom_fighter': 'Kittur Rani Chennamma'})


# We can also pass 'id' parameter
```

```python
docs = [doc1, doc2, doc3, doc4, doc5]
```

```python
# Like chroma, we cannot create an empty vector store in FAISS, 
# so while creating an object itself, we pass our documents

faiss = FAISS.from_documents(documents = docs, embedding = OpenAIEmbeddings(), ids = ['1', '2', '3', '4', '5'])
```

```python
# Adding a new document

doc6 = Document(page_content = '''She led her army personally against the Mughal general Asaf Khan
in 1564, fighting valiantly with bow and arrows. When wounded in battle and facing defeat, she chose 
to stab herself rather than surrender, becoming a symbol of honor and courage.''',
               metadata = {'freedom_fighter': 'Rani Durgavati'})

doc7 = Document(page_content = '''She inspired Rajput warriors by refusing Alauddin Khilji’s advances,
defending Chittorgarh’s dignity with unshakable resolve. During the siege of Chittor in 1303, she led 
thousands of women into jauhar (mass self-immolation), embracing fire over dishonor.''',
               metadata = {'freedom_fighter': 'Rani Padmini'}) 

faiss.add_documents(documents = [doc6, doc7], ids = ['6', '7'])
```

```python
faiss.similarity_search_with_score(
    query = 'Who fought tying her baby on her back?',
    k = 1)
```

```python
faiss.delete(ids = ['7'])
```

Other methods:
1. delete(id = []): To delete documents
2. add_documents(documents = []): To add documents
3. similarity_search_with_score(query, k): To search a document based on a query.

Note: There is no method to upate a document in FAISS. So first delete the old doc and enter the new one
