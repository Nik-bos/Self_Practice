# Not, we will create a small document similarity app.

**Scenerio**: We will have a document and a question. We will know to which document, the question is realted. <br>

`How will this work?`:<br>
We will have embeddings of both, documents' and our question's. Then we will find our question's simillarity with all the documents, whose simillarity is highest, that's the document.

```python
from langchain_openai import OpenAIEmbeddings
from dotenv import load_dotenv

# We will need cosine_similarity() to generate cosine_similarity
from sklearn.metrics.pairwise import cosine_similarity

# may need numpy
import numpy as np
```

```python
load_dotenv()
```

```python
emb = OpenAIEmbeddings(
    model = 'text-embedding-3-large',
    dimensions = 300
)
```

```python
document = ['Chandrashekhar Azad embraced martyrdom with honor at Alfred Park in 1931, choosing death over surrender to the British.',
'Bhagat Singh ignited the spirit of revolution by fearlessly throwing bombs in the Central Assembly, shouting Inquilab Zindabad!',
'Sukhdev Thapar displayed unmatched bravery by planning the assassination of Saunders to avenge Lala Lajpat Rai’s death.',
'Durga Bhabhi showed extraordinary courage by disguising Bhagat Singh as her husband and aiding his heroic escape.',
'Rajguru proved his patriotism by firing the revolutionary shot that avenged the injustice done to Lala Lajpat Rai.']
```

```python
document[0]
```

```python
query = 'Tell me about Azad.'
```

```python
# Creating document embeddings
doc_emb = emb.embed_documents(document)
query_emb = emb.embed_query(query)
```

```python
print(len(doc_emb[0]))
print(len(query_emb))
```

```python
# finding query's simillarity with all 5 documents.
# Remember, values passed to cosine_simillarity should be 2-D list.
# our query_emb is 1d.

sim_scores = cosine_similarity([query_emb], doc_emb) [0]
# See, our first sentence is of Chandra shekhar azad and our query as related with him.
# So we are getting highest similarity with the 1st document i.e 0.41x

print(sim_scores)
```

```python
# Getting the index of the highest simillarity score.

high_score_index = np.argmax(sim_scores)
```

```python
document[high_score_index]
```

```python
# Let; try another query

def get_document(query, documents):
    doc_emb = emb.embed_documents(documents)
    query_emb = emb.embed_query(query)

    sim_scores = cosine_similarity([query_emb], doc_emb)

    high_sim_index = np.argmax(sim_scores)

    return documents[high_sim_index]
    
```

```python
get_document(
    query = 'Which woman helped BhagatSingh?',
    documents = document)
```
