```python
from langchain_openai import OpenAI    # To work with OpenAI models
from dotenv import load_dotenv    # To load .env file where we've stored our OpenAI's API key 
```

```python
load_dotenv()
```

```python
llm = OpenAI(model = 'gpt-3.5-turbo-instruct',  # Name of the model we wanna use
            temperature = 0.2,    # Effects the creativity of the response. Maximum 2
            max_tokens = 15)    # Maximum tokens in output

# model: Name of the model we wanna use. There are many other models available n OpenAi. We will see then later

```

# For creativity (temperature):
- Factual answers (Maths, facts, code) --> 0- 0.3 (No much creativity, want accurate answers)
- Balanced answers (General quest & answer) --> 0.5- 0.7
- Creative writting, story telling, jokes --> 0.9- 1.2
- Max randomness (Great ideas) --> 1.5+


```python
result = llm.invoke('Who is the known as captain cool? Just give me name')

# invoke() is a very importand method in lagchain.
# It is available in each component of the langchain.

# What will happen here is, invoke() will go and hit the model and give the entered prompt to the model.
# Model will process the prompt and generate a reply, which we've stored in variable 'result' 
```

```python
print(result.strip())
```

# Now, we know the above models were paid, but what if we don't wanna pay? In this case, we also have open source models which we can download in our system, can use as we want, can fine tune and can also deploy anywhere

`Difference between Paid and Open source models:`

`Cost`: <br> 
--> Open source models are free to use as there's no api usage. <br>
--> We need to pay for API usage. (OpenAI charges per token) <br>

`Control`: <br>
--> Can modify, fine-tune, deploy anywhere. <br>
--> Locked to provider's infrastructure, we cannot modify. <br>

`Data privacy`: <br>
--> We have high data privacy as we are downloading model into our system and then using it.<br>
--> No data privacy. We are sending data to the model through API for processing. 

**Some popular open source models**

- LLaMA by Meta AI (General purpose text generation)
- Mixtral by Mistral AI (Effecient and fast response)
- Mistral by Mistral AI (Best small scale model)
- Falcon by Tll UAE (High speed inference)
- Bloom by BigScience (Multilingual text generation)
- GPT-J-6B by EleutherAI (Light weight and effecient)
- GPT-NeoX-20B by EleutherAI (LAege scale application)
- StableLM by StabilityAI (Compact model for chatbot)

**Where can we find these models?** <br>
We will find these and many more open sourse models in **Huggingface** <br>
Huggingface is the largest repository for opensource LLMs. <br>
We can visit its website and find thousands of open source models for various tasks.

`We have two ways to use these models. They're` <br>
1. We can download the model in our system and then use it.
2. We can generate Hugging face API and use the models using API. We get limited trials in API and after exhausting free trials, we need to pay. But it's enough if we are working for small project or learning as a student

**Disadvantages**
1. Require expensive GPUs: Running large models in our own system requires high GPUs.
2. Setup complexities: Requires to install dependencies like Pytorch, CUDA, Transformers etc, which we're not needed if we use API.
3. Lack of RLHF: Resonses by open source model is not up to the mark like closed source models. But we can handle this as we can fine tune them. (We will learn this later)
4. Limited multimodel ability: Open models don't support images, audios, videos like GPT-4 and above.
