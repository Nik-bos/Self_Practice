# What is abstraction:

Abstraction means hiding the implementation from the user. 
For e.g:

1) When we search something on google, we get results, but we don't know what is going on behind the scene.

2) When we withdraw or deposit money on ATM or CDM, we get our money withdrawn or deposited, but we don't know how is this happening.

All such cases are of 'Abstraction'

In actual, Abstraction is attained through 'Encapsulation', but here we will just understand what Abstraction is.

# Create Abstract Class

```python
from abc import ABC, abstractmethod
```

```python
class Vehical(ABC):         # This is still not an abstract class. To make it an abs class, it should have atleast one abstract method
                            # Morewver we cannot create an object of abstract class
    
    def __init__(self, n):
        self.no_of_tyres = n

    @ abstractmethod
    def start(self):
        pass
    
```

```python
v = Vehical()    # We cannot create an object of abstract class
```
