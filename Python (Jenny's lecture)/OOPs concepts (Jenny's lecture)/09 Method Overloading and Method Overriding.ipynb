# Method Overloading.

# `Note:` Python doesn't support `Method Overloading`, but can achive it with some technique 

* `Method Overloading` means in same class, we are defining more than one methods with same name but with different no.of parameters. 

```python
class Demo:
    def add(self, a, b):
        return a + b
```

```python
d = Demo()
d.add(5, 6)
```

```python
class Demo:
    def add(self, a, b):
        return a + b

    def add(self, a,b,c):
        return a + b + c
```

```python
d = Demo()
print(d.add(5, 6, 7))
print(d.add(5, 6))

'''
Here, in 'print(d.add(5, 6))', we will receive error bcz if it does'nt give error, it is Method Overloading. But as Method Overloading
is not supported in Python, we are encountering an error that 'missing one argument c'. This is bcz Python considers the latest method
everytime and in this case, the latest add method has 3 parameters.
'''
```

# Let's achieve Method Overloading Through some other manner.

1) By giving a default argument.
2) By using *args as 2nd parameter, after self

```python
class Demo:

    def add(self, a,b,c = 0):
        return a + b + c
```

```python
d = Demo()
print(d.add(5, 6))
print(d.add(5, 6, 7))

# See, both are working, unlike in above code where add() method with 3 arguments was only working.
```

```python
# Using *args as 2nd parameter

class Demo:
    def add(self, *args):
        
        total = 0

        for i in args:
            total = total + i
        return total
```

```python
d = Demo()
print(d.add(5, 6))
print()

print(d.add(5, 6, 7))
print()

print(d.add(5, 6, 7, 8))
print()
```

# `Method Overriding:`

Method Overriding means two different classes has a method with same name and same no.of parameters but definition is different. Moreover classes must be inherited

```python
class Father:
    def sleep(self):
        print("Father sleeps from 10 p.m to 5 a.m")

    def eat(self):
        print("Father eats only healthy food.")



class Son(Father):
    def sleep(self):    # Method Overriding
        print("Son sleeps from 12 a.m to 8 a.m")

    def eat(self):    # Method Overriding
        print('Son eats more junk food.')
        # super().eat()
```

```python
s = Son()

s.sleep()
s.eat()
```

```python
f = Father()

f.sleep()
f.eat()
```
