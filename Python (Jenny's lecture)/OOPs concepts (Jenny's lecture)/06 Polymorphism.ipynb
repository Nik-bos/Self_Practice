Poly means `many` and `morph` means forms. So `Polymorphism` means many forms.

Imagine a person, he has many forms. A person is a  father, son, friend, teacher, gurdian, husband etc.
His roles changes according to the other person he is dealing with. But ultimately he is a Person. This is called `Polymorphism`.



`Now let's understand this with the example related to programing`.

Consider a shape superclass with subclasses circle, square, and triangle. Each subclass has its own implementation of the area() method. When you call the area() method on a shape object, polymorphism allows the appropriate area() method of the specific subclass to be invoked based on the type of the object.

# There're four ways to achieve Polymorphism. They're,
1) Duck typing
2) Method overloading
3) Method overriding and
4) Operator overloading
