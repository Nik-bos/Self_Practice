```python
class Student_info:
    def __init__(self, name, div, subject):
        self.name = name    # self.name is a variable and name is a parameter
        self.div = div
        self.sub = subject
    
    def info(self):
        print(f"""Name: {self.name}
Division: {self.div}
Subject: {self.sub}""")
```

```python
nikunj = Student_info(name = 'Nikunj',
                      div = 'A',
                      subject = "Python")
```

```python
nikunj.info()
```

```python
jugal = Student_info('Jugal', 'A', 'English')
```

```python
jugal.info()
```

```python
harshit = Student_info("Harshit", 'B', 'Anatomy')
```

```python
harshit.info()
```

```python
class ATM:
    def __init__(self, balance):
        self.pin = 0000
        self.balance = balance
        self.menu()

    def menu(self):
        user_input = int(input("""
Hiii! How can I help you?

a. Press 1 to create_pin
b. Press 2 to change_pin
d. Press 3 to check balance
e. Press 4 to withdraw
f. Press 5 to exit"""))

        if user_input == 1:
            self.create_pin()
        if user_input == 2:
            pass    # change_pin
        if user_input == 3:
            pass    # check_balance
        if user_input == 4:
            pass    # withdraw
        if user_input == 4:
            pass    # exit

#========== Create Pin ==========
    def create_pin(self):
        new_pin = int(input("Enter new pin: "))
        self.pin = new_pin

        print("Pin created successsfully.")
        self.menu()
```

```python
nikunj = ATM(10000)
```

```python
print(nikunj.pin)
print(nikunj.balance)
```
