# Types of Inheritance.
1) **Single inheritance**: One child class inheriting one parent class `(Father/mother to son)`
2) **Multiple inheritance**: One child class inherits the properties from more than one parent class`(Father's and mother's both's property to son)`.
3) **Multilevel inheritance**: A child class inherits from parent class and forms a chain `(Great grandfather > Grandfather > father > son)`.
4) **Hierarchical inheritance**: Multiple child class inherits from one parent class `(Father's property to his all 3 sons)`.
5) **Hybrid inheritance**: Hybrid inheritance is a combination of two or more types of inheritance. It can involvemultiple, multilevel, or hierarchical inheritance

# Multiple inheritance.
2) Multiple inheritance: One child class inherits the properties from more than one parent class.(Father's and mother's both's property to son)

Multiple inheritance is an example of `Diamond problem`.

```python
class Yogesh:
    def funny(self):
        print("Father is funny in nature")

class Bhavna:
    def calm(self):
        print("Mother is calm in nature")

class Nikunj(Yogesh, Bhavna):
    def angry(self):
        print('Nikunj is angry in nature')
```

```python
n = Nikunj()

n.calm()
n.funny()
n.angry()
```

# Multi level inheritance:
Multilevel inheritance: A child class inherits from parent class and forms a chain.(Great grandfather -> Grandfather -> father -> son)


```python
class Dharsibhai:    # Grandfather
    def account(self):
        print("Grandfather knows account")

class Yogesh(Dharsibhai):    # Father
    def history(self):
        print('Father knows history')

class Nikunj(Yogesh):    # Son
    def maths(self):
        print("Son knows maths.")


# We can notice that this srtucture is creating a level. Hence it is called multi-level inheritance
```

```python
n = Nikunj()
n.history()
n.account()
n.maths()
```

# 4) Hierarchical inheritance.

Multiple child class inherits from one parent class. (Father's property to his all 3 sons)


```python
class Yogesh:    # Father
    def maths(self):
        print("Yogesh knows maths")

class Nikunj(Yogesh):    # Son 1
    pass

class Jugal(Yogesh):    # Son 2
    pass
        
        
```

```python
n = Nikunj()
n.maths()
```

# Method Resolution Order(MRO)

This will tell us the order of searching the methods or attributes in different class in case of inheritance

```python
# This code is an example of a Diamond Problem

class A:
    def display(self):
        print("Display from class A")

class B(A):
    def display(self):
        print("Display from class B")

class C (A):
    def display(self):
        print("Display from class C")

class D(B, C):
    def display(self):
        print("Display from class D")
```

```python
d = D()
```

```python
(super(D, d).display())    
# Here, super(D, d) returns a proxy object that represents the superclass of class D (B in this case) 
# for the instance d.
```

```python
D.mro()
```

```python
d = D()
d.mro()
```

# super() keyword.

`super()` keyword is used to access the methods of parent class even though they are present in child class.
For eg. Supose that display() method is present in parent and child class both. When we call display method through the object of child class,
it will execute display() of child class. Right?? But we want such that upon calling the display method of child class, 
it should call the display() method of parent class as well. 

Here comes the super() keyword. 
See the below code.

# Note: Remember, we cannot use super() keyword outside the class. We can only use it inside the class. Moreover, super() cannot acces attributes of the parent class, it can only access methods of parent class.

# Summary:

1) super() can only be used inside the class and not outside
2) It can only access methods of parent class and not the attributes.

```python
class Phone:
    def __init__(self, price, brand):
        self.price = price
        self.brand = brand

    def display(self):
        return 'From Phone(Parent) class.'

class Smartphone(Phone):
    
    def display(self):
        print('From Smartphone(Child) classs.')
        print(super().display())
```

```python
s = Smartphone(20000, 'Samsung')
```

```python
s.display()
```

```python
class Phone:
    def __init__(self, price, brand, camera):
        print("Inside Phone constructor.")
        self.price = price
        self.brand = brand
        self.camera = camera

    def display(self):
        return 'From Phone(Parent) class.'

class Smartphone(Phone):
    def __init__(self, price, brand, camera, os, ram):
        print('Inside Smartphone constructor.')
        super().__init__(price, brand, camera)
        self.os = os
        self.ram = ram
        
    
    def display(self):
        print('From Smartphone(Child) classs.')
        print(super().display())
```

```python
s = Smartphone(20000, 'Samsung', 108, 'Android', 8)
print(s.ram)
print(s.camera)
print(s.brand)
```

# Understand the working of above code.

When we created the object of Smartphone class, it immediately triggered its constructor and printed 'Inside Smartphone constructor'. <br>
Then it called the constructor of Phone(Parent) class and printed "Inside Phone constructor" and initialised attributes price and brand and camera.<br>
Then it again came to Smartphone's constructor and initialised the attributes os and ram
