# 1) Arithmetic Operators:

 # Rules of Precendancy:
 `PEMDAS` <br>

1) `Parenthesis`: 
2) `Exponent`:
3) `Multiplication->Division-> Floor Division-> Modulus`
4) `Addition-> Substraction`

**Working:**
- First, Parenthesis will get solved
- Then Exponent
- Now understand carefully. MDFM has same preference, so which will get solved first? The one which appears first, gets solved first.
- Same for Addition and Substraction: Which appears first, gets solved first

```python
print(5+2)    # 7
print(5-2)    # 3
print(5*2)    # 10
print(4/2)    # 2.0
print(4//2)    # 2
print(5%2)    # 1
print(2**5)    # 32
```

```python
# Rule of operation
# PEMDAS

print(
    (5+2)+2-3*4/2**2//3
)

# 7+2-3*4/4//3
# 7+2-12/4//3
# 7+2-3//3
# 7+2-1
# 9-1
# 8

# Working

# Parenthesis-> Exponent-> Multiplication->Division->FloorDivision->Additiion->Substraction
```

2) **Assignment** and **Comparison** operator

- `Assignment Operators`
    - Used to assign a value to the variable
    - Types are:
        - =, +=, -=, *=, /=, //=, %= and **=

- `Comparison Operators`
    - Used to compare values or variables (LHS = RHS)
    - Types are:
        -  <, >,  >=, <=, !=, ==
    - Result of these operators will be either True or False


```python
# Assignment Operators

a = 5
print(a)    # 5

a+=2    #5+2 = 7
print(a)

a-=2    # 7-2 = 5
print(a)

a*=2    # 5*2 = 10
print(a)

a/=2    # 10/2 = 5.0
print(a)

a//=2   # 5.0//2 = 2.0 
print(a)

a%=2    # 2.0 % 2 = 0
print(a)

a**=2    # 0 Raised to 0 equals to 0
print(a)

a, b = 5, 6
c = a+b

a+=5
c+=a

print(c)    # 21

# Exercise:
"""
a, b = 1, 2
c = a+b
print(c)

c+=a
print(c)

c%=b
print(c)

c//=a
print(c)
"""
```

```python
# Comparison Operators

a = 5

print(a>10)    # False
print(a<10)    # True
print(a>=10)    # False
print(a<=10)    # True
print(a==10)    # False
print(a!=10)    # True
```

# 3) **Logical Operators**:

* Logical Operators are used to combine two or more conditions or say, conditional statements.
* Types are:
    * `and`: Returns True only if both the conditions are True.
    * `or`: Returns True if any one condition is True.
    * `not`: Negates the conditions


```python
a, b, c = 7, 8, 0

print((a>5) and (b<10))    # True
print((a>5) and (b<5))    # Flase bcz b<5 is False
print((a>5) or (b<5))   # True bcz a>5 is True

print(not a)    # Flase: Other than 0 everything is True
print(not c)    # True
```

# 4) **Bitwise Operators**

* This works with bits
* Types are:
    * `&` : If both bits are 1, then only returns 1, else 0.
    * `|` : If any one bit is 1, returns 1.
    * `^` : If both the bits are same, returns 0, else 1.
    * `<<(Left Shift)`:
        - Shifts the bits to left and inputs 0 on the left.
        - This does not delete any bits
    * `>>(Right shift)`:
        - Shifts the bits to right.
        - This deletes the bits.

```python
a = 5
b = 4

print(a & b)    # 4
print(a | b)    # 5
print(a ^ b)    # 1
print(a << 2)   # 20
print(a >> 2)   # 1
```

# 5) **Identity Operators**

* This is used to compare the memory location of the objects.
* `For example`: 5 is 5, check if location of both 5's are same.
* There're two types:
    - is
        - Return True if location of both the values are same, else False.
    - is  not
        - Returns True if location of both the values are different, else True if same.

# Difference between 
* `is` and `==`: 
    * `is` compares memory location.
    * `==` compares values
* `is not` and `!=`:
    - `is not` compares memory location.
    - `!=` compares values.

`We use identity operators mostly with conditional statements.`

```python
a = b = 5
print(a is b)    # True. bcz both a and b are at same memory location
print(a is not b)    # False bcz a and b both are at same memory location
print(id(a))
print(id(b))
```

# 6) Membership Operator
* Checks if a value present in the sequence.
* Sequence means string, list, tuple, dictionary, set.

```python
a = 'Nikunj'
print("k" in a)    # True
print("K" in a)    # False

b = ("Nikunj", 'Harshit', 'Jenish', "Parth")
print("Harshit" in b)    # True

print('Nikunj' not in b)    # False, bcz Nikunj is in b
print("y" not in a)    # True, bcz 'y' not in b
```

Question:
How much days, weeks, months and years are left in your life, if u are to live 90 years.

`Input`: Your age <br>
`Output`: You are left with 365 days, 52 weeks, 12 months and 1 year.

```python
age = int(input("Enter your age"))
total_age = 90

days_remaining = (total_age*365) - (age*365)
weeks_remaining = (total_age*52) - (age*52)
months_remaining = (total_age*12) - (age*12)
years_remaining = (total_age - age)

print(f""" You are left with
{days_remaining} days,
{weeks_remaining} weeks,
{months_remaining} months or
{years_remaining} years

""")
```
